Ejemplo sin usar I — Segregación de Interfaces (ISP)

```python
class ImpresoraDeOficina:
    def imprimir(self):
        pass

    def escanear(self):
        pass

    def enviar_fax(self):
        pass

class ImpresoraBasica(ImpresoraDeOficina):
    def imprimir(self):
        print("Imprimiendo documento")

    def escanear(self):
        print("Escaneando documento")

    def enviar_fax(self):
        pass

impresora = ImpresoraBasica()
impresora.imprimir()
impresora.escanear()
impresora.enviar_fax()
```

Ejemplo usando I — Segregación de Interfaces (ISP)

```python
from abc import ABC, abstractmethod


class Imprimir(ABC):
    @abstractmethod
    def imprimir(self):
        pass


class Escanear(ABC):
    @abstractmethod
    def escanear(self):
        pass

class EnviarFax(ABC):
    @abstractmethod
    def enviar_fax(self):
        pass

class ImpresoraBasica(Imprimir, Escanear):
    def imprimir(self):
        print("Imprimiendo documento")

    def escanear(self):
        print("Escaneando documento")

class ImpresoraAvanzada(Imprimir, Escanear, EnviarFax):
    def imprimir(self):
        print("Imprimiendo documento")

    def escanear(self):
        print("Escaneando documento")

    def enviar_fax(self):
        print("Enviando fax")

print("Usando Impresora Básica:\n")
impresora_basica = ImpresoraBasica()
impresora_basica.imprimir()
impresora_basica.escanear()

print("\nUsando Impresora Avanzada:\n")
impresora_avanzada = ImpresoraAvanzada()
impresora_avanzada.imprimir()
impresora_avanzada.escanear()
impresora_avanzada.enviar_fax()

```
