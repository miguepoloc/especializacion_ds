Ejemplo sin usar O — Abierto/Cerrado (OCP)

```python
class CalculadoraDescuento:
    def calcular(self, monto: float, tipo_cliente: str) -> float:
        if tipo_cliente == "normal":
            return monto
        elif tipo_cliente == "vip":
            return monto * 0.80



calculadora = CalculadoraDescuento()
print(f"Total Normal: ${calculadora.calcular(100.0, 'normal')}")
print(f"Total VIP: ${calculadora.calcular(100.0, 'vip')}")
```

Ejemplo usando O — Abierto/Cerrado (OCP) 

```python
from abc import ABC, abstractmethod

class Descuento(ABC):
    @abstractmethod
    def calcular(self, monto: float) -> float:
        pass


class DescuentoNormal(Descuento):
    def calcular(self, monto: float) -> float:
        return monto


class DescuentoVIP(Descuento):
    def calcular(self, monto: float) -> float:
        return monto * 0.80


class DescuentoMayorista(Descuento):
    def calcular(self, monto: float) -> float:
        return monto * 0.70


class CalculadoraDescuento:
    def __init__(self, estrategia_descuento: Descuento):
        self.estrategia = estrategia_descuento

    def aplicar_descuento(self, monto: float) -> float:
        return self.estrategia.calcular(monto)


monto_compra = 3000.0

calc_normal = CalculadoraDescuento(DescuentoNormal())
print(f"Total Normal: ${calc_normal.aplicar_descuento(monto_compra)}")


calc_vip = CalculadoraDescuento(DescuentoVIP())
print(f"Total VIP: ${calc_vip.aplicar_descuento(monto_compra)}")


calc_mayorista = CalculadoraDescuento(DescuentoMayorista())
print(f"Total Mayorista: ${calc_mayorista.aplicar_descuento(monto_compra)}")
```
