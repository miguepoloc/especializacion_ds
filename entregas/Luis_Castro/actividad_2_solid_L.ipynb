EJemplo sin usar L — Sustitución de Liskov (LSP)

```python
class Trabajador:
    def __init__(self):
        self.salario = 1000.0

class Voluntario(Trabajador):
    def __init__(self):
        self.salario = "Sin sueldo"

def calcular_bono_navideño(trabajador: Trabajador):
    bono = trabajador.salario * 0.10
    print(f"El bono es de: ${bono}")

empleado_normal = Trabajador()
calcular_bono_navideño(empleado_normal)

voluntario = Voluntario()
calcular_bono_navideño(voluntario)
```

Ejemplo usando L — Sustitución de Liskov (LSP)

```python
from abc import ABC, abstractmethod

class Trabajador(ABC):
    @property
    @abstractmethod
    def salario(self) -> float:
        pass


class EmpleadoNormal(Trabajador):
    def __init__(self):
        self._salario = 1000.0

    @property
    def salario(self) -> float:
        return self._salario


class Voluntario(Trabajador):
    def __init__(self):
        self._salario = 0.0

    @property
    def salario(self) -> float:
        return self._salario


def calcular_bono_navideño(trabajador: Trabajador):
    bono = trabajador.salario * 0.10
    print(f"El bono asignado es de: ${bono:.2f}")


empleado = EmpleadoNormal()
voluntario = Voluntario()


calcular_bono_navideño(empleado)
calcular_bono_navideño(voluntario)
```
