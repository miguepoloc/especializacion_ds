Ejemplo sin usar S — Responsabilidad Única (SRP)


```python
class CarritoDeCompras:
    def __init__(self):
        self.items = []

    def agregar_producto(self, producto: str, precio: float):
        self.items.append({"producto": producto, "precio": precio})

    def quitar_producto(self, producto: str):
        self.items = [item for item in self.items if item["producto"] != producto]

    def calcular_total(self) -> float:
        return sum(item["precio"] for item in self.items)

    def procesar_pago(self, tarjeta: str):
        print(f"Procesando pago con tarjeta {tarjeta} por un total de {self.calcular_total()}...")

    def enviar_confirmacion_email(self, email: str):
        print(f"Enviando correo de confirmación de la compra a {email}...")

Carrito = CarritoDeCompras()
Carrito.agregar_producto("Laptop", 1200.00)
Carrito.agregar_producto("Mouse", 25.00)
Carrito.quitar_producto("Mouse")
print(f"Total a pagar: ${Carrito.calcular_total()}")
Carrito.procesar_pago("1234-5678-9012-3456")
Carrito.enviar_confirmacion_email("cliente@gmail.com")

```

Ejemplo aplicando S — Responsabilidad Única (SRP)

```python
from abc import ABC, abstractmethod

class CarritoDeCompras:
    def __init__(self):
        self.items = []

    def agregar_producto(self, producto: str, precio: float):
        self.items.append({"producto": producto, "precio": precio})

    def quitar_producto(self, producto: str):
        self.items = [item for item in self.items if item["producto"] != producto]

    def obtener_items(self):
        return self.items

class CalculadoraCarrito:
    @staticmethod
    def calcular_total(carrito: CarritoDeCompras) -> float:
        return sum(item["precio"] for item in carrito.obtener_items())

class ProcesadorPago(ABC):
    @abstractmethod
    def procesar_pago(self, carrito: CarritoDeCompras):
        pass

class ProcesadorPagoTarjeta(ProcesadorPago):
    def __init__(self, tarjeta: str):
        self.tarjeta = tarjeta

    def procesar_pago(self, carrito: CarritoDeCompras):
        total = CalculadoraCarrito.calcular_total(carrito)
        print(f"Procesando pago con tarjeta no: {self.tarjeta} por un total de ${total}...")

class notificador(ABC):
    @abstractmethod
    def enviar_confirmacion_email(self, email: str):
        pass

class NotificadorEmail(notificador):
    def enviar_confirmacion_email(self, email: str):
        print(f"Enviando correo de confirmación de la compra a {email}...")

carrito = CarritoDeCompras()
carrito.agregar_producto("Laptop", 1200.0)
carrito.agregar_producto("Mouse", 25.0)
carrito.quitar_producto("Mouse")

total = CalculadoraCarrito.calcular_total(carrito)
print(f"Total a pagar: ${total}")

pago = ProcesadorPagoTarjeta("1234-5678-9012-3456")
pago.procesar_pago(carrito)

email = NotificadorEmail()
email.enviar_confirmacion_email("cliente@email.com")
```
