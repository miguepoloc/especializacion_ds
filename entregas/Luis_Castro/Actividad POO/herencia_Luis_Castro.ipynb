```python
class VideojuegoPC:
    def __init__(self, titulo, generos, desarrollador):
        self.titulo = titulo
        self.generos = generos
        self.desarrollador = desarrollador

class EarlyAccess(VideojuegoPC):
    def __init__(self, titulo, generos, desarrollador, version_build):
        super().__init__(titulo, generos, desarrollador)
        self.version_build = version_build

class VersionOficial(VideojuegoPC):
    def __init__(self, titulo, generos, desarrollador, version_lanzamiento):
        super().__init__(titulo, generos, desarrollador)
        self.version_lanzamiento = version_lanzamiento

juego1 = EarlyAccess("MENACE", "RPG", "Overhype Studios", "0.3.0")
juego2 = VersionOficial("The Witcher 3", "Aventura", "CD Projekt Red", "1.32")

print(f"{juego1.titulo} Genero: {juego1.generos} - Build: {juego1.version_build}")
print(f"{juego2.titulo} Genero: {juego2.generos} - Version: {juego2.version_lanzamiento}")
```
