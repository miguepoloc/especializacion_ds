Encapsulamiento

```python
class VideojuegoPC:
    def __init__(self, titulo):
        self.titulo = titulo
        self.__calificaciones = []

    def agregar_calificacion(self, nota):
        if 1.0 <= nota <= 5.0:
            self.__calificaciones.append(nota)

    def calcular_promedio(self):
        if len(self.__calificaciones) == 0:
            return 0.0
        suma = sum(self.__calificaciones)
        return round(suma / len(self.__calificaciones), 1)

    @property
    def total_reseñas(self):
        return len(self.__calificaciones)


juego = VideojuegoPC("Cyberpunk 2077")
juego.agregar_calificacion(5.0)
juego.agregar_calificacion(4.5)
juego.agregar_calificacion(4.8)

print(f"Juego: {juego.titulo}")
print(f"Total Reseñas: {juego.total_reseñas}")
print(f"Promedio: {juego.calcular_promedio()} / 5.0")
```
