```python
class VideojuegoTienda:
    def __init__(self, titulo, precio, plataforma):
        self.titulo = titulo
        self.precio = precio
        self.plataforma = plataforma

    def obtener_ficha_venta(self):
        return f"{self.titulo}\n Plataforma= {self.plataforma} \n -Precio= ${self.precio}"


class Videojuego_ficha_tecnica:
    def __init__(self, titulo, fps_objetivo, motor_grafico):
        self.titulo = titulo
        self.fps_objetivo = fps_objetivo
        self.motor_grafico = motor_grafico

    def obtener_configuracion(self):
        return f"Ficha tecnica \n {self.titulo}\n  Motor grafico= {self.motor_grafico}\n  Fps= {self.fps_objetivo} "


juego_tienda = VideojuegoTienda("The Witcher 3", 20, "PC")
juego_ficha = Videojuego_ficha_tecnica("The Witcher 3", 60, "REDengine")

print(juego_tienda.obtener_ficha_venta())
print(juego_ficha.obtener_configuracion())
```
