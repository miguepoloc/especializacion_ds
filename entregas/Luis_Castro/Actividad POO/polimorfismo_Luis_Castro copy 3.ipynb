Polimorfismo

```python
from abc import ABC, abstractmethod

class VideojuegoPC(ABC):
    def __init__(self, titulo, desarrollador):
        self.titulo = titulo
        self.desarrollador = desarrollador
        self.__calificaciones = []

    def agregar_calificacion(self, nota):
        self.__calificaciones.append(nota)

    def calcular_promedio(self):
        if len(self.__calificaciones) == 0:
            return 0.0
        return round(sum(self.__calificaciones) / len(self.__calificaciones), 1)

    @abstractmethod
    def resumen(self):
        pass


class EarlyAccess(VideojuegoPC):
    def __init__(self, titulo, desarrollador, version_build):
        super().__init__(titulo, desarrollador)
        self.version_build = version_build

    def resumen(self):
        return f"(Early Access) {self.titulo} ({self.desarrollador}) - Build: {self.version_build} - Nota: {self.calcular_promedio()}"


class VersionOficial(VideojuegoPC):
    def __init__(self, titulo, desarrollador, version_lanzamiento):
        super().__init__(titulo, desarrollador)
        self.version_lanzamiento = version_lanzamiento

    def resumen(self):
        return f" {self.titulo} ({self.desarrollador}) - Versión: {self.version_lanzamiento} - Nota: {self.calcular_promedio()}"


juego1 = VersionOficial("The Witcher 3", "CD Projekt Red", "1.32")
juego1.agregar_calificacion(5.0)
juego1.agregar_calificacion(4.0)

juego2 = EarlyAccess("MENACE", "Overhype Studios", "0.3.0")
juego2.agregar_calificacion(4.0)
juego2.agregar_calificacion(3.5)

biblioteca = [juego1, juego2]

for juego in biblioteca:
    print(juego.resumen())
```
