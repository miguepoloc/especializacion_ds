Ejemplo sin usar D — Inversión de Dependencias (DIP)

```python
class Teclado():
    def leer_teclado(self) -> str:
        return "tecla w,a,s,d presionada -> mover personaje"

class Juego():
    def __init__(self, teclado: Teclado):
        self.teclado = teclado

    def jugar(self):
        accion = self.teclado.leer_teclado()
        print(f"Acción del juego: {accion}")

juego = Juego(Teclado())
juego.jugar()
```

EJmplo usando  D — Inversión de Dependencias (DIP)

```python
from abc import ABC, abstractmethod


class DispositivoEntrada(ABC):
    @abstractmethod
    def leer_entrada(self) -> str:
        pass

class Teclado(DispositivoEntrada):
    def leer_entrada(self) -> str:
        return "tecla w,a,s,d presionada -> mover personaje"

class MandoXbox(DispositivoEntrada):
    def leer_entrada(self) -> str:
        return "Joystick izquierdo movido -> mover personaje"

class MandoPlaystation(DispositivoEntrada):
    def leer_entrada(self) -> str:
        return "Joystick derecho movido -> mover personaje"

class Juego():
    def __init__(self, dispositivo: DispositivoEntrada):
        self.dispositivo = dispositivo

    def jugar(self):
        accion = self.dispositivo.leer_entrada()
        print(f"Acción del juego: {accion}")


print("Usando teclado:")
juego_teclado = Juego(Teclado())
juego_teclado.jugar()

print("\nUsando mando Xbox:")
juego_xbox = Juego(MandoXbox())
juego_xbox.jugar()

print("\nUsando mando PlayStation:")
juego_playstation = Juego(MandoPlaystation())
juego_playstation.jugar()
```
