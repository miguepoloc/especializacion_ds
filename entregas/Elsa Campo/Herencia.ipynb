```python
class Empleado:
    def __init__(self, nombre: str, salario: int) -> None:
        self.nombre: str = nombre
        self.salario: int = salario

    def presentarse(self) -> None:
        print(f"Hola, soy {self.nombre} y laboro como {type(self).__name__}")


class Diseñador(Empleado):
    def __init__(self, nombre: str, salario: int, programa: str) -> None:
        super().__init__(nombre=nombre, salario=salario)
        self.programa: str = programa

    def diseñar(self) -> None:
        print(f"Estoy diseñando usando {self.programa}")


class Programador(Empleado):
    def __init__(self, nombre: str, salario: int, lenguaje: str) -> None:
        super().__init__(nombre=nombre, salario=salario)
        self.lenguaje: str = lenguaje

    def programar(self) -> None:
        print(f"Estoy programando en {self.lenguaje}")
```

```python
mi_diseñador = Diseñador(
    nombre="Pedro", salario=5000000,  programa="Photoshop"
)

mi_diseñador.presentarse()
mi_diseñador.diseñar()

print(
    f"Nombre: {mi_diseñador.nombre}, "
    f"Salario: ${mi_diseñador.salario}, "
    f"Programa: {mi_diseñador.programa}"
)


mi_programador = Programador(
    nombre="David",  salario=7500000,  lenguaje="Python, Java, C++"
)

mi_programador.presentarse()
mi_programador.programar()

print(
    f"Nombre: {mi_programador.nombre}, "
    f"Salario: ${mi_programador.salario}, "
    f"Lenguaje: {mi_programador.lenguaje}"
)
```
