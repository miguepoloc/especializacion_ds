# Abierto-Cerrado

#codigo correcto

```python
from abc import ABC, abstractmethod

class Envio(ABC):

    @abstractmethod
    def calcular_costo(self) -> float:
        pass

class EnvioTerrestre(Envio):

    def __init__(self, distancia: float) -> None:
        self.distancia = distancia

    def calcular_costo(self) -> float:
        return self.distancia * 1500

class EnvioAereo(Envio):

    def __init__(self, peso: float) -> None:
        self.peso = peso

    def calcular_costo(self) -> float:
        return self.peso * 8000


envios: list[Envio] = [
    EnvioTerrestre(10),
    EnvioAereo(5),
]

for envio in envios:
    print(f"Costo: ${envio.calcular_costo()}")
```

#codigo incorrecto

```python
class Envio:

    def __init__(self, tipo: str, distancia: float = 0, peso: float = 0) -> None:

        self.tipo = tipo
        self.distancia = distancia
        self.peso = peso

    def calcular_costo(self) -> float:

        if self.tipo == "terrestre":
            return self.distancia * 1500

        elif self.tipo == "aereo":
            return self.peso * 8000

        else:
            return 0


terrestre = Envio(
    tipo="terrestre",
    distancia=10
)

aereo = Envio(
    tipo="aereo",
    peso=5
)

envios: list[Envio] = [terrestre, aereo]

for envio in envios:
    print(f"Costo: ${envio.calcular_costo()}")
```
