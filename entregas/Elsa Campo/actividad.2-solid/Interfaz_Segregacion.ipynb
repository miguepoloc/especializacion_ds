# Segregacion-Interfaces

#codigo correcto

```python

class Vehiculo:

    def conducir(self) -> None:
        pass

    def volar(self) -> None:
        pass

    def navegar(self) -> None:
        pass

class Carro(Vehiculo):

    def conducir(self) -> None:
        print("El carro está conduciendo.")

class Avion(Vehiculo):

    def conducir(self) -> None:
        print("El avión se desplaza por la pista.")

    def volar(self) -> None:
        print("El avión está volando.")

class Barco(Vehiculo):

    def navegar(self) -> None:
        print("El barco esta navegando por el agua.")
```

```python

carro = Carro()
avion = Avion()
barco = Barco()

carro.conducir()

avion.conducir()
avion.volar()

barco.navegar()
```
