# Principio de Inversion de Dependencia 

#codigo correcto

```python
import abc

class Almacenamiento(abc.ABC):

    @abc.abstractmethod
    def guardar(self, archivo: str) -> None:
        pass

class DiscoLocal(Almacenamiento):

    def guardar(self, archivo: str) -> None:
        print(f"Archivo {archivo} guardado en el disco local.")

class Nube(Almacenamiento):

    def guardar(self, archivo: str) -> None:
        print(f"Archivo {archivo} guardado en la nube.")

class MemoriaUSB(Almacenamiento):

    def guardar(self, archivo: str) -> None:
        print(f"Archivo {archivo} guardado en la memoria USB.")

class Archivos:

    def __init__(self, almacenamiento: Almacenamiento) -> None:
        self.almacenamiento = almacenamiento

    def guardar_archivo(self, archivo: str) -> None:
        self.almacenamiento.guardar(archivo)
```

```python
disco = DiscoLocal()
nube = Nube()
usb = MemoriaUSB()

gestor_disco = GestorArchivos(disco)
gestor_nube = GestorArchivos(nube)
gestor_usb = GestorArchivos(usb)

gestor_disco.guardar_archivo("informe.pdf")
gestor_nube.guardar_archivo("proyecto.py")
gestor_usb.guardar_archivo("presentacion.pptx")
```
