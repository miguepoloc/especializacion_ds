# Principio de Sustitucion

#codigo correcto

```python
class Empleado:
    def __init__(self, nombre: str, salario: float) -> None:
        self.nombre: str = nombre
        self.salario: float = salario

    def calcular_bono(self) -> float:
        return self.salario * 0.10


class Gerente(Empleado):
    def calcular_bono(self) -> float:
        return self.salario * 0.20

class Nomina:
    def __init__(self, empleado: Empleado) -> None:
        self.empleado: Empleado = empleado

    def obtener_bono(self) -> float:
        return self.empleado.calcular_bono()
```

```python
empleado = Empleado(
    nombre="Camilo",
    salario=8000000
)

gerente = Gerente(
    nombre="Andrea",
    salario=15000000
)

nomina_empleado = Nomina(empleado=empleado)
nomina_gerente = Nomina(empleado=gerente)

print(f"El empleado {empleado.nombre} recibe un bono de ", f"${nomina_empleado.obtener_bono():,.0f}")

print(f"La gerente {gerente.nombre} recibe un bono de ", f"${nomina_gerente.obtener_bono():,.0f}")
```
