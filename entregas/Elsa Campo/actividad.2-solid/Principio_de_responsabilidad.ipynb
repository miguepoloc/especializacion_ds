# Principio de Responsabilidad Unica


#codigo correcto


```python
class TemperaturaAire:
    def __init__(self, temperatura: float) -> None:
        self.temperatura: float = temperatura


class LectorTemperatura:
    def mostrar_temperatura(self, sensor: TemperaturaAire) -> None:
        print(f"Temperatura actual: {sensor.temperatura}°C")


class ControlAire:
    def controlar_aire(self, sensor: TemperaturaAire) -> None:
        if sensor.temperatura >= 31:
            print("Temperatura no aceptable")
        elif sensor.temperatura >= 16:
            print("Aire acondicionado encendido")
        else:
            print("Aire acondicionado apagado")
```

```python
sensor = TemperaturaAire(temperatura=17)

lector = LectorTemperatura()
control = ControlAire()

lector.mostrar_temperatura(sensor)
control.controlar_aire(sensor)
```

```python
class AireAcondicionado:
    def __init__(self, temperatura: float) -> None:
        self.temperatura: float = temperatura

    def mostrar_temperatura(self) -> None:
        print(f"Temperatura actual: {self.temperatura}°C")

    def controlar_aire(self) -> None:
        if self.temperatura >= 31:
            print("Temperatura no aceptable")
        elif self.temperatura >= 16:
            print("Aire acondicionado encendido")
        else:
            print("Aire acondicionado apagado")
```

```python
aire = AireAcondicionado(temperatura=29)

aire.mostrar_temperatura()
aire.controlar_aire()
```
