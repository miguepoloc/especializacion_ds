```python
class Pago:
    def __init__(self, monto: float, cliente: str) -> None:
        self.monto = monto
        self.cliente = cliente

    def procesar(self) -> str:
        pass

    def describir(self) -> str:
        return f"{self.cliente} - ${self.monto}"


class Tarjeta(Pago):
    def procesar(self) -> str:
        return "Pago realizado con tarjeta"


class Transferencia(Pago):
    def procesar(self) -> str:
        return "Pago realizado por transferencia"


class Efectivo(Pago):
    def procesar(self) -> str:
        return "Pago realizado en efectivo"


def realizar_pago(pago: Pago) -> None:
    print(f"Procesando pago de {pago.describir()}:")
    print(pago.procesar())
    print()
```

```python
pago1 = Tarjeta(500000, "Laura")
pago2 = Transferencia(850000, "José")
pago3 = Efectivo(200000, "Paul")

realizar_pago(pago1)
realizar_pago(pago2)
realizar_pago(pago3)
```

```python
def iniciar_pago(pago: Pago) -> None:
    print(f"Iniciando pago de {pago.describir()}:")
    print(pago.procesar())
    print()
```
