```python
class SensorTemperatura:

    def __init__(self, nombre, unidad, pin):
        self.nombre = nombre
        self.unidad = unidad
        self.pin = pin
        self.temperatura = 0

    def leer_temperatura(self):
        pass

    def mostrar_datos(self):
        print(f"Sensor: {self.nombre}")
        print(f"Pin: {self.pin}")
        print(f"Temperatura: {self.temperatura} {self.unidad}")

class SensorLM35(SensorTemperatura):

    def leer_temperatura(self):
        self.temperatura = 25.5
        return self.temperatura


class SensorDHT11(SensorTemperatura):

    def leer_temperatura(self):
        self.temperatura = 28.2
        return self.temperatura


sensor1 = SensorLM35("LM35", "°C", "A0")
sensor2 = SensorDHT11("DHT11", "°C", "D2")


sensor1.leer_temperatura()
sensor2.leer_temperatura()


sensor1.mostrar_datos()
sensor2.mostrar_datos()
```
