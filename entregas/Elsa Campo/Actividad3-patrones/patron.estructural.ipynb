#codigo correcto

```python
class Controlador:
    def __init__(self, luz):
        self.luz = luz

    def activar(self):
        self.luz.encender()

class Luz:
    def encender(self):
        pass

class LuzSala(Luz):
    def encender(self):
        print("Luz de la sala encendida")

class LuzJardin(Luz):
    def encender(self):
        print("Luz del jardín encendida")

class AppMovil(Controlador):
    def controlar(self):
        print("Controlando desde la aplicación móvil")
        self.activar()

class ControlVoz(Controlador):
    def controlar(self):
        print("Controlando por comando de voz")
        self.activar()


control = AppMovil(LuzSala())
control.controlar()

control2 = ControlVoz(LuzJardin())
control2.controlar()
```
