#codigo correcto

```python
class Computador:
    def __init__(self):
        self.procesador = None
        self.ram = None
        self.almacenamiento = None
        self.tarjeta_grafica = None

    def mostrar(self):
        print(f"Procesador: {self.procesador}")
        print(f"RAM: {self.ram}")
        print(f"Almacenamiento: {self.almacenamiento}")
        print(f"Tarjeta gráfica: {self.tarjeta_grafica}")


class ComputadorBuilder:

    def __init__(self):
        self.computador = Computador()

    def agregar_procesador(self, procesador):
        self.computador.procesador = procesador
        return self

    def agregar_ram(self, ram):
        self.computador.ram = ram
        return self

    def agregar_almacenamiento(self, almacenamiento):
        self.computador.almacenamiento = almacenamiento
        return self

    def agregar_tarjeta_grafica(self, tarjeta):
        self.computador.tarjeta_grafica = tarjeta
        return self

    def construir(self):
        return self.computador


pc = (
    ComputadorBuilder()
    .agregar_procesador("Intel Core i7")
    .agregar_ram("16 GB")
    .agregar_almacenamiento("SSD 1 TB")
    .agregar_tarjeta_grafica("RTX 4060")
    .construir()
)

pc.mostrar()
```
