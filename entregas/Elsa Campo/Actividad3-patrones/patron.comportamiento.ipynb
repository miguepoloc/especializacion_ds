#codigo incorrecto

```python
class CuentaUsuario:

    def __init__(self):
        self.estado = "activa"

    def iniciar_sesion(self):

        if self.estado == "activa":
            print("Acceso permitido")

        elif self.estado == "bloqueada":
            print("Acceso denegado: cuenta bloqueada")

        elif self.estado == "suspendida":
            print("Acceso denegado: cuenta suspendida")

usuario = CuentaUsuario()

usuario.iniciar_sesion()
usuario.estado = "bloqueada"
usuario.iniciar_sesion()
```

#codigo correcto

```python
class EstadoCuenta:

    def iniciar_sesion(self, cuenta):
        pass

class Activa(EstadoCuenta):

    def iniciar_sesion(self, cuenta):
        print("Acceso permitido")
        print("Bienvenido al sistema")

class Bloqueada(EstadoCuenta):

    def iniciar_sesion(self, cuenta):
        print("Acceso denegado")
        print("La cuenta está bloqueada")

class Suspendida(EstadoCuenta):

    def iniciar_sesion(self, cuenta):
        print("Acceso denegado")
        print("La cuenta está suspendida")

class CuentaUsuario:

    def __init__(self):
        self.estado = Activa()

    def cambiar_estado(self, estado):
        self.estado = estado

    def iniciar_sesion(self):
        self.estado.iniciar_sesion(self)


usuario = CuentaUsuario()
usuario.iniciar_sesion()

usuario.cambiar_estado(Bloqueada())
usuario.iniciar_sesion()

usuario.cambiar_estado(Suspendida())
usuario.iniciar_sesion()
```
