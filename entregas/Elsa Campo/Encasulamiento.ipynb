```python
class ReservaHotel:
    def __init__(self, habitacion: int, cliente: str,
                 noches: int, precio: float) -> None:
        self.__habitacion = habitacion
        self.__cliente = cliente
        self.__noches = noches
        self.__precio = precio

    def obtener_costo(self) -> float:
        return self.__noches * self.__precio

    def actualizar_estadia(self, dias: int) -> bool:
        if dias > 0:
            self.__noches = dias
            return True
        return False

    def obtener_reserva(self) -> str:
        return (f"Cliente: {self.__cliente}, "
                f"Habitación: {self.__habitacion}, "
                f"Noches: {self.__noches}, "
                f"Costo: ${self.obtener_costo()}")
```

```python
reserva = ReservaHotel(
    habitacion=308,
    cliente="Jose Robles",
    noches=6,
    precio=385.0
)

print(f"Reserva inicial: {reserva.obtener_reserva()}")
print(f"Costo total: ${reserva.obtener_costo()}")

reserva.actualizar_estadia(5)

print(f"Reserva actualizada: {reserva.obtener_reserva()}")
print(f"Nuevo costo total: ${reserva.obtener_costo()}")
```
