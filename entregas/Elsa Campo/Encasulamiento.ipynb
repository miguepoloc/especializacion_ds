## Ejemplo práctico

En este ejemplo:

1. `CuentaBancaria` encapsula los datos (`__titular` y `__saldo`) y los métodos que operan sobre ellos.

2. Los atributos `__titular` y `__saldo` son privados (indicado por el doble guion bajo).

3. Se proporcionan métodos públicos (`depositar`, `retirar`, `obtener_saldo`, `obtener_titular`) para interactuar con los datos privados.

4. Los métodos `depositar` y `retirar` incluyen lógica para mantener la integridad de los datos.

5. El acceso directo a `__saldo` desde fuera de la clase resulta en un error.

class CuentaBancaria {

    - saldo: float

    + retirar(cantidad): bool
    + obtener_saldo(): float
    + obtener_titular(): str

@startuml
class Motor{
    - velocidad: float
    - velocidad_maxima: float
    - encendido: bool
    + encender()
    + apagar()
    + frenar()
}
@enduml

```python
class Motor:
    def __init__(self, marca: str, velocidad_maxima: float) -> None:
        self.__marca: str = marca
        self.__velocidad: float = 0
        self.__velocidad_maxima: float = velocidad_maxima
        self.__encendido: bool = False

    def encender(self) -> bool:
        if not self.__encendido:
            self.__encendido = True
            return True
        return False

    def apagar(self) -> bool:
        if self.__encendido and self.__velocidad == 0:
            self.__encendido = False
            return True
        return False

    def acelerar(self, cantidad: float) -> bool:
        if self.__encendido and cantidad > 0:
            if self.__velocidad + cantidad <= self.__velocidad_maxima:
                self.__velocidad += cantidad
                return True
        return False

    def frenar(self, cantidad: float) -> bool:
        if cantidad > 0 and cantidad <= self.__velocidad:
            self.__velocidad -= cantidad
            return True
        return False

    def obtener_velocidad(self) -> float:
        return self.__velocidad

    def obtener_marca(self) -> str:
        return self.__marca

    def esta_encendido(self) -> bool:
        return self.__encendido
```

```python
motor = Motor(marca="Honda", velocidad_maxima=200)

print(f"Marca: {motor.obtener_marca()}")
print(f"Motor encendido: {motor.esta_encendido()}")
print(f"Velocidad: {motor.obtener_velocidad()} km/h")

motor.encender()
print(f"Motor encendido: {motor.esta_encendido()}")

motor.acelerar(50)
print(f"Velocidad después de acelerar: {motor.obtener_velocidad()} km/h")

motor.acelerar(30)
print(f"Velocidad después de acelerar: {motor.obtener_velocidad()} km/h")

motor.frenar(20)
print(f"Velocidad de frenado: {motor.obtener_velocidad()} km/h")

motor.frenar(60)
print(f"Velocidad de frenado: {motor.obtener_velocidad()} km/h")

motor.apagar()
print(f"Motor encendido: {motor.esta_encendido()}")
```
