# Principio de Inversión de Dependencias (Dependency Inversion Principle, DIP)


### incorrecto

```python
class MysqlRepository:
    def guardar(self, data):
        # guarda la información en la base de datos MySQL
        return True


class Uservice:
    def __init__(self):
        self.repository = MysqlRepository()

    def guardar_usuario(self, user):
        # lógica para guardar un usuario
        if self.repository.guardar(user):
            return "Usuario guardado exitosamente"
        else:
            return "Error al guardar el usuario"


service = Uservice()
print(service.guardar_usuario("juan"))
```

### Correcto

```python
from abc import ABC, abstractmethod
class RepositoryITF():
    @abstractmethod
    def guardar(self, data):
        pass

class MysqlRepository(RepositoryITF):
    def guardar(self, data):
        # guarda la información en la base de datos MySQL
        return True
class PostgresRepository(RepositoryITF):
    def guardar(self, data):
        # guarda la información en la base de datos PostgreSQL
        return True
class Service:
    def __init__(self, repository: RepositoryITF):
        self.repository = repository

    def guardar_usuario(self, user):
        # lógica para guardar un usuario
        if self.repository.guardar(user):
            return "Usuario guardado exitosamente"
        else:
            return "Error al guardar el usuario"
service = Service(MysqlRepository())
print(service.guardar_usuario("juan"))
service = Service(PostgresRepository())
print(service.guardar_usuario("juan"))
```
