# Principio de Sustitución de Liskov (Liskov Substitution Principle)

```python
class Calculo:
    def __init__(self, a:int, b:int):
        self.a = a
        self.b = b

    def calcular(self) -> int:
        return 0

class Sumar(Calculo):
    def __init__(self, a:int, b:int):
        super().__init__(a, b)
    def calcular(self) -> str:
        return "el resultado es " + str(self.a + self.b)
sumadora = Sumar(5, 10)
print(sumadora.calcular() + 10)
```

### correcto

```python
class Calculo:
    def __init__(self, a:int, b:int):
        self.a = a
        self.b = b

    def calcular(self) -> int:
        return 0

class Sumar(Calculo):
    def __init__(self, a:int, b:int):
        super().__init__(a, b)
    def calcular(self) -> int:
        return self.a + self.b

def promedio_nota(a,b,c):
    sumadora = Sumar(a, b)
    return sumadora.calcular() + c

promedio = promedio_nota(5, 10, 15)
print(promedio)
```
