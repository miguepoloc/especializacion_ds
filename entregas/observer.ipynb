```python
class Observador:
    def actualizar(self,titulo):
        pass

class Suscriptor(Observador):
    def __init__(self,nombre):
        self.nombre=nombre
    def actualizar(self,titulo):
        print(f'{self.nombre} vio: {titulo}')

class Canal:
    def __init__(self):
        self.observadores=[]
    def suscribir(self,observador):
        self.observadores.append(observador)
    def nuevo_video(self,titulo):
        for observador in self.observadores:
            observador.actualizar(titulo)

Canal1 = Canal()
Sub1=Suscriptor('Camila')
Sub2=Suscriptor('Julian')
Canal1.suscribir(Sub1)
Canal1.suscribir(Sub2)
Canal1.nuevo_video('Como aprender patrones de diseño')
```
