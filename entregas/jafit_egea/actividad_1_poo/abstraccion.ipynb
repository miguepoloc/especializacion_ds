## Ejercicio de Abstracción: Restaurante

### Ejemplo 1: Sistema de pedidos.

Para este contexto, la clase Restaurant es representada desde el punto de vista de los cocineros como un sistema de pedidos/órdenes.

```python
class Restaurant:
    def __init__(self, name: str, orders: int, is_kitchen_open: bool) -> None:
        self.name = name
        self.orders = orders
        self.is_kitchen_open = is_kitchen_open

    def prepare_order(self, order_id: int) -> str:
        if self.is_kitchen_open:
            return f"Order {order_id} is being prepared..."
        else:
            return f"Kitchen is currently closed. Cannot prepare order {order_id}."

    def __str__(self) -> str:
        return f"Restaurant Name: {self.name}, Orders: {self.orders}, Kitchen Open: {self.is_kitchen_open}"
```

```python
restaurant = Restaurant("Gourmet Restaurant", 8, True)
print(restaurant)
restaurant.prepare_order(2)
```

### Ejemplo 2: Sistema de reservas.

Para este contexto, la clase Restaurant es representada desde el punto de vista de los meseros como un sistema de reservas de mesas.

```python
class Restaurant:
    def __init__(self, name: str, tables: int, capacity: int) -> None:
        self.name = name
        self.tables = tables
        self.capacity = capacity

    def reserve_table(self, people: int) -> None:
        if people > self.capacity:
            print("No table available")
            return

        print(f"Table reserved for {people} people")

    def __str__(self) -> str:
        return f"{self.name} - {self.tables} tables"
```

```python
restaurant = Restaurant("Gourmet Haven", 5, 3)
print(restaurant)
restaurant.reserve_table(4)
restaurant.reserve_table(2)
```
