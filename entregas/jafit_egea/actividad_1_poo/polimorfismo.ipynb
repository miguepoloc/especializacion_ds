## Ejercicio de polimorfismo: tipos de producto en un restaurante

Usando el mismo ejemplo de los tipos de producto en un restaurante, podemos abordar el tema del polimorfismo

```python
from abc import ABC, abstractmethod

class Product(ABC):
    def __init__(self, name: str, price: float, available: bool) -> None:
            self.name: str = name
            self.price: float = price
            self.available: bool = available

    @abstractmethod
    def get_final_price(self) -> float:
        pass

    def show_info(self) -> str:
            status = "Disponible" if self.available else "No disponible"
            return f"{self.name} - ${self.price:.2f} - {status}"
```

```python
class Food(Product):
    def get_final_price(self) -> float:
        return self.price * 1.10  # la comida tiene un 10% de impuesto adicional

class Beverage(Product):
    def get_final_price(self) -> float:
        return self.price * 1.03  # las bebidas tienen un 3% de impuesto adicional

class Dessert(Product):
    def get_final_price(self) -> float:
        return self.price * 1.05  # los postres tienen un 5% de impuesto adicional
```

```python
food: Food = Food("Pizza", 15.0, True)
beverage: Beverage = Beverage("Coca-Cola", 3.0, True)
dessert: Dessert = Dessert("Tiramisu", 6.0, False)

for product in [food, beverage, dessert]:
    print(product.show_info())
    print(f"Precio final: ${product.get_final_price():.2f}\n")
```
