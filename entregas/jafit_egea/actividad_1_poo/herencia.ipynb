## Ejercicio de herencia: tipos de producto en un restaurante

Podemos aplicar la herencia para describir los tipos de producto que pueden haber en un restaurante.

### Clase padre Product

```python
class Product:
    def __init__(self, name: str, price: float, available: bool) -> None:
        self.name: str = name
        self.price: float = price
        self.available: bool = available

    def show_info(self) -> str:
        status = "Disponible" if self.available else "No disponible"
        return f"{self.name} - ${self.price:.2f} - {status}"

    def apply_discount(self, percentage: float) -> None:
        self.price -= self.price * (percentage / 100)
```

```python
class Food(Product):
    def __init__(self,name: str, price: float, available: bool, ingredients: list[str]) -> None:
        super().__init__(name=name, price=price, available=available)
        self.ingredients: list[str] = ingredients

    def show_ingredients(self) -> str:
        return f"{self.name} - Ingredientes: {', '.join(self.ingredients)}"
```

### Clase hija Beverage

```python
class Beverage(Product):
    def __init__(self,name: str, price: float, available: bool, size: str) -> None:
        super().__init__(name=name, price=price, available=available)
        self.size: str = size

    def set_size(self, size: str) -> None:
        self.size = size

    def show_info(self) -> str:
            return super().show_info() + f" - Tamaño: {self.size}"
```

### Clase hija Dessert

```python
class Dessert(Product):
    def __init__(self,name: str, price: float, available: bool, is_cool: bool) -> None:
        super().__init__(name=name, price=price, available=available)
        self.is_cool: bool = is_cool

    def serve(self) -> str:
        if self.is_cool:
            return f"{self.name} se sirve frío."
        else:
            return f"{self.name} se sirve a temperatura ambiente."
```

```python
food: Food = Food(name="Pizza", price=15.0, available=True, ingredients=["Harina", "Tomate", "Queso"])
beverage: Beverage = Beverage(name="Coca-Cola", price=2.5, available=True, size="500ml")
dessert: Dessert = Dessert(name="Helado", price=5.0, available=True, is_cool=True)

products: list[Product] = [food, beverage, dessert]

for p in products:
    print(p.show_info())

print(food.show_ingredients())

beverage.set_size("1L")
print(beverage.show_info())

print(dessert.serve())
```
