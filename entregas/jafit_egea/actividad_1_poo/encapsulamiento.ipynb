## Ejercicio de encapsulamiento: Producto

Podemos aplicar encapsulamiento para describir un producto bien sea de un supermercado, de un almacen, de una tienda de ropa, etc.

```python
class Product:
    def __init__(self, name: str, price: float, available: bool) -> None:
        self.name = name
        self.__price = price
        self.available = available

    def get_price(self) -> float:
        return self.__price

    def set_price(self, price: float) -> None:
        if price > 0:
            self.__price = price
```

```python
product = Product("Laptop", 1000.0, True)
print(f"Product: {product.name}, Price: {product.get_price()}, Available: {product.available}")

product.set_price(1200.0)
print(f"Product: {product.name}, Price: {product.get_price()}, Available: {product.available}")
```

```python
# Intentando acceder directamente a los atributos privados
try:
    print(product.__price)
except AttributeError as e:
    print(f"Error: {e}")
```
