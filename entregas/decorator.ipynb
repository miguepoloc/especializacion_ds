Crea tu propio Decorator para agregar funcionalidades como compresión o registro de logs a un sistema de mensajes.


```python
class SistemaMensajeria:
    def enviar(self, texto):
        print(f"Enviando mensaje: {texto}")


class DecoradorMensajeria(SistemaMensajeria):
    def __init__(self, mensajeria):
        self.mensajeria = mensajeria

    def enviar(self, texto):
        self.mensajeria.enviar(texto)


class Decoradorlogs(DecoradorMensajeria):
    def enviar(self, texto):
        print(f"registrando mensaje: {texto}")
        super().enviar(texto)

msg = Decoradorlogs(SistemaMensajeria())
msg.enviar("hola, mundo")
```

```python
class SistemaMensajeria:
    def enviar(self, texto):
        print(f"Enviando mensaje: {texto}")

# Clase base para los decoradores
class DecoradorMensajeria(SistemaMensajeria):
    def __init__(self, mensajeria):
        self.mensajeria = mensajeria  # ✅ Guardamos la referencia correctamente

    def enviar(self, texto):
        self.mensajeria.enviar(texto)  # ✅ Delegamos el comportamiento

# Decorador específico para registros/logs
class DecoradorLogs(DecoradorMensajeria):
    def enviar(self, texto):
        print(f"Registrando mensaje: {texto}")  # ✨ Funcionalidad añadida
        super().enviar(texto)                    # ✅ Llama al método de la clase padre

# Ejemplo de uso
msg = DecoradorLogs(SistemaMensajeria())
msg.enviar("hola, mundo")
```
