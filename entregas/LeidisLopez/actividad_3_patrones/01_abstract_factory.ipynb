# Abstract Factory - Patrón creacional

## Contexto

Un concesionario ofrece familias de vehículos: vehículos automáticos y vehículos mecánicos. Cada familia está compuesta por productos relacionados, como un carro y una moto.

## Problema

Si el cliente crea directamente cada clase concreta, debe conocer los nombres de todas las clases y puede combinar productos de familias diferentes por error. Además, cada vez que se agrega una nueva familia, es necesario modificar el código cliente.

En la primera parte se muestra la solución directa, sin patrón. En la segunda parte se aplica **Abstract Factory** para crear familias completas de productos relacionados.

````python
from abc import ABC, abstractmethod


class Carro(ABC):
    @abstractmethod
    def arrancar(self):
        pass

    @abstractmethod
    def detener(self):
        pass


class Moto(ABC):
    @abstractmethod
    def arrancar(self):
        pass

    @abstractmethod
    def detener(self):
        pass


# Solución directa, sin patrón:
# el cliente conoce y crea cada clase concreta.
# Esto acopla el código cliente a las implementaciones y permite mezclar
# productos de familias diferentes sin ninguna validación.
class CarroAutomatico(Carro):
    def __init__(self, marca, modelo, precio):
        self.marca = marca
        self.modelo = modelo
        self.precio = precio

    def arrancar(self):
        print(f"El carro {self.marca} {self.modelo} ha arrancado automáticamente.")

    def detener(self):
        print(f"El carro {self.marca} {self.modelo} se ha detenido automáticamente.")


class CarroMecanico(Carro):
    def __init__(self, marca, modelo, precio):
        self.marca = marca
        self.modelo = modelo
        self.precio = precio

    def arrancar(self):
        print(f"El carro {self.marca} {self.modelo} ha arrancado manualmente.")

    def detener(self):
        print(f"El carro {self.marca} {self.modelo} se ha detenido manualmente.")


class MotoAutomatica(Moto):
    def __init__(self, marca, modelo, precio):
        self.marca = marca
        self.modelo = modelo
        self.precio = precio

    def arrancar(self):
        print(f"La moto {self.marca} {self.modelo} ha arrancado automáticamente.")

    def detener(self):
        print(f"La moto {self.marca} {self.modelo} se ha detenido automáticamente.")


class MotoMecanica(Moto):
    def __init__(self, marca, modelo, precio):
        self.marca = marca
        self.modelo = modelo
        self.precio = precio

    def arrancar(self):
        print(f"La moto {self.marca} {self.modelo} ha arrancado manualmente.")

    def detener(self):
        print(f"La moto {self.marca} {self.modelo} se ha detenido manualmente.")
````

````python
# El cliente debe conocer las clases concretas y decidir qué crear.
carro = CarroMecanico("Toyota", "Camry", 25000)
moto = MotoAutomatica("Yamaha", "MT-07", 7500)

carro.arrancar()
carro.detener()
moto.arrancar()
moto.detener()
````

## Por qué la solución directa es problemática

El código anterior funciona, pero el cliente queda acoplado a `CarroMecanico` y `MotoAutomatica`. También es posible combinar accidentalmente un carro mecánico con una moto automática, aunque pertenezcan a familias diferentes.

La solución es delegar la creación de toda la familia a una fábrica abstracta. Así, el cliente solicita un carro y una moto a la misma fábrica, y recibe productos compatibles entre sí.

## Con el patrón Abstract Factory implementado

La implementación contiene todos los roles principales del patrón:

- **Fábrica abstracta:** `VehiculoFactory`, que declara `crear_carro` y `crear_moto`.
- **Fábricas concretas:** `FamiliaAutomaticaFactory` y `FamiliaMecanicaFactory`.
- **Productos abstractos:** las interfaces comunes de carro y moto quedan representadas por las operaciones `arrancar` y `detener`.
- **Productos concretos:** `CarroAutomatico`, `CarroMecanico`, `MotoAutomatica` y `MotoMecanica`.
- **Cliente:** `mostrar_familia`, que trabaja con `VehiculoFactory` sin conocer las clases concretas.

````python
from abc import ABC, abstractmethod


class VehiculoFactory(ABC):
    """Fábrica abstracta para crear una familia de vehículos compatibles."""

    @abstractmethod
    def crear_carro(self, marca, modelo, precio):
        pass

    @abstractmethod
    def crear_moto(self, marca, modelo, precio):
        pass


class FamiliaAutomaticaFactory(VehiculoFactory):
    """Fábrica concreta de la familia automática."""

    def crear_carro(self, marca, modelo, precio):
        return CarroAutomatico(marca, modelo, precio)

    def crear_moto(self, marca, modelo, precio):
        return MotoAutomatica(marca, modelo, precio)


class FamiliaMecanicaFactory(VehiculoFactory):
    """Fábrica concreta de la familia mecánica."""

    def crear_carro(self, marca, modelo, precio):
        return CarroMecanico(marca, modelo, precio)

    def crear_moto(self, marca, modelo, precio):
        return MotoMecanica(marca, modelo, precio)


# El cliente depende de VehiculoFactory y no conoce las clases concretas.
def mostrar_familia(factory):
    carro = factory.crear_carro("Toyota", "Camry", 25000)
    moto = factory.crear_moto("Yamaha", "MT-07", 7500)

    print(f"Familia creada: {carro.__class__.__name__} y {moto.__class__.__name__}")
    carro.arrancar()
    moto.arrancar()

    return carro, moto
````

````python
# Se puede cambiar la familia completa sin modificar el cliente.
familia_automatica = mostrar_familia(FamiliaAutomaticaFactory())
familia_mecanica = mostrar_familia(FamiliaMecanicaFactory())
````

## UML del patrón Abstract Factory

```plantuml
@startuml

    + crear_moto(marca, modelo, precio)

FamiliaMecanicaFactory ..> MotoMecanica : crea
@enduml
```

## ¿Por qué Abstract Factory y no otro patrón?

Se eligió **Abstract Factory** porque el concesionario debe crear familias completas de productos relacionados: un carro y una moto que pertenezcan a la misma familia automática o mecánica. La fábrica garantiza que los productos creados sean compatibles entre sí.

**Factory Method** sería apropiado para delegar la creación de un solo tipo de producto mediante subclases, pero aquí se necesita crear varios productos relacionados. **Builder** tampoco es el más adecuado, porque su objetivo principal es construir paso a paso un objeto complejo, no coordinar familias de objetos compatibles.
