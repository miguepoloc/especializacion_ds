# Herencia

## Aplicación en el Modelo Bancario

```python
from abc import ABC, abstractmethod
from datetime import datetime
from typing import List

class Cuenta(ABC):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0):
        # Atributos comunes
        self.numero_cuenta = numero_cuenta
        self.titular = titular
        self._saldo = saldo_inicial
        self.fecha_creacion = datetime.now()
        self.movimientos = []

    def depositar(self, cantidad: float) -> bool:
        if cantidad <= 0:
            print(f"Error: cantidad debe ser positiva")
            return False
        self._saldo += cantidad
        self.movimientos.append({"tipo": "deposito", "cantidad": cantidad})
        print(f"✓ Depósito ${cantidad:.2f} realizado")
        return True

    def retirar(self, cantidad: float) -> bool:
        if cantidad <= 0:
            print(f"Error: cantidad debe ser positiva")
            return False
        if cantidad > self._saldo:
            print(f"Error: saldo insuficiente")
            return False
        self._saldo -= cantidad
        self.movimientos.append({"tipo": "retiro", "cantidad": cantidad})
        print(f"✓ Retiro ${cantidad:.2f} realizado")
        return True

    def obtener_saldo(self) -> float:
        return self._saldo

    @abstractmethod
    def procesar_comisiones(self):
        pass


class CuentaAhorros(Cuenta):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0, tasa_interes: float = 0.05):
        super().__init__(numero_cuenta, titular, saldo_inicial)
        self.tasa_interes = tasa_interes

    def procesar_comisiones(self):
        interes = self._saldo * (self.tasa_interes / 12)
        self._saldo += interes
        print(f"Interés acreditado: ${interes:.2f}")


class CuentaCorriente(Cuenta):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0, comision_mensual: float = 10):
        super().__init__(numero_cuenta, titular, saldo_inicial)
        self.comision_mensual = comision_mensual

    def procesar_comisiones(self):
        self._saldo -= self.comision_mensual
        print(f"Comisión cobrada: ${self.comision_mensual:.2f}")
```

```python
ahorros = CuentaAhorros("AH-001", "Leidis", 1000, 0.05)
corriente = CuentaCorriente("CC-001", "Leidis", 2000, 15)

print(f"Saldo inicial: ${ahorros.obtener_saldo():.2f}  ← Método heredado")
ahorros.depositar(500)  # ← Método heredado
print(f"Saldo después de depósito: ${ahorros.obtener_saldo():.2f}")
ahorros.retirar(200)  # ← Método heredado
print(f"Saldo después de retiro: ${ahorros.obtener_saldo():.2f}")


print(f"Saldo inicial: ${corriente.obtener_saldo():.2f}  ← Método heredado")
corriente.depositar(1000)  # ← Método heredado
print(f"Saldo después de depósito: ${corriente.obtener_saldo():.2f}")
corriente.retirar(300)  # ← Método heredado
print(f"Saldo después de retiro: ${corriente.obtener_saldo():.2f}")


print(f"¿ahorros es instancia de Cuenta? {isinstance(ahorros, Cuenta)}")
print(f"¿corriente es instancia de Cuenta? {isinstance(corriente, Cuenta)}")
print(f"¿ahorros es instancia de CuentaAhorros? {isinstance(ahorros, CuentaAhorros)}")
print(f"¿corriente es instancia de CuentaCorriente? {isinstance(corriente, CuentaCorriente)}")


print("CuentaAhorros hereda estos atributos de Cuenta:")
print(f"  • numero_cuenta: {ahorros.numero_cuenta}")
print(f"  • titular: {ahorros.titular}")
print(f"  • _saldo: ${ahorros._saldo:.2f}")
print(f"  • fecha_creacion: {ahorros.fecha_creacion}")
print(f"  • movimientos: {ahorros.movimientos}")
print("\nCuentaAhorros agrega su propio atributo:")
print(f"  • tasa_interes: {ahorros.tasa_interes}")
```
