# Polimorfismo

## Aplicación en el Modelo Bancario

```python
from abc import ABC, abstractmethod
from datetime import datetime
from typing import List

class Cuenta(ABC):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0):
        self.numero_cuenta = numero_cuenta
        self.titular = titular
        self._saldo = saldo_inicial
        self.fecha_creacion = datetime.now()
        self.movimientos = []

    def depositar(self, cantidad: float) -> bool:
        if cantidad <= 0:
            return False
        self._saldo += cantidad
        return True

    def retirar(self, cantidad: float) -> bool:
        if cantidad <= 0 or cantidad > self._saldo:
            return False
        self._saldo -= cantidad
        return True

    def obtener_saldo(self) -> float:
        return self._saldo


    @abstractmethod
    def procesar_comisiones(self):
        pass


class CuentaAhorros(Cuenta):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0, tasa_interes: float = 0.05):
        super().__init__(numero_cuenta, titular, saldo_inicial)
        self.tasa_interes = tasa_interes

    def procesar_comisiones(self):
        interes = self._saldo * (self.tasa_interes / 12)
        self._saldo += interes
        print(f"[AHORROS] Interés de ${interes:.2f} acreditado. Saldo: ${self._saldo:.2f}")


class CuentaCorriente(Cuenta):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0, comision_mensual: float = 10):
        super().__init__(numero_cuenta, titular, saldo_inicial)
        self.comision_mensual = comision_mensual

    def procesar_comisiones(self):
        self._saldo -= self.comision_mensual
        print(f"[CORRIENTE] Comisión de ${self.comision_mensual:.2f} cobrada. Saldo: ${self._saldo:.2f}")


class CuentaInfantil(Cuenta):

    def __init__(self, numero_cuenta: str, titular: str, tutor: str, saldo_inicial: float = 0):
        super().__init__(numero_cuenta, titular, saldo_inicial)
        self.tutor = tutor

    def procesar_comisiones(self):
        bonificacion = 5.0
        self._saldo += bonificacion
        print(f"[INFANTIL] Bonificación de ${bonificacion:.2f} acreditada. Saldo: ${self._saldo:.2f}")


class CuentaEmpresa(Cuenta):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0, comision_transaccional: float = 0.01):
        super().__init__(numero_cuenta, titular, saldo_inicial)
        self.comision_transaccional = comision_transaccional

    def procesar_comisiones(self):
        comision = self._saldo * self.comision_transaccional
        self._saldo -= comision
        print(f"[EMPRESA] Comisión transaccional de ${comision:.2f} cobrada. Saldo: ${self._saldo:.2f}")
```

```python
class Cliente:
    def __init__(self, nombre: str, cedula: str):
        self.nombre = nombre
        self.cedula = cedula
        self._cuentas: List[Cuenta] = []

    def agregar_cuenta(self, cuenta: Cuenta):
        self._cuentas.append(cuenta)

    def procesar_todas_comisiones(self):
        print(f"\nCliente: {self.nombre}")
        print("-" * 70)
        for cuenta in self._cuentas:
            cuenta.procesar_comisiones()


cliente = Cliente("Leidis López", "1234567890")
cliente.agregar_cuenta(CuentaAhorros("AH-001", "Leidis López", 5000, 0.05))
cliente.agregar_cuenta(CuentaCorriente("CC-001", "Leidis López", 8000, 15))
cliente.agregar_cuenta(CuentaEmpresa("EM-001", "Mi Empresa", 20000, 0.015))


cliente.procesar_todas_comisiones()
```
