## Definición de la Clase Abstracta Base

```python
from abc import ABC, abstractmethod
from datetime import datetime
from typing import List


class Cuenta(ABC):

    def __init__(self, numero_cuenta: str, titular: str, saldo_inicial: float = 0):
        self.numero_cuenta = numero_cuenta
        self.titular = titular
        self._saldo = saldo_inicial          # PRIVADO - encapsulado, solo accesible mediante métodos
        self.fecha_creacion = datetime.now()
        self.movimientos: List[dict] = []


    def obtener_saldo(self) -> float:
        return self._saldo

    def obtener_numero_cuenta(self) -> str:
        return self.numero_cuenta

    def obtener_titular(self) -> str:
        return self.titular

    def depositar(self, cantidad: float) -> bool:
        if cantidad <= 0:
            print(f"Error: La cantidad debe ser positiva. Se intentó depositar {cantidad}")
            return False

        self._saldo += cantidad
        self._registrar_movimiento("deposito", cantidad)
        print(f"Depósito de ${cantidad:.2f} realizado. Nuevo saldo: ${self._saldo:.2f}")
        return True

    def retirar(self, cantidad: float) -> bool:
        if cantidad <= 0:
            print(f"Error: La cantidad debe ser positiva. Se intentó retirar {cantidad}")
            return False


        if cantidad > self._saldo:
            print(f" Error: Saldo insuficiente. Saldo actual: ${self._saldo:.2f}, se intentó retirar: ${cantidad:.2f}")
            return False

        self._saldo -= cantidad
        self._registrar_movimiento("retiro", cantidad)
        print(f"✓ Retiro de ${cantidad:.2f} realizado. Nuevo saldo: ${self._saldo:.2f}")
        return True

    def _registrar_movimiento(self, tipo: str, cantidad: float):
        self.movimientos.append({
            "tipo": tipo,
            "cantidad": cantidad,
            "fecha": datetime.now(),
            "saldo_resultado": self._saldo
        })

    def mostrar_estado(self):
        print("Estado de la cuenta:")
        print(f"Cuenta: {self.numero_cuenta}")
        print(f"Titular: {self.titular}")
        print(f"Saldo: ${self._saldo:.2f}")
        print(f"Tipo: {self.__class__.__name__}")
        print(f"Fecha creación: {self.fecha_creacion.strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
Cuenta = Cuenta("123456789", "Leidis Lopez", 1000)

Cuenta.depositar(500)
Cuenta.retirar(200)
Cuenta.mostrar_estado()
```
