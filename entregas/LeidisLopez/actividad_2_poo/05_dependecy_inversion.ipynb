# Dependency Inversion (Inversión de Dependencias)

El **Principio de Inversión de Dependencias** establece que **los módulos de alto nivel no deben depender de módulos de bajo nivel. Ambos deben depender de abstracciones**. En lugar de que las clases dependan de implementaciones concretas, deben depender de interfaces o clases abstractas.


## Ejemplo 1: Violación del Principio de Inversión de Dependencias

En este ejemplo, la clase `Transferencia` depende directamente de `BrebTransferencia`, una implementación concreta. Si en el futuro necesitamos usar otro método de transferencia (como Nequi), tendríamos que modificar la clase `Transferencia`.



```python
class BrebTransferencia:
    def transferir(self, cuenta_origen, cuenta_destino, monto):
        print(f"Transferencia de {monto} desde la cuenta {cuenta_origen} a la cuenta {cuenta_destino} realizada con éxito.")

class Transferencia:
    def __init__(self):
        self.breb_transferencia = BrebTransferencia()

    def transferir(self, cuenta_origen, cuenta_destino, monto):
        self.breb_transferencia.transferir(cuenta_origen, cuenta_destino, monto)


```

La clase `Transferencia` está acoplada a `BrebTransferencia`. Si queremos usar otro método de transferencia, tendríamos que modificar el constructor y la lógica de la clase `Transferencia`, violando el Principio de Abierto/Cerrado y este principio.


```python
transfer = Transferencia()
transfer.transferir("123456789", "987654321", 1000)
```

## Ejemplo 2: Aplicando el Principio de Inversión de Dependencias

Aquí creamos una interfaz abstracta `Transferir` que define el contrato para todos los métodos de transferencia. La clase `Transferencia` no depende de una implementación concreta, sino de la abstracción `Transferir`. Esto permite inyectar cualquier método de transferencia sin modificar la clase principal.



```python
class Transferir:
    def transferir(self, cuenta_origen, cuenta_destino, monto):
        pass

class BrebTransferencia(Transferir):
    def transferir(self, cuenta_origen, cuenta_destino, monto):
        print(f"Transferencia de {monto} desde la cuenta {cuenta_origen} a la cuenta {cuenta_destino} realizada con éxito a través de breb.")

class NequiTransferencia(Transferir):
    def transferir(self, cuenta_origen, cuenta_destino, monto):
        print(f"Transferencia de {monto} desde la cuenta {cuenta_origen} a la cuenta {cuenta_destino} realizada con éxito a través de Nequi.")

class NubankTransferencia(Transferir):
    def transferir(self, cuenta_origen, cuenta_destino, monto):
        print(f"Transferencia de {monto} desde la cuenta {cuenta_origen} a la cuenta {cuenta_destino} realizada con éxito a través de Nubank.")


class Transferencia:
    def __init__(self, metodo_transferencia: Transferir):
        self.metodo_transferencia = metodo_transferencia

    def transferir(self, cuenta_origen, cuenta_destino, monto):
        self.metodo_transferencia.transferir(cuenta_origen, cuenta_destino, monto)
```

Cada clase de transferencia (`BrebTransferencia`, `NequiTransferencia`, `NubankTransferencia`) implementa la interfaz `Transferir`. La clase `Transferencia` recibe cualquiera de estas implementaciones a través de inyección de dependencias, sin necesidad de conocer detalles específicos de cada una.


```python
transferencia = Transferencia(NequiTransferencia())
transferencia.transferir("123456789", "987654321", 1000)
```

Ahora podemos crear una instancia de `Transferencia` con `NequiTransferencia()` sin cambiar la clase `Transferencia`. Si mañana necesitamos usar `NubankTransferencia` o un nuevo método de transferencia, solo cambiamos qué clase inyectamos, no la clase principal. El código está desacoplado y sigue siendo flexible.
