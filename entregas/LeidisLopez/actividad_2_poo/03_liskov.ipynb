# Liskov Substitution (Principio de Sustitución de Liskov)

El **Principio de Sustitución de Liskov (LSP)** establece que los objetos de una clase derivada deben poder usarse en lugar de objetos de la clase base sin afectar el comportamiento correcto del programa. En otras palabras, una subclase debe ser completamente compatible con su clase padre.


## Ejemplo 1: Violación del Principio de Liskov

En este ejemplo, `CuentaAhorros` hereda de `CuentaBancaria`, pero cambia el comportamiento del método `retirar()` para devolver un string en lugar de un número. Esto viola el principio porque el código que espera un número recibirá un string, causando errores.


```python
class CuentaBancaria:
    def __init__(self):
        self.saldo = 0
        self.titular = ""

    def depositar(self, cantidad):
        self.saldo += cantidad
        return self.saldo

    def retirar(self, cantidad):
        if cantidad <= self.saldo:
            self.saldo -= cantidad
            return self.saldo
        else:
            print("Saldo insuficiente")

```

La clase `CuentaBancaria` establece que el método `retirar()` debe devolver un número (el nuevo saldo). Este es el contrato que debe cumplir cualquier subclase.


```python
class CuentaAhorros(CuentaBancaria):
    def __init__(self, tasa_interes):
        super().__init__()
        self.tasa_interes = tasa_interes

    def depositar(self, cantidad):
        super().depositar(cantidad)
        interes = self.calcular_interes()
        self.saldo += interes
        return f"Depósito de {cantidad} realizado. Interés ganado: {interes}. Saldo actual: {self.saldo}"

    def retirar(self, cantidad):
        if cantidad <= self.saldo:
            super().retirar(cantidad)
            return f"Retiro de {cantidad} realizado. Saldo actual: {self.saldo}"
        else:
            return "Saldo insuficiente"


    def calcular_interes(self):
        return self.saldo * self.tasa_interes
```

Aquí, `CuentaAhorros` cambia el contrato: `retirar()` devuelve un string con un mensaje en lugar de un número. Esto viola el LSP porque un cliente que use una referencia de `CuentaBancaria` esperará un número, no un string.


```python
Cuenta = CuentaAhorros(0.05)
Cuenta.depositar(1000)
print(Cuenta.retirar(500) - 10)
```

## Ejemplo 2: Aplicando el Principio de Liskov

En esta versión corregida, `CuentaAhorros` mantiene el contrato establecido por `CuentaBancaria`. El método `retirar()` devuelve el mismo tipo (un número), permitiendo que se use como sustituto de la clase base sin problemas.


```python
class CuentaAhorros(CuentaBancaria):
    def __init__(self, tasa_interes):
        super().__init__()
        self.tasa_interes = tasa_interes

    def depositar(self, cantidad):
        super().depositar(cantidad)
        interes = self.calcular_interes()
        self.saldo += interes
        return f"Depósito de {cantidad} realizado. Interés ganado: {interes}. Saldo actual: {self.saldo}"

    def retirar(self, cantidad):
        if cantidad <= self.saldo:
            super().retirar(cantidad)
            return self.saldo
        else:
            return "Saldo insuficiente"

    def calcular_interes(self):
        return self.saldo * self.tasa_interes
```

Ahora, `CuentaAhorros` respeta el contrato de `CuentaBancaria`. El método `retirar()` devuelve un número (el saldo), igual que en la clase base. Esto cumple el Principio de Liskov: una instancia de `CuentaAhorros` puede usarse en cualquier lugar donde se espere una `CuentaBancaria`.


```python
Cuenta = CuentaAhorros(0.05)
Cuenta.depositar(1000)
print(Cuenta.retirar(500) - 10)
```
