# Open/Closed (Abierto/Cerrado)

El **Principio de Abierto/Cerrado** establece que una clase debe estar **abierta para extensión, pero cerrada para modificación**. Esto significa que podemos agregar nuevas funcionalidades sin cambiar el código existente.




## Ejemplo 1: Violando el Principio Open/Closed

En el siguiente ejemplo, la clase `Envio` usa múltiples `if-elif` para manejar diferentes tipos de envío. Esto **viola el principio** porque cada vez que queremos agregar un nuevo tipo de envío, debemos modificar el método `calcular_costo()`.



```python
class Envio:
    def __init__(self, tipo_envio, peso):
        self.tipo_envio = tipo_envio
        self.peso = peso;


    def calcular_costo(self):
        if self.tipo_envio == "estandar":
            return self.peso * 5
        elif self.tipo_envio == "expreso":
            return self.peso * 10
        elif self.tipo_envio == "internacional":
            return self.peso * 20
        else:
            return 0
```

```python
envio1 = Envio("estandar", 3)

print(envio1.calcular_costo())
```

```python
envio2 = Envio("terrestre", 2)
print(envio2.calcular_costo())
```

Ahora el nuevo tipo de envío "terrestre" retorna 0 porque no está contemplado en el método. Para agregarlo, tendríamos que modificar la clase `Envio`, lo que **viola el principio Open/Closed**. Cada nueva necesidad requiere editar el código existente.


## Ejemplo 2: Aplicando el Principio Open/Closed

Aquí usamos una clase abstracta `TipoEnvio` como base, y creamos clases especializadas para cada tipo de envío. De esta forma, la clase original está **cerrada para modificación** (no cambia), pero está **abierta para extensión** (agregamos nuevos tipos heredando).


```python
from abc import ABC, abstractmethod

class TipoEnvio(ABC):
    @abstractmethod
    def calcular_costo(self, peso):
        pass


class EnvioEstandar(TipoEnvio):
    def calcular_costo(self, peso):
        return peso * 5


class EnvioExpreso(TipoEnvio):
    def calcular_costo(self, peso):
        return peso * 10


class EnvioInternacional(TipoEnvio):
    def calcular_costo(self, peso):
        return peso * 20

class EnvioTerrestre(TipoEnvio):
    def calcular_costo(self, peso):
        return 15
```

Cada clase especializada implementa el método `calcular_costo()` con su propia lógica. Si queremos agregar un nuevo tipo de envío, solo creamos una nueva clase sin modificar las existentes.


```python
class Envio:
    def calcular_costo(self, envio, peso):
        return envio.calcular_costo(peso)


envio = Envio()

print(envio.calcular_costo(EnvioEstandar(), 3))
print(envio.calcular_costo(EnvioExpreso(), 3))
print(envio.calcular_costo(EnvioInternacional(), 3))
print(envio.calcular_costo(EnvioTerrestre(), 3))
```

La clase `Envio` ahora solo recibe un objeto de tipo `TipoEnvio` y delega el cálculo a él. Si en el futuro necesitamos un nuevo tipo de envío (por ejemplo, `EnvioDrone`), solo creamos esa clase sin modificar `Envio`. El código original sigue siendo válido y no cambia.
