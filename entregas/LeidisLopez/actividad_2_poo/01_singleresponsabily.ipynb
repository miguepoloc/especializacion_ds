# Single Responsability



## Ejemplo 1: Violación del Principio de Responsabilidad Única

En el siguiente ejemplo, la clase `Concesionario` asume múltiples responsabilidades: gestionar vehículos, gestionar clientes y gestionar ventas. Esto es un problema porque:

- Si necesitamos cambiar cómo se gestionan los clientes, modificamos la clase `Concesionario`
- Si necesitamos cambiar la lógica de ventas, modificamos la misma clase
- Esto viola el SRP porque la clase tiene más de una razón para cambiar


```python
class Consesionario():
    def __init__(self, nombre, direccion, telefono):
        self.nombre = nombre
        self.direccion = direccion
        self.telefono = telefono
        self.vehiculos = []
        self.clientes = []

    def agregar_vehiculo(self, vehiculo):
        self.vehiculos.append(vehiculo)

    def mostrar_vehiculos(self):
        for vehiculo in self.vehiculos:
            print(
                f"Marca: {vehiculo['marca']}, "
                f"Modelo: {vehiculo['modelo']}, "
                f"Año: {vehiculo['año']}, "
                f"Precio: {vehiculo['precio']}"
            )

    def agregar_cliente(self, cliente):
        self.clientes.append(cliente)

    def mostrar_clientes(self):
        for cliente in self.clientes:
            print(
                f"Nombre: {cliente['nombre']}, "
                f"Dirección: {cliente['direccion']}, "
                f"Teléfono: {cliente['telefono']}"
            )

    def vender_vehiculo(self, vehiculo, cliente):
        if vehiculo in self.vehiculos:
            self.vehiculos.remove(vehiculo)
            print(f"Vehículo {vehiculo['marca']} {vehiculo['modelo']} vendido a {cliente['nombre']}.")
        else:
            print("El vehículo no está disponible en el concesionario.")
```

```python
Consesionario1 = Consesionario("AutoMax", "Calle 123", "555-1234")
Consesionario1.agregar_vehiculo({"marca": "Toyota", "modelo": "Corolla", "año": 2020, "precio": 20000})
Consesionario1.agregar_vehiculo({"marca": "Honda", "modelo": "Civic", "año": 2019, "precio": 18000})
Consesionario1.mostrar_vehiculos()
Consesionario1.agregar_cliente({"nombre": "Juan Perez", "direccion": "Calle 456", "telefono": "555-5678"})
Consesionario1.mostrar_clientes()
```

Como vemos en la ejecución anterior, la clase `Concesionario` maneja todo: vehículos, clientes y ventas. Esto causa sobrecarga de responsabilidades, dificil mantenimiento y alto acoplamiento.


## Ejemplo 2: Aplicando el Principio de Responsabilidad Única

Para cumplir con el SRP, refactorizamos el código anterior en tres clases independientes, cada una con una única responsabilidad:

### Clase 1: Concesionario
La clase `Concesionario` ahora **solo se encarga de gestionar el inventario de vehículos**. Sus responsabilidades son:
- Agregar vehículos al inventario
- Retirar vehículos del inventario
- Mostrar los vehículos disponibles


```python
class Consesionario():
    def __init__(self, nombre, direccion, telefono):
        self.nombre = nombre
        self.direccion = direccion
        self.telefono = telefono
        self.vehiculos = []

    def agregar_vehiculo(self, vehiculo):
        self.vehiculos.append(vehiculo)

    def retirar_vehiculo(self, vehiculo):
        if vehiculo in self.vehiculos:
            self.vehiculos.remove(vehiculo)
            return True
        return False

    def mostrar_vehiculos(self):
        for vehiculo in self.vehiculos:
            print(
                f"Marca: {vehiculo['marca']}, "
                f"Modelo: {vehiculo['modelo']}, "
                f"Año: {vehiculo['año']}, "
                f"Precio: {vehiculo['precio']}"
            )
```

### Clase 2: Clientes
La clase `Clientes` **solo se encarga de gestionar la base de datos de clientes**. Sus responsabilidades son:
- Agregar nuevos clientes
- Mostrar clientes registrados


```python
class Clientes():
    def __init__(self):
        self.clientes = []

    def agregar_cliente(self, cliente):
        self.clientes.append(cliente)

    def mostrar_clientes(self):
        for cliente in self.clientes:
            print(
                f"Nombre: {cliente['nombre']}, "
                f"Dirección: {cliente['direccion']}, "
                f"Teléfono: {cliente['telefono']}"
            )
```

### Clase 3: Ventas
La clase `Ventas` **solo se encarga de gestionar el proceso de ventas**. Sus responsabilidades son:
- Registrar transacciones de venta
- Ejecutar la lógica de venta (transferencia de vehículo del concesionario al cliente)
- Mostrar el historial de ventas



```python
class Ventas():
    def __init__(self):
        self.ventas = []

    def agregar_venta(self, venta):
        self.ventas.append(venta)

    def vender_vehiculo(self, concesionario, vehiculo, cliente):
        if vehiculo in concesionario.vehiculos:
            concesionario.vehiculos.remove(vehiculo)
            print(f"Vehículo {vehiculo['marca']} {vehiculo['modelo']} vendido a {cliente['nombre']}.")
        else:
            print("El vehículo no está disponible en el concesionario.")

    def mostrar_ventas(self):
        for venta in self.ventas:
            print(
                f"Vehículo: {venta['vehiculo']['marca']} {venta['vehiculo']['modelo']}, "
                f"Cliente: {venta['cliente']['nombre']}, "
                f"Fecha: {venta['fecha']}"
            )
```

```python
Consesionario1 = Consesionario("AutoMax", "Calle Falsa 123", "555-1234")
Consesionario1.agregar_vehiculo({"marca": "Toyota", "modelo": "Camry", "año": 2020, "precio": 25000})
Consesionario1.mostrar_vehiculos()

clientes = Clientes()
clientes.agregar_cliente({"nombre": "Juan Perez", "direccion": "Calle 456", "telefono": "555-5678"})
clientes.mostrar_clientes()

ventas = Ventas()
ventas.agregar_venta({"vehiculo": {"marca": "Toyota", "modelo": "Camry"}, "cliente": {"nombre": "Juan Perez"}, "fecha": "2024-06-01"})
ventas.vender_vehiculo(Consesionario1, {"marca": "Toyota", "modelo": "Camry", "año": 2020, "precio": 25000}, {"nombre": "Juan Perez", "direccion": "Calle 456", "telefono": "555-5678"})
ventas.mostrar_ventas()
```
