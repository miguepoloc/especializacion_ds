# Interface Segregation (Segregación de Interfaces)

El **Principio de Segregación de Interfaces** establece que **no se debe obligar a un cliente a depender de interfaces que no utiliza**. Las interfaces deben ser específicas y pequeñas, en lugar de grandes y generales.


## Ejemplo 1: Violando el principio de Interface Segregation

```python
from abc import ABC, abstractmethod

class Persona(ABC):
    @abstractmethod
    def get_students(self):
        pass

    @abstractmethod
    def get_curses(self):
        pass

    @abstractmethod
    def get_salary(self):
        pass

class Student(Persona):
    def __init__(self, name, curses):
        self.name = name
        self.curses = curses

    def get_students(self):
        pass

    def get_curses(self):
        print(f"The student {self.name} has the following courses: {', '.join(self.curses)}")

    def get_salary(self):
        pass


class Teacher(Persona):
    def __init__(self, name, salary, curses, students):
        self.name = name
        self.salary = salary
        self.curses = curses
        self.students = students

    def get_students(self):
        print(f"The teacher {self.name} has the following students: {', '.join(self.students)}")

    def get_curses(self):
        print(f"The teacher {self.name} teaches the following courses: {', '.join(self.curses)}")

    def get_salary(self):
        print(f"The teacher {self.name} has a salary of {self.salary}")

```

La clase abstracta `Persona` define una interfaz con tres responsabilidades completamente diferentes. El `Student` implementa `get_curses()` correctamente pero deja `get_students()` y `get_salary()` vacíos, lo que es un desperdicio.


```python
teacher1 = Teacher("John Doe", 50000, ["Math", "Science"], ["Alice Johnson", "Bob Williams"])
student1 = Student("Jane Smith", ["Math", "Science"])

teacher1.get_curses()
teacher1.get_salary()
teacher1.get_students()

student1.get_curses()
```

Observamos que `Student` implementa métodos innecesarios como `get_salary()`, lo cual no tiene sentido. Esto viola el Principio de Segregación de Interfaces porque estamos forzando a una clase a depender de métodos que nunca utilizará.


## Ejemplo 2: Aplicando el Principio de Segregación

En esta solución, dividimos la interfaz `Persona` en interfaces más pequeñas y específicas. Ahora tenemos `Persona` (solo para cursos) y `Worker` (solo para salarios). Cada clase implementa solo las interfaces que realmente necesita.

```python
from abc import ABC, abstractmethod

class Persona(ABC):
    @abstractmethod
    def get_curses(self):
        pass

class Worker(ABC):
    @abstractmethod
    def get_salary(self):
        pass



class Student(Persona):

    def __init__(self, name, grades):
        self.name = name
        self.grades = []
        self.curses = []

    def get_curses(self):
        print(f"The student {self.name} has the following courses: {', '.join(self.curses)}")


class Teacher(Persona, Worker):
    def __init__(self, name, salary, curses):
        self.name = name
        self.salary = salary
        self.curses = curses

    def get_curses(self):
        print(f"The teacher {self.name} teaches the following courses: {', '.join(self.curses)}")

    def get_salary(self):
        print(f"The teacher {self.name} has a salary of {self.salary}")
```

Ahora `Student` implementa solo `Persona` (necesita `get_curses()`), mientras que `Teacher` implementa ambas interfaces (necesita `get_curses()` y `get_salary()`). Cada clase depende solo de los métodos que realmente usa, cumpliendo así el Principio de Segregación de Interfaces.
