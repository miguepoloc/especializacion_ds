```python
# 1. El Flyweight: Guarda el estado intrínseco (compartido)
class TipoPieza:
    def __init__(self, nombre: str, color: str, simbolo: str):
        self.nombre = nombre
        self.color = color
        self.simbolo = simbolo  # Símbolo Unicode para renderizar en consola

    def mostrar(self, x: int, y: int):
        """Combina el estado intrínseco con el extrínseco (x, y)"""
        print(f"Pieza: {self.color} {self.nombre} en la posición ({x},{y}) -> Símbolo: {self.simbolo}")


# 2. La Fábrica Flyweight: Garantiza que los estados intrínsecos se compartan
class FabricaPiezas:
    _piezas = {}

    @classmethod
    def obtener_pieza(cls, nombre: str, color: str, simbolo: str) -> TipoPieza:
        # Creamos una clave única combinando nombre y color
        clave = (nombre, color)

        if clave not in cls._piezas:
            cls._piezas[clave] = TipoPieza(nombre, color, simbolo)
            print(f"🆕 [Fábrica] Creando nuevo objeto intrínseco para: {color} {nombre}")

        return cls._piezas[clave]

    @classmethod
    def total_objetos_en_memoria(cls) -> int:
        return len(cls._piezas)


# 3. El Contexto: Guarda el estado extrínseco (único) y referencia al Flyweight
class PiezaTablero:
    def __init__(self, x: int, y: int, tipo_pieza: TipoPieza):
        self.x = x  # Estado extrínseco
        self.y = y  # Estado extrínseco
        self.tipo_pieza = tipo_pieza  # Referencia al estado compartido

    def dibujar(self):
        self.tipo_pieza.mostrar(self.x, self.y)


# 4. El Cliente: Gestiona el tablero completo
class TableroAjedrez:
    def __init__(self):
        self.piezas_tablero = []

    def colocar_pieza(self, nombre: str, color: str, simbolo: str, x: int, y: int):
        # La fábrica nos da un objeto compartido (Flyweight)
        tipo = FabricaPiezas.obtener_pieza(nombre, color, simbolo)
        # Creamos el contenedor con su posición única
        pieza = PiezaTablero(x, y, tipo)
        self.piezas_tablero.append(pieza)

    def inicializar_partida_demo(self):
        """Coloca algunas piezas en el tablero para la demostración"""
        # Fila de peones blancos (Comparten el mismo objeto de tipo "Peón Blanco")
        for x in range(8):
            self.colocar_pieza("Peón", "Blanco", "♙", x, 1)

        # Fila de peones negros (Comparten el mismo objeto de tipo "Peón Negro")
        for x in range(8):
            self.colocar_pieza("Peón", "Negro", "♟", x, 6)

        # Torres (Mismo objeto intrínseco para las dos torres del mismo color)
        self.colocar_pieza("Torre", "Blanco", "♖", 0, 0)
        self.colocar_pieza("Torre", "Blanco", "♖", 7, 0)
        self.colocar_pieza("Torre", "Negro", "♜", 0, 7)
        self.colocar_pieza("Torre", "Negro", "♜", 7, 7)

    def renderizar_tablero(self):
        print("\n--- ♟️ ESTADO ACTUAL DEL TABLERO ♟️ ---")
        for pieza in self.piezas_tablero:
            pieza.dibujar()


# --- PRUEBA DEL SISTEMA ---
tablero = TableroAjedrez()

print("--- ⚙️ Inicializando fábrica y colocando piezas ---")
tablero.inicializar_partida_demo()

# Dibujamos el tablero
tablero.renderizar_tablero()

# Verificamos la optimización de memoria
print("\n--- 📊 REPORTE DE MEMORIA ---")
print(f"Piezas totales colocadas en el tablero: {len(tablero.piezas_tablero)}")
print(f"Objetos reales creados en memoria (Flyweights): {FabricaPiezas.total_objetos_en_memoria()}")
```

```python
class FlyweightCaracter:
    _c = {}
    def __new__(cls, simbolo, fuente):
        key = (simbolo, fuente)
        if key not in cls._c:
            obj = super().__new__(cls)
            cls._c[key] = obj
        return cls._c[key]
    def __init__(self, simbolo, fuente):
        self.simbolo = simbolo

texto = 'aac'
caracteres = [FlyweightCaracter(s, 'Arial')
              for s in texto]
```

```python
class TipoArbol:
    _tipos = {}
    def __new__(cls, modelo):
        if modelo not in cls._tipos:
            obj = super().__new__(cls)
            cls._tipos[modelo] = obj
        return cls._tipos[modelo]
    def __init__(self, modelo):
        self.modelo = modelo

class Arbol:
    def __init__(self, x, y, tipo):
        self.x, self.y = x, y
        self.tipo = tipo  # compartido

tipo_pino = TipoArbol('pino')
bosque = [Arbol(x, y, tipo_pino)
          for x, y in [(1, 2), (5, 7)]]
```
