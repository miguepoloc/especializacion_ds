```python
class Organizacion:
    def mostrar_organizacion(self, nivel=0):
        pass

class Empleados(Organizacion):
    def __init__(self, nombre, puesto):
        self.nombre = nombre
        self.puesto = puesto

    def mostrar_organizacion(self, nivel=0):
        # Multiplica los espacios por el nivel para crear la jerarquía visual
        print("    " * nivel + f"• {self.nombre} ({self.puesto})")

class Departamentos(Organizacion):
    def __init__(self, nombre):
        self.nombre = nombre
        self.componentes = [] # Cambiado a componentes para aceptar empleados o subdepartamentos

    def agregar(self, componente):
        self.componentes.append(componente)

    def mostrar_organizacion(self, nivel=0):
        print("    " * nivel + f"■ Departamento: {self.nombre}")
        # Se eliminó el bucle duplicado
        for componente in self.componentes:
            componente.mostrar_organizacion(nivel + 1)

# Creación de empleados
empleado1 = Empleados("Juan Perez", "Programador")
empleado2 = Empleados("Maria García", "Diseñadora")
empleado3 = Empleados("Juliana Torres", "Gerente de Recursos Humanos")

# Creación de departamentos de nivel inferior
tecnologia = Departamentos("Tecnologia")
Recursos_Humanos = Departamentos("Recursos Humanos")

# Asignación de empleados a sus departamentos
tecnologia.agregar(empleado1)
tecnologia.agregar(empleado2)
Recursos_Humanos.agregar(empleado3)

# Estructura principal (Empresa)
empresa = Departamentos("Empresa Z")
empresa.agregar(tecnologia)
empresa.agregar(Recursos_Humanos)

# Mostrar la estructura completa
empresa.mostrar_organizacion()
```
