# Patrón creacional: Factory Method

En este ejemplo una plataforma necesita crear reportes en distintos formatos. Primero veremos qué pasa cuando el código cliente decide todas las clases y después lo organizaremos con Factory Method.

## Ejemplo malo: el cliente crea cada reporte

El cliente está lleno de `if` y conoce los detalles de cada reporte. Si aparece un formato nuevo, hay que modificar esta función.

````python
class ReportePDF:
    def exportar(self, datos):
        return f'PDF: {datos}'

class ReporteCSV:
    def exportar(self, datos):
        return f'CSV: {datos}'

def exportar_reporte(formato, datos):
    if formato == 'pdf':
        reporte = ReportePDF()
    elif formato == 'csv':
        reporte = ReporteCSV()
    else:
        raise ValueError('Formato no soportado')
    print(reporte.exportar(datos))

exportar_reporte('pdf', 'ventas de agosto')
exportar_reporte('csv', 'ventas de agosto')
````

## Ejemplo bueno: Factory Method

Ahora cada creador sabe qué producto debe construir. El flujo general está en la clase base y el cliente solo usa `generar_reporte()`.

````python
from abc import ABC, abstractmethod

class Reporte(ABC):
    @abstractmethod
    def exportar(self, datos):
        pass

class ReportePDF(Reporte):
    def exportar(self, datos):
        return f'PDF: {datos}'

class ReporteCSV(Reporte):
    def exportar(self, datos):
        return f'CSV: {datos}'

class GeneradorReporte(ABC):
    @abstractmethod
    def crear_reporte(self):
        pass

    def generar_reporte(self, datos):
        reporte = self.crear_reporte()
        print(reporte.exportar(datos))

class GeneradorPDF(GeneradorReporte):
    def crear_reporte(self):
        return ReportePDF()

class GeneradorCSV(GeneradorReporte):
    def crear_reporte(self):
        return ReporteCSV()

GeneradorPDF().generar_reporte('ventas de agosto')
GeneradorCSV().generar_reporte('ventas de agosto')
````

## UML

```plantuml
@startuml
abstract class GeneradorReporte {
  + crear_reporte()
  + generar_reporte(datos)
}
class GeneradorPDF
class GeneradorCSV
GeneradorReporte <|-- GeneradorPDF
GeneradorReporte <|-- GeneradorCSV
abstract class Reporte {
  + exportar(datos)
}
class ReportePDF
class ReporteCSV
Reporte <|-- ReportePDF
Reporte <|-- ReporteCSV
GeneradorReporte ..> Reporte
@enduml
```

## ¿Por qué elegí este patrón?

Lo elegí porque el problema principal era decidir qué objeto crear. Factory Method deja esa decisión en los creadores concretos y evita que el código que genera reportes crezca con más condiciones.

## Conclusión

El ejemplo malo funciona, pero queda muy unido a PDF y CSV. Con Factory Method puedo agregar, por ejemplo, `GeneradorJSON` sin cambiar el flujo general de generación.
