# Patrón estructural: Adapter

Una aplicación de viajes espera consultar el clima con `consultar_ciudad()`, pero el servicio externo usa `get_weather()` y devuelve otras claves. Adapter sirve para conectar ambas interfaces sin modificar el servicio externo.

## Ejemplo malo: el cliente conoce al servicio externo

Para usar el servicio directamente, el cliente debe recordar otro nombre de método y traducir la respuesta. Si se cambia de proveedor, hay que buscar y cambiar todo el código cliente.

````python
class ServicioClimaExterno:
    def get_weather(self, city):
        return {'city': city, 'temp_c': 24}

def mostrar_clima_malo(servicio, ciudad):
    respuesta = servicio.get_weather(ciudad)
    print(f"Clima en {respuesta['city']}: {respuesta['temp_c']} C")

mostrar_clima_malo(ServicioClimaExterno(), 'Santa Marta')
````

## Ejemplo bueno: Adapter

El adaptador ofrece la interfaz que nuestra aplicación espera y por dentro traduce la llamada y la respuesta del servicio externo.

````python
class ServicioClimaExterno:
    def get_weather(self, city):
        return {'city': city, 'temp_c': 24}

class ClimaAdapter:
    def __init__(self, servicio_externo):
        self.servicio_externo = servicio_externo

    def consultar_ciudad(self, ciudad):
        datos = self.servicio_externo.get_weather(ciudad)
        return {'ciudad': datos['city'], 'temperatura': datos['temp_c']}

def mostrar_clima(servicio, ciudad):
    datos = servicio.consultar_ciudad(ciudad)
    print(f"Clima en {datos['ciudad']}: {datos['temperatura']} C")

clima = ClimaAdapter(ServicioClimaExterno())
mostrar_clima(clima, 'Santa Marta')
````

## UML

```plantuml
@startuml
class AplicacionViajes {
  + mostrar_clima(servicio, ciudad)
}
class ClimaAdapter {
  - servicio_externo
  + consultar_ciudad(ciudad)
}
class ServicioClimaExterno {
  + get_weather(city)
}
AplicacionViajes --> ClimaAdapter
ClimaAdapter --> ServicioClimaExterno
@enduml
```

## ¿Por qué elegí este patrón?

Lo elegí porque el servicio externo ya existe y no conviene modificarlo. Adapter concentra en un solo lugar las diferencias de nombres y formato.

## Conclusión

Sin Adapter, la integración se repite en el cliente. Con Adapter, la aplicación trabaja con una interfaz propia y cambiar de proveedor requiere cambiar principalmente el adaptador.
