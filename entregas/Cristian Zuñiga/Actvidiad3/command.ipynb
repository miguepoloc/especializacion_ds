# Patrón de comportamiento: Command

Una cafetería recibe órdenes para preparar bebidas. Veremos primero un código donde el menú ejecuta todo directamente y después convertiremos cada orden en un objeto que se puede guardar y ejecutar.

## Ejemplo malo: el menú hace todo directamente

El menú conoce cada operación y no puede guardar órdenes para procesarlas después. Agregar una nueva bebida obliga a modificar la lógica del menú.

````python
class Cafetera:
    def preparar_cafe(self):
        print('Preparando cafe')

    def preparar_te(self):
        print('Preparando te')

def atender_pedido(tipo):
    cafetera = Cafetera()
    if tipo == 'cafe':
        cafetera.preparar_cafe()
    elif tipo == 'te':
        cafetera.preparar_te()

atender_pedido('cafe')
atender_pedido('te')
````

## Ejemplo bueno: Command

Cada pedido se convierte en un comando. La cola puede guardarlo, ejecutarlo más tarde y conservar un historial de lo que ya se procesó.

````python
from abc import ABC, abstractmethod

class Cafetera:
    def preparar_cafe(self):
        print('Preparando cafe')

    def preparar_te(self):
        print('Preparando te')

class ComandoBebida(ABC):
    @abstractmethod
    def ejecutar(self):
        pass

class PrepararCafe(ComandoBebida):
    def __init__(self, cafetera):
        self.cafetera = cafetera

    def ejecutar(self):
        self.cafetera.preparar_cafe()

class PrepararTe(ComandoBebida):
    def __init__(self, cafetera):
        self.cafetera = cafetera

    def ejecutar(self):
        self.cafetera.preparar_te()

class ColaPedidos:
    def __init__(self):
        self.pendientes = []
        self.historial = []

    def agregar(self, comando):
        self.pendientes.append(comando)

    def procesar(self):
        while self.pendientes:
            comando = self.pendientes.pop(0)
            comando.ejecutar()
            self.historial.append(comando)

cafetera = Cafetera()
cola = ColaPedidos()
cola.agregar(PrepararCafe(cafetera))
cola.agregar(PrepararTe(cafetera))
cola.procesar()
print(f'Pedidos procesados: {len(cola.historial)}')
````

## UML

```plantuml
@startuml

  + ejecutar()
}

  + preparar_cafe()
  + preparar_te()
}

  - pendientes
  - historial
  + agregar(comando)
  + procesar()
}
PrepararCafe --> Cafetera
PrepararTe --> Cafetera
ColaPedidos --> ComandoBebida
@enduml
```

## ¿Por qué elegí este patrón?

Lo elegí porque una orden puede necesitar guardarse y ejecutarse después. Command separa quién pide la acción de quién la realiza, y facilita agregar una cola, un historial o reintentos.

## Conclusión

El código malo mezcla el menú con las acciones de la cafetera. Con Command, cada pedido tiene su propio objeto y la cola puede manejar todos los pedidos de la misma forma.
