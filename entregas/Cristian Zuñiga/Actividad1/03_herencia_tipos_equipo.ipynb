# Actividad 1 - Herencia en Tipos de Equipo

Este notebook extiende una clase base Equipo para crear tipos especializados.

## Objetivo 

1. Reutilizar comportamiento comun en una clase base.
2. Agregar atributos propios en subclases.
3. Sobrescribir metodos para especializar el resultado.

```python
class Equipo:
    def __init__(self, nombre: str, ciudad: str) -> None:
        self.nombre: str = nombre
        self.ciudad: str = ciudad
        self._puntos: int = 0

    def sumar_victoria(self) -> None:
        self._puntos += 3

    def resumen(self) -> str:
        return f"{self.nombre} ({self.ciudad}) - puntos: {self._puntos}"


class EquipoProfesional(Equipo):
    def __init__(self, nombre: str, ciudad: str, categoria: str, patrocinador: str) -> None:
        super().__init__(nombre, ciudad)
        self.categoria: str = categoria
        self.patrocinador: str = patrocinador

    def resumen(self) -> str:
        base = super().resumen()
        return f"{base} - categoria: {self.categoria} - sponsor: {self.patrocinador}"


class EquipoUniversitario(Equipo):
    def __init__(self, nombre: str, ciudad: str, universidad: str, facultad: str) -> None:
        super().__init__(nombre, ciudad)
        self.universidad: str = universidad
        self.facultad: str = facultad

    def resumen(self) -> str:
        base = super().resumen()
        return f"{base} - universidad: {self.universidad} - facultad: {self.facultad}"
```

```python
equipo_profesional = EquipoProfesional(
    nombre="Tigres FC",
    ciudad="Medellin",
    categoria="Primera A",
    patrocinador="Marca X",
)
equipo_universitario = EquipoUniversitario(
    nombre="Leones U",
    ciudad="Cali",
    universidad="Universidad del Valle",
    facultad="Ingenieria",
)

equipo_profesional.sumar_victoria()
equipo_universitario.sumar_victoria()

print(equipo_profesional.resumen())
print(equipo_universitario.resumen())

assert "categoria: Primera A" in equipo_profesional.resumen()
assert "universidad: Universidad del Valle" in equipo_universitario.resumen()
assert equipo_profesional._puntos == 3
assert equipo_universitario._puntos == 3
print("Herencia verificada")
```

## Por que aqui usamos herencia

- Porque ambos tipos de equipo comparten estructura y reglas basicas.
- Porque super evita duplicar inicializacion y logica comun.
- Porque cada subclase puede especializar el resumen con su contexto.
