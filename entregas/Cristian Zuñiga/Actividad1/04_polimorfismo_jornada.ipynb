# Actividad 1 - Polimorfismo en una Jornada

Este notebook muestra polimorfismo aplicando la misma operacion a equipos de tipos diferentes.

## Objetivo 

1. Definir una interfaz comun para el estilo de juego.
2. Hacer que varias subclases respondan distinto al mismo mensaje.
3. Procesar una lista homogenea de tipo base sin preguntar tipo concreto.

```python
from abc import ABC, abstractmethod


class EquipoBase(ABC):
    def __init__(self, nombre: str) -> None:
        self.nombre: str = nombre

    @abstractmethod
    def presentar_estilo(self) -> str:
        raise NotImplementedError


class EquipoProfesional(EquipoBase):
    def presentar_estilo(self) -> str:
        return f"{self.nombre}: presion alta y transiciones rapidas"


class EquipoUniversitario(EquipoBase):
    def presentar_estilo(self) -> str:
        return f"{self.nombre}: bloque medio y salida asociativa"


class EquipoFormativo(EquipoBase):
    def presentar_estilo(self) -> str:
        return f"{self.nombre}: prioridad al desarrollo tecnico"


def mostrar_estilos(equipos: list[EquipoBase]) -> list[str]:
    return [equipo.presentar_estilo() for equipo in equipos]
```

```python
equipos: list[EquipoBase] = [
    EquipoProfesional(nombre="Tigres FC"),
    EquipoUniversitario(nombre="Leones U"),
    EquipoFormativo(nombre="Cantera Norte"),
]

mensajes = mostrar_estilos(equipos)
for mensaje in mensajes:
    print(mensaje)

assert mensajes == [
    "Tigres FC: presion alta y transiciones rapidas",
    "Leones U: bloque medio y salida asociativa",
    "Cantera Norte: prioridad al desarrollo tecnico",
]
print("Polimorfismo verificado")
```

## Por que aqui usamos polimorfismo

- Porque una sola funcion trabaja con varios tipos sin condicionales por tipo.
- Porque cada subclase implementa su propia respuesta al mismo metodo.
- Porque el sistema queda abierto para agregar nuevos tipos de equipo.
