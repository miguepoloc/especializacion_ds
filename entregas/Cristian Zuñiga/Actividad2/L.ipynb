# Principio de Sustitución de Liskov (LSP)


## Caso problemático: mala práctica

La clase `Cuadrado` hereda de `Rectangulo`, pero cambia el comportamiento esperado.
Esto rompe la lógica de la superclase.

```python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def calcular_area(self):
        return self.ancho * self.alto


class Cuadrado(Rectangulo):
    def __init__(self, lado):
        super().__init__(lado, lado)

    def cambiar_dimensiones(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto


rectangulo = Rectangulo(4, 5)
print("Área del rectángulo:", rectangulo.calcular_area())

cuadrado = Cuadrado(4)
cuadrado.cambiar_dimensiones(2, 8)
print("Área del cuadrado:", cuadrado.calcular_area())
```

## Caso correcto: buena práctica

Cada figura mantiene su comportamiento sin romper el contrato general.
La subclase `Cuadrado` no se comporta como un rectángulo con dimensiones variables.
Por eso es más correcto que cada clase implemente su propia lógica.

```python
class Figura:
    def calcular_area(self):
        raise NotImplementedError("Debe implementar calcular_area")


class Rectangulo(Figura):
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def calcular_area(self):
        return self.ancho * self.alto


class Cuadrado(Figura):
    def __init__(self, lado):
        self.lado = lado

    def calcular_area(self):
        return self.lado * self.lado


rectangulo = Rectangulo(4, 5)
print("Área del rectángulo:", rectangulo.calcular_area())
cuadrado = Cuadrado(4)
print("Área del cuadrado:", cuadrado.calcular_area())
```
