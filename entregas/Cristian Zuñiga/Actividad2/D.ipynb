# Principio de Inversión de Dependencias (DIP)



## Caso problemático: mala práctica

La clase `Reporte` depende directamente de `BaseDatosMySQL`.
Si se cambia a PostgreSQL o a un archivo, hay que modificar la clase principal.

```python
class BaseDatosMySQL:
    def guardar(self, informacion):
        print("Guardando en MySQL:", informacion)


class Reporte:
    def __init__(self):
        self.base_datos = BaseDatosMySQL()

    def generar(self, datos):
        self.base_datos.guardar(datos)
        print("Reporte generado")


reporte = Reporte()
reporte.generar({"usuario": "Cristian", "estado": "activo"})
```

## Caso correcto: buena práctica

Se define una abstracción `Repositorio` y luego se inyectan distintas implementaciones.
La clase `Reporte` depende de la abstracción, no de la implementación concreta.

```python
class Repositorio:
    def guardar(self, informacion):
        raise NotImplementedError("Debes implementar guardar")


class RepositorioMySQL(Repositorio):
    def guardar(self, informacion):
        print("Guardando en MySQL:", informacion)


class RepositorioArchivo(Repositorio):
    def guardar(self, informacion):
        print("Guardando en archivo:", informacion)


class Reporte:
    def __init__(self, repositorio:Repositorio):
        self.repositorio = repositorio

    def generar(self, datos):
        self.repositorio.guardar(datos)
        print("Reporte generado")


reporte_sql = Reporte(RepositorioMySQL())
reporte_sql.generar({"usuario": "Cristian", "estado": "activo"})

reporte_archivo = Reporte(RepositorioArchivo())
reporte_archivo.generar({"usuario": "Cristian", "estado": "activo"})
```
