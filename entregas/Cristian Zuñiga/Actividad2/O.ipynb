# Principio de Abierto/Cerrado (OCP)



## Caso problemático: mala práctica

La clase `CalculadorDescuento` tiene muchos `if` para cada tipo de cliente.
Cada nueva categoría exige cambiar ese mismo código.

```python
class CalculadorDescuento:
    def __init__(self, tipo_cliente, valor_compra):
        self.tipo_cliente = tipo_cliente
        self.valor_compra = valor_compra

    def calcular(self):
        if self.tipo_cliente == "regular":
            return self.valor_compra * 0.05
        elif self.tipo_cliente == "estudiante":
            return self.valor_compra * 0.10
        elif self.tipo_cliente == "premium":
            return self.valor_compra * 0.20
        else:
            return 0

precio = 500
calculador = CalculadorDescuento("premium", precio)
print("Descuento:", calculador.calcular())
```

## Caso correcto: buena práctica

Se crean diferentes clases para cada estrategia de descuento.
La clase `Cliente` usa una estrategia sin depender de muchos condicionales.

```python
class Descuento:
    def aplicar(self, valor_compra):
        raise NotImplementedError("Debes implementar aplicar.")

class DescuentoRegular(Descuento):
    def aplicar(self, valor_compra):
        return valor_compra * 0.05

class DescuentoEstudiante(Descuento):
    def aplicar(self, valor_compra):
        return valor_compra * 0.10

class DescuentoPremium(Descuento):
    def aplicar(self, valor_compra):
        return valor_compra * 0.20

class Cliente:
    def __init__(self, estrategia_descuento):
        self.estrategia_descuento = estrategia_descuento

    def pagar(self, valor_compra):
        descuento = self.estrategia_descuento.aplicar(valor_compra)
        total = valor_compra - descuento
        return total

cliente = Cliente(DescuentoPremium())
print("Total a pagar:", cliente.pagar(500))

cliente_estudiante = Cliente(DescuentoEstudiante())
print("Total para estudiante:", cliente_estudiante.pagar(500))
```
