# Principio de Responsabilidad Única (SRP)



## Caso problemático: mala práctica

La clase `SistemaEstudiante` hace varias tareas a la vez: valida datos, calcula promedios y guarda información.
Si cambia una parte, quizá tengamos que tocar la clase completa.

```python
class SistemaEstudiante:
    def __init__(self):
        self.estudiantes = []

    def registrar_estudiante(self, nombre, edad, notas):
        if not nombre or len(nombre.strip()) < 2:
            raise ValueError("El nombre no es válido.")

        if edad < 0 or edad > 120:
            raise ValueError("La edad no es válida.")

        if len(notas) != 3:
            raise ValueError("Se requieren exactamente tres notas.")

        promedio = sum(notas) / len(notas)
        estudiante = {
            "nombre": nombre,
            "edad": edad,
            "notas": notas,
            "promedio": promedio
        }

        self.estudiantes.append(estudiante)
        print("Estudiante registrado:", estudiante)

sistema = SistemaEstudiante()
sistema.registrar_estudiante("Ana", 20, [4.5, 4.0, 5.0])
```

## Caso correcto: buena práctica

Ahora separamos responsabilidades:
- `ValidadorEstudiante` valida datos
- `CalculadorPromedio` calcula el promedio
- `RepositorioEstudiante` guarda la información
- `RegistroEstudiante` coordina todo

```python
class ValidadorEstudiante:
    def validar_nombre(self, nombre):
        if not nombre or len(nombre.strip()) < 2:
            raise ValueError("El nombre no es válido.")

    def validar_edad(self, edad):
        if edad < 0 or edad > 120:
            raise ValueError("La edad no es válida.")

    def validar_notas(self, notas):
        if len(notas) != 3:
            raise ValueError("Se requieren exactamente tres notas.")

class CalculadorPromedio:
    def calcular(self, notas):
        return sum(notas) / len(notas)

class RepositorioEstudiante:
    def __init__(self):
        self.estudiantes = []

    def guardar(self, estudiante):
        self.estudiantes.append(estudiante)
        print("Se guardó el estudiante:", estudiante)

class RegistroEstudiante:
    def __init__(self):
        self.validador = ValidadorEstudiante()
        self.calculador = CalculadorPromedio()
        self.repositorio = RepositorioEstudiante()

    def registrar(self, nombre, edad, notas):
        self.validador.validar_nombre(nombre)
        self.validador.validar_edad(edad)
        self.validador.validar_notas(notas)

        promedio = self.calculador.calcular(notas)
        estudiante = {
            "nombre": nombre,
            "edad": edad,
            "notas": notas,
            "promedio": promedio
        }

        self.repositorio.guardar(estudiante)
        return estudiante

registro = RegistroEstudiante()
registro.registrar("Luis", 22, [3.5, 4.0, 4.5])
```
