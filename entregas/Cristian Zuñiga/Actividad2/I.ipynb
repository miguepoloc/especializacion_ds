# Principio de Segregación de Interfaces (ISP)



## Caso problemático: mala práctica

Se crea una interfaz grande llamada `Empleado` con todos los métodos posibles.
Pero un analista no necesita procesar pagos, y un contador no necesita generar reportes complejos.

```python
class Empleado:
    def registrar_asistencia(self):
        pass

    def enviar_correo(self):
        pass

    def generar_reporte(self):
        pass

    def procesar_pago(self):
        pass


class Analista(Empleado):
    def registrar_asistencia(self):
        print("Analista registra asistencia")

    def enviar_correo(self):
        print("Analista envía correo")

    def generar_reporte(self):
        print("Analista genera reporte")

    def procesar_pago(self):
        raise NotImplementedError("El analista no procesa pagos")


class Contador(Empleado):
    def registrar_asistencia(self):
        print("Contador registra asistencia")

    def enviar_correo(self):
        print("Contador envía correo")

    def generar_reporte(self):
        print("Contador genera reporte")

    def procesar_pago(self):
        print("Contador procesa pago")
```

## Caso correcto: buena práctica

Se separan responsabilidades en interfaces pequeñas: asistencia, correos, reportes y pagos.
Cada clase implementa solo lo que necesita.

```python
class RegistroAsistencia:
    def registrar_asistencia(self):
        pass


class EnvioCorreo:
    def enviar_correo(self):
        pass


class GenerarReporte:
    def generar_reporte(self):
        pass


class ProcesarPago:
    def procesar_pago(self):
        pass


class Analista(RegistroAsistencia, EnvioCorreo, GenerarReporte):
    def registrar_asistencia(self):
        print("Analista registra asistencia")

    def enviar_correo(self):
        print("Analista envía correo")

    def generar_reporte(self):
        print("Analista genera reporte")


class Contador(RegistroAsistencia, ProcesarPago):
    def registrar_asistencia(self):
        print("Contador registra asistencia")

    def procesar_pago(self):
        print("Contador procesa pago")


analista = Analista()
contado = Contador()

analista.registrar_asistencia()
analista.generar_reporte()
contado.registrar_asistencia()
contado.procesar_pago()
```
