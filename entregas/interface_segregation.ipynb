### Correcto

```python
from abc import ABC, abstractmethod

class ProfesorITF(ABC):
    @abstractmethod
    def obtener_nombre(self) -> str:
        pass

    @abstractmethod
    def dictar_clase(self, curso: str) -> str:
        pass
class EstudianteITF(ABC):
    @abstractmethod
    def obtener_nombre(self) -> str:
        pass

    @abstractmethod
    def estudiar(self, curso: str) -> str:
        pass

class Profesor(ProfesorITF):
    def __init__(self, nombre: str, curso: str):
        self.nombre = nombre
        self.curso = curso

    def obtener_nombre(self) -> str:
        return self.nombre

    def dictar_clase(self, curso: str) -> str:
        return self.obtener_nombre() + " está dictando la clase de " + curso

class Estudiante(EstudianteITF):
    def __init__(self, nombre: str, cursos: list):
        self.nombre = nombre
        self.cursos = cursos

    def obtener_nombre(self) -> str:
        return self.nombre

    def estudiar(self, curso: str) -> str:
        if curso in self.cursos:
            return self.obtener_nombre() + " está estudiando " + curso
        else:
            raise Exception("El estudiante no está inscrito en el curso " + curso)

profesor = Profesor("Juan", "Matemáticas")
print(profesor.dictar_clase("Matemáticas"))

estudiante = Estudiante("Pedro", ["Matemáticas", "Física", "Química"])
try:
    print(estudiante.estudiar("Matemáticas"))
    print(estudiante.estudiar("Historia"))
except Exception as e:
    print(e)
```
