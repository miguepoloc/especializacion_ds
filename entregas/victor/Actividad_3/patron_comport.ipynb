### Forma errónea: Sin patrón Strategy

```python
class Compra:

    def __init__(self, total):
        self.total = total

    def calcular_total_final(self, tipo_descuento):

        if tipo_descuento == "estudiante":
            descuento = self.total * 0.10

        elif tipo_descuento == "vip":
            descuento = self.total * 0.20

        elif tipo_descuento == "temporada":
            descuento = self.total * 0.15

        else:
            descuento = 0

        return self.total - descuento


# Ejemplo de uso

compra = Compra(100000)

print("Total original:", compra.total)

print(
    "Total con descuento de estudiante:",
    compra.calcular_total_final("estudiante")
)

print(
    "Total con descuento VIP:",
    compra.calcular_total_final("vip")
)

print(
    "Total con descuento de temporada:",
    compra.calcular_total_final("temporada")
)
```

### Forma correcta: Con patrón Strategy

```python

class EstrategiaDescuento:


    def calcular_descuento(self, total):
        pass


# Estrategia 1
class DescuentoEstudiante(EstrategiaDescuento):

    def calcular_descuento(self, total):
        return total * 0.10


# Estrategia 2
class DescuentoVIP(EstrategiaDescuento):

    def calcular_descuento(self, total):
        return total * 0.20


# Estrategia 3
class DescuentoTemporada(EstrategiaDescuento):

    def calcular_descuento(self, total):
        return total * 0.15


# Contexto
class Compra:

    def __init__(self, total, estrategia):
        self.total = total
        self.estrategia = estrategia

    def calcular_total_final(self):

        descuento = self.estrategia.calcular_descuento(
            self.total
        )

        return self.total - descuento

    def cambiar_estrategia(self, nueva_estrategia):

        self.estrategia = nueva_estrategia


# EJEMPLO DE US

# Creamos una compra con descuento de estudiante
compra = Compra(
    100000,
    DescuentoEstudiante()
)

print("Total original:", compra.total)

print(
    "Total con descuento de estudiante:",
    compra.calcular_total_final()
)


# Cambiamos la estrategia a VIP
compra.cambiar_estrategia(
    DescuentoVIP()
)

print(
    "Total con descuento VIP:",
    compra.calcular_total_final()
)


# Cambiamos nuevamente la estrategia
compra.cambiar_estrategia(
    DescuentoTemporada()
)

print(
    "Total con descuento de temporada:",
    compra.calcular_total_final()
)
```
