## Patrón Creacional: Singleton
Este ejemplo demuestra claramente que el patrón Singleton garantiza que exista una sola configuración global, evitando que diferentes módulos de la aplicación trabajen con configuraciones inconsistentes.

### Forma incorrecta: sin Singleton

```python
class ConfiguracionEstacion:

    def __init__(self, nombre, ubicacion, unidad_temperatura):
        self.nombre = nombre
        self.ubicacion = ubicacion
        self.unidad_temperatura = unidad_temperatura

    def mostrar_configuracion(self):
        print(f"Estación: {self.nombre}")
        print(f"Ubicación: {self.ubicacion}")
        print(f"Temperatura: {self.unidad_temperatura}")


# Se crean dos configuraciones diferentes
config1 = ConfiguracionEstacion(
    "Estación Santa Marta",
    "Santa Marta",
    "Celsius"
)

config2 = ConfiguracionEstacion(
    "Estación Bogotá",
    "Bogotá",
    "Fahrenheit"
)

config1.mostrar_configuracion()

print("------------------")

config2.mostrar_configuracion()

print("------------------")

print(config1 is config2)
```

### Forma correcta: aplicando Singleton

```python
class ConfiguracionEstacion:

    _instancia = None

    def __new__(cls, nombre, ubicacion, unidad_temperatura):

        if cls._instancia is None:

            cls._instancia = super().__new__(cls)

            cls._instancia.nombre = nombre
            cls._instancia.ubicacion = ubicacion
            cls._instancia.unidad_temperatura = unidad_temperatura

        return cls._instancia


    def mostrar_configuracion(self):

        print("CONFIGURACIÓN DE LA ESTACIÓN")
        print(f"Estación: {self.nombre}")
        print(f"Ubicación: {self.ubicacion}")
        print(f"Unidad de temperatura: {self.unidad_temperatura}")


# Primera creación de la configuración
config1 = ConfiguracionEstacion(
    "Estación Meteorológica Santa Marta",
    "Santa Marta",
    "Celsius"
)

config1.mostrar_configuracion()

print("________________________________")

# Se intenta crear otra configuración
config2 = ConfiguracionEstacion(
    "Otra Estación",
    "Bogotá",
    "Fahrenheit"
)

config2.mostrar_configuracion()

print("________________________________")
```
