### Forma errónea:
Aquí el cliente tiene que conocer y utilizar cuatro clases diferentes. Esto aumenta la complejidad del código.

```python
class Inventario:

    def verificar_producto(self, producto):
        print(f"Verificando disponibilidad de {producto}")
        return True


class Pago:

    def realizar_pago(self, monto):
        print(f"Realizando pago de ${monto}")


class Factura:

    def generar_factura(self, producto, monto):
        print(f"Generando factura de {producto} por ${monto}")


class Envio:

    def programar_envio(self, producto):
        print(f"Programando envío de {producto}")


# Cliente

inventario = Inventario()
pago = Pago()
factura = Factura()
envio = Envio()

producto = "Computador"

if inventario.verificar_producto(producto):
    pago.realizar_pago(2500000)
    factura.generar_factura(producto, 2500000)
    envio.programar_envio(producto)
```

### Con patrón Facade (forma correcta)
Creamos una clase CompraFacade que se encarga de coordinar todos los subsistemas.

```python
class Inventario:

    def verificar_producto(self, producto):
        print(f"Verificando disponibilidad de {producto}")
        return True


class Pago:

    def realizar_pago(self, monto):
        print(f"Realizando pago de ${monto}")


class Factura:

    def generar_factura(self, producto, monto):
        print(f"Generando factura de {producto} por ${monto}")


class Envio:

    def programar_envio(self, producto):
        print(f"Programando envío de {producto}")


class CompraFacade:

    def __init__(self):
        self.inventario = Inventario()
        self.pago = Pago()
        self.factura = Factura()
        self.envio = Envio()

    def comprar(self, producto, monto):

        if self.inventario.verificar_producto(producto):
            self.pago.realizar_pago(monto)
            self.factura.generar_factura(producto, monto)
            self.envio.programar_envio(producto)

            print("Compra realizada correctamente")


# Cliente

compra = CompraFacade()

compra.comprar("Computador", 2500000)
```
