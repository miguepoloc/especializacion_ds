# Principio de Inversión de Dependencias

### Ejemplo correcto 
sistema de transporte para una aplicación de viajes.

```python

class Transporte:
    def viajar(self, destino):
        pass

class Taxi(Transporte):

    def viajar(self, destino):
        print(f"Viajando en taxi hacia {destino}")


class Motocicleta(Transporte):

    def viajar(self, destino):
        print(f"Viajando en motocicleta hacia {destino}")


class AplicacionViajes:

    def __init__(self, transporte: Transporte):
        self.transporte = transporte

    def realizar_viaje(self, destino):
        self.transporte.viajar(destino)


taxi = Taxi()
app = AplicacionViajes(taxi)
app.realizar_viaje("Centro de la ciudad")

moto = Motocicleta()
app = AplicacionViajes(moto)
app.realizar_viaje("Universidad")



```
