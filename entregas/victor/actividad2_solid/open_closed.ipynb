# Principio Abierto/Cerrado (Open/Closed Principle, OCP)



### Ejemplo Correcto

```python
class UsuarioEPS:

    def __init__(self, nombre: str, documento: int) -> None:
        self.nombre = nombre
        self.documento = documento

    def mostrar_informacion(self) -> str:
        return f"Usuario: {self.nombre} - Documento: {self.documento}"


class Afiliado(UsuarioEPS):

    def mostrar_informacion(self) -> str:
        return f"Afiliado: {self.nombre} - Documento: {self.documento}"


class Beneficiario(UsuarioEPS):

    def mostrar_informacion(self) -> str:
        return f"Beneficiario: {self.nombre} - Documento: {self.documento}"


usuarios: list[UsuarioEPS] = [
    Afiliado("Juliana Campo", "1221986613"),
    Beneficiario("Carlos Anaya", "57419305")
]

for usuario in usuarios:
    print(usuario.mostrar_informacion())
```
