## Principio de Sustitución de Liskov

#### ejemplo correcto 

```python
class Cuenta:
    def __init__(self, saldo: float):
        self.saldo = saldo

    def consultar_saldo(self) -> float:
        return self.saldo

    def retirar(self, cantidad: float) -> float:
        if cantidad <= self.saldo:
            self.saldo -= cantidad
            return self.saldo
        else:
            raise ValueError("Saldo insuficiente")


class CuentaAhorros(Cuenta):
    def retirar(self, cantidad: float) -> float:
        if cantidad <= self.saldo:
            self.saldo -= cantidad
            return self.saldo
        else:
            raise ValueError("Saldo insuficiente")


def realizar_retiro(cuenta: Cuenta):
    cuenta.retirar(100)
    print("Saldo restante:", cuenta.consultar_saldo())


cuenta = Cuenta(500)
ahorros = CuentaAhorros(500)

realizar_retiro(cuenta)
realizar_retiro(ahorros)
```

### ejemplo incorrecto

```python
class Cuenta:
    def __init__(self, saldo: float):
        self.saldo = saldo

    def retirar(self, cantidad: float) -> float:
        if cantidad <= self.saldo:
            self.saldo -= cantidad
            return self.saldo
        else:
            raise ValueError("Saldo insuficiente")


class CuentaBloqueada(Cuenta):
    def retirar(self, cantidad: float) -> float:
        raise Exception("Esta cuenta no permite retiros")


def realizar_retiro(cuenta: Cuenta):
    cuenta.retirar(100)
    print("Retiro realizado")


cuenta = Cuenta(500)
bloqueada = CuentaBloqueada(500)

realizar_retiro(cuenta)       # Funciona
realizar_retiro(bloqueada)    # Error
```
