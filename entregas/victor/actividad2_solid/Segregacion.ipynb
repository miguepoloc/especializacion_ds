# Segregación de Interfaces (ISP)


### ejemplo correcto


```python

class Encendible:


    def encender(self):
        pass


    def apagar(self):
        pass


class ControlVolumen:

    def subir_volumen(self):
        pass


class ControlCanales:

    def cambiar_canal(self):
        pass


class Televisor(Encendible, ControlVolumen, ControlCanales):

    def encender(self):
        print("El televisor está encendido.")

    def apagar(self):
        print("El televisor está apagado.")

    def subir_volumen(self):
        print("Subiendo el volumen del televisor.")

    def cambiar_canal(self):
        print("Cambiando el canal del televisor.")


class Bombillo(Encendible):

    def encender(self):
        print("El bombillo está encendido.")

    def apagar(self):
        print("El bombillo está apagado.")


class Parlante(Encendible, ControlVolumen):

    def encender(self):
        print("El parlante está encendido.")

    def apagar(self):
        print("El parlante está apagado.")

    def subir_volumen(self):
        print("Subiendo el volumen del parlante.")
```

```python
televisor = Televisor()
bombillo = Bombillo()
parlante = Parlante()

televisor.encender()
televisor.subir_volumen()
televisor.cambiar_canal()

bombillo.encender()
bombillo.apagar()

parlante.encender()
parlante.subir_volumen()
```
