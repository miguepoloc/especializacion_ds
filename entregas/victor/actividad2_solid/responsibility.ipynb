# principio de resposabilidad unica 


En la siguenite actividad aplicaremos el principio de responsabilidad unicá 

#### Análisis del Ejemplo correcto

```python
class HotelBooking:
    def __init__(self, hotel_name: str, room_type: str, check_in_date: str, check_out_date: str,num_guests: int ):
        self.hotel_name = hotel_name
        self.room_type = room_type
        self.check_in_date = check_in_date
        self.check_out_date = check_out_date
        self.num_guests = num_guests
```

```python
class BookingPrinter:
    def print_booking(self, booking: HotelBooking) -> None:
        print(
            f"Hotel: {booking.hotel_name}, "
            f"Tipo de habitación: {booking.room_type}, "
            f"Entrada: {booking.check_in_date}, "
            f"Salida: {booking.check_out_date}, "
            f"Huéspedes: {booking.num_guests}"
        )
```

```python
class BookingRepository:
    def save_booking(self, booking: HotelBooking) -> None:
        print("Reserva guardada correctamente.")
```

```python
booking = HotelBooking(
    hotel_name="Hotel Paraíso",
    room_type="Habitación Deluxe",
    check_in_date="28/08/2026",
    check_out_date="31/08/2026",
    num_guests=2
)

printer = BookingPrinter()
repository = BookingRepository()

printer.print_booking(booking)
repository.save_booking(booking)
```
