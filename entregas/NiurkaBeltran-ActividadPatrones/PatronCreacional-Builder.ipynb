# __BUILDER__

En este ejemplo se está creando un sistema solar personalizado, donde hay muchas opciones y todas estas tienen que ser rellenadas, así no todas apliquen (por ejemplo, un cliente que no quiere batería igual tenía que pasar `False` y `0` como relleno). Como solución implementé el patrón Builder, porque las opciones que no aplican pueden omitirse y toman un valor predeterminado, evitando que el usuario tenga que proporcionar valores de relleno como False o 0 

__¿Por qué se usó BUILDER y no otro patrón?__

Se descartó Factory Method porque su objetivo principal es delegar la creación de objetos, mientras que en este caso el problema consiste en configurar progresivamente un mismo objeto. Abstract Factory tampoco es necesario porque está orientado a crear familias de objetos relacionados. Finalmente, Singleton no aplica porque se necesitan crear diferentes sistemas solares con configuraciones independientes. Por estas razones, Builder es adecuado porque permite construir el sistema paso a paso y omitir las características que no sean necesarias.

# EJEMPLO SIN PATRÓN

````python
class SistemaSolar:
    def __init__(self, cantidad_paneles, tipo_panel, inversor, tiene_bateria, capacidad_bateria, tipo_instalacion):
        self.cantidad_paneles = cantidad_paneles
        self.tipo_panel = tipo_panel
        self.inversor = inversor
        self.tiene_bateria = tiene_bateria
        self.capacidad_bateria = capacidad_bateria
        self.tipo_instalacion = tipo_instalacion

    def mostrar(self):
        print(f"Sistema con {self.cantidad_paneles} paneles solares de tipo {self.tipo_panel}, inversor {self.inversor}, batería de {self.capacidad_bateria} kWh, instalado como {self.tipo_instalacion}")
````

````python
sistema = SistemaSolar(6, "monocristalino", "InversorOnGrid", True, 10, "techo")
sistema.mostrar()
````

acá se observa como toca rellenar cada opción asi sea que no se requiera, ademas como las caracteristicas van en un cierto orden.

````python
sistema_sin_bateria = SistemaSolar(4, "policristalino", "InversorOffGrid", False, 0, "suelo")
sistema_sin_bateria.mostrar()
````

# EJEMPLO CON EL PATRÓN BUILDER

````python
class SistemaSolar:
    def __init__(self, cantidad_paneles=None, tipo_panel=None, inversor=None, capacidad_bateria=None, tipo_instalacion=None):
        self.cantidad_paneles = cantidad_paneles
        self.tipo_panel = tipo_panel
        self.inversor = inversor
        self.capacidad_bateria = capacidad_bateria
        self.tipo_instalacion = tipo_instalacion

    def mostrar(self):
        tiene_bateria = "si" if self.capacidad_bateria > 0 else "no"
        print(f"Sistema con {self.cantidad_paneles} paneles solares de tipo {self.tipo_panel}, inversor {self.inversor}, batería de {self.capacidad_bateria} kWh, instalado como {self.tipo_instalacion}, tiene batería: {tiene_bateria}")

class SistemaSolarBuilder:
    def __init__(self):
        self.cantidad_paneles = None
        self.tipo_panel = None
        self.inversor = None
        self.capacidad_bateria = 0
        self.tipo_instalacion = None

    def con_paneles(self, cantidad, tipo):
        self.cantidad_paneles = cantidad
        self.tipo_panel = tipo
        return self

    def con_inversor(self, inversor):
        self.inversor = inversor
        return self

    def con_bateria(self, capacidad):
        self.capacidad_bateria = capacidad
        return self

    def con_instalacion(self, tipo):
        self.tipo_instalacion = tipo
        return self

    def build(self):
        return SistemaSolar(self.cantidad_paneles, self.tipo_panel, self.inversor, self.capacidad_bateria, self.tipo_instalacion)
````

````python
sistema_desordenado = (SistemaSolarBuilder().con_bateria(10).con_instalacion("techo").con_paneles(6, "monocristalino").con_inversor("InversorOnGrid").build())

sistema_desordenado.mostrar()

sistema_sin_bateria = (SistemaSolarBuilder().con_paneles(4, "policristalino").con_inversor("InversorOffGrid").con_instalacion("suelo").build())
sistema_sin_bateria.mostrar()
````
