# __FACADE__

Para este ejemplo se está haciendo un diagnóstico de un sistema solar. El problema real es que quien quiera generar un diagnóstico debe conocer las 3 clases (`SensorPanel`, `CalculadorEnergia`, `ReporteDiagnostico`), crearlas por separado y llamarlas en el orden correcto, por eso opte por facade, ya que permite ocultar esa coordinación detrás de un solo método. 

__¿Por qué se usó FACADE y no otro patrón?__

Se descartó Adapter porque no existe ninguna interfaz incompatible que necesite ser adaptada. También se descartó Decorator porque el objetivo no es agregar funcionalidades adicionales de manera dinámica, sino simplificar el acceso a un proceso compuesto por varias operaciones. Por estas razones, Facade es adecuado porque permite ocultar la complejidad de las clases internas y ofrecer una única interfaz para ejecutar el diagnóstico.

# EJEMPLO SIN PATRÓN 

````python
class SensorPanel:
    def leer_potencia(self):
        return 300


class CalculadorEnergia:
    def calcular_energia(self, potencia, horas):
        return potencia * horas * 0.20


class ReporteDiagnostico:
    def generar(self, energia):
        print(f"El sistema generó {energia} kWh de energía hoy.")
````

 acá la persona debe de conocer y coordinar las tres clases para poder ejecutar el código

````python
sensor = SensorPanel()
potencia = sensor.leer_potencia()
calculador = CalculadorEnergia()
energia = calculador.calcular_energia(potencia, 5)
reporte = ReporteDiagnostico()
reporte.generar(energia)
````

# EJEMPLO CON PATRÓN FACADE

````python
class SensorPanel:
    def leer_potencia(self):
        return 300


class CalculadorEnergia:
    def calcular_energia(self, potencia, horas):
        return potencia * horas * 0.20


class ReporteDiagnostico:
    def generar(self, energia):
        print(f"El sistema generó {energia} kWh de energía hoy.")


class DiagnosticoSolarFacade:

    def __init__(self):
        self.sensor = SensorPanel()
        self.calculador = CalculadorEnergia()
        self.reporte = ReporteDiagnostico()

    def ejecutar_diagnostico(self, horas):
        potencia = self.sensor.leer_potencia()
        energia = self.calculador.calcular_energia(potencia, horas)
        self.reporte.generar(energia)
````

````python
diagnostico = DiagnosticoSolarFacade()
diagnostico.ejecutar_diagnostico(5)
````
