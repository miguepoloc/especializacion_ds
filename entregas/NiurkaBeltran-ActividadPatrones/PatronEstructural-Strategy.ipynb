# __STRATEGY__

En este ejercicio se trabajó con el patrón Strategy porque el problema era un `if/elif` que crecía cada vez que aparecía un tipo de tarifa nuevo, obligando a modificar código ya existente como solución, se separó cada forma de calcular el ahorro en una estrategia independiente. De esta manera, `TarifaResidencial`, `TarifaComercial` y `TarifaIndustrial` implementan el mismo método `calcular()`, pero cada una utiliza una fórmula diferente. 

__¿Por qué se usó STRATEGY y no otro patrón?__

Descarté Factory Method porque el problema no consiste en decidir qué objeto crear, sino en seleccionar qué algoritmo utilizar para calcular el ahorro de una factura. También descarté State porque las tarifas no representan estados por los que pasa una factura durante su funcionamiento, sino diferentes algoritmos de cálculo que pueden seleccionarse según el tipo de tarifa. Por estas razones, Strategy es adecuado porque permite intercambiar el algoritmo de cálculo sin modificar la `clase Factura`.

# EJEMPLO SIN PATRÓN

````python
class Factura:
    def __init__(self, energia_generada, tipo_tarifa):
        self.energia_generada = energia_generada
        self.tipo_tarifa = tipo_tarifa

    def calcular_ahorro(self):
        if self.tipo_tarifa == "residencial":
            return self.energia_generada * 796
        elif self.tipo_tarifa == "comercial":
            return self.energia_generada * 650
        else:
            raise ValueError("Tipo de tarifa no válido")
````

````python
factura_residencial = Factura(energia_generada=50, tipo_tarifa="residencial")
factura_comercial = Factura(energia_generada=50, tipo_tarifa="comercial")

print("Ahorro residencial:", factura_residencial.calcular_ahorro())
print("Ahorro comercial:", factura_comercial.calcular_ahorro())

factura_industrial = Factura(energia_generada=50, tipo_tarifa="industrial")
factura_industrial.calcular_ahorro()
````

# EJEMPLO CON PATRÓN STRATEGY

````python
from abc import ABC, abstractmethod


class EstrategiaTarifa(ABC):
    @abstractmethod
    def calcular(self, energia_generada):
        pass


class TarifaResidencial(EstrategiaTarifa):
    def calcular(self, energia_generada):
        return energia_generada * 796


class TarifaComercial(EstrategiaTarifa):
    def calcular(self, energia_generada):
        return energia_generada * 650

class TarifaIndustrial(EstrategiaTarifa):
    def calcular(self, energia_generada):
        return energia_generada * 550

class Factura:
    def __init__(self, energia_generada, estrategia: EstrategiaTarifa):
        self.energia_generada = energia_generada
        self.estrategia = estrategia

    def calcular_ahorro(self):
        return self.estrategia.calcular(self.energia_generada)
````

````python
factura_residencial = Factura(energia_generada=50, estrategia=TarifaResidencial())
factura_comercial = Factura(energia_generada=50, estrategia=TarifaComercial())

print("Ahorro residencial:", factura_residencial.calcular_ahorro())
print("Ahorro comercial:", factura_comercial.calcular_ahorro())

factura_industrial = Factura(energia_generada=50, estrategia=TarifaIndustrial())
print("Ahorro industrial:", factura_industrial.calcular_ahorro())
````
