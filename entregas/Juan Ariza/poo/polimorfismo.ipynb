## Ejemplo polimorfismo

abstract class Lanzamiento {
    + titulo: str
    + artista: str
    + publicar(): str
    + describir(): str
}

class Sencillo {
    + publicar(): str
}

class Album {
    + publicar(): str
}

class VideoMusical {
    + publicar(): str
}

Lanzamiento <|-- Sencillo
Lanzamiento <|-- Album
Lanzamiento <|-- VideoMusical
@startuml

```python
from abc import ABC, abstractmethod

class Lanzamiento(ABC):
    def __init__(self, titulo: str, artista: str) -> None:
        self.titulo: str = titulo
        self.artista: str = artista

    @abstractmethod
    def publicar(self) -> str:
        pass

    def describir(self) -> str:
        return f"'{self.titulo}' de {self.artista}"


class Sencillo(Lanzamiento):
    def publicar(self) -> str:
        return "Distribuyendo track individual a plataformas de streaming (Spotify, Apple Music)..."


class Album(Lanzamiento):
    def publicar(self) -> str:
        return "Enviando máster completo de 12 tracks a distribución digital y enviando arte a prensa para vinilos..."


class VideoMusical(Lanzamiento):
    def publicar(self) -> str:
        return "Subiendo video en 4K a YouTube y programando estreno con chat en vivo..."


def ejecutar_lanzamiento(lanzamiento: Lanzamiento) -> None:
    print(f"Procesando el lanzamiento de {lanzamiento.describir()}:")
    print(lanzamiento.publicar())
    print()



sencillo: Sencillo = Sencillo(titulo="Luna llena", artista="Aura")
album: Album = Album(titulo="Horizontes", artista="Los Viajeros")
video: VideoMusical = VideoMusical(titulo="Luna llena (Video Oficial)", artista="Aura")


for lanzamiento in [sencillo, album, video]:
    ejecutar_lanzamiento(lanzamiento=lanzamiento)
```
