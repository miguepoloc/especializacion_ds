### Ejemplo de herencia
####  reutiliza codigo de la clase padre animal

```python
class Animal:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def hacer_sonido(self):
        print(f"{self.nombre} hace un sonido genérico.")

    def describir(self):
        print(f"{self.nombre} tiene {self.edad} años.")


class Perro(Animal):
    def __init__(self, nombre, edad, raza):
        super().__init__(nombre, edad)
        self.raza = raza

    def hacer_sonido(self):
        print(f"{self.nombre} dice: ¡Guau!")


class Gato(Animal):
    def hacer_sonido(self):
        print(f"{self.nombre} dice: ¡Miau!")


perro = Perro("Rex", 3, "Labrador")
gato = Gato("Michi", 2)

perro.describir()
perro.hacer_sonido()

gato.describir()
gato.hacer_sonido()
```
