

### Ejemplo Trabajador

#### Desde la empresa en la que labora
En este ejemplo, la clase `Trabajador` es una abstracción de un trabajador desde la perspectiva de una empresa que lo emplea. Observemos cómo se aplica la abstracción:

1. **Atributos relevantes**: Se han seleccionado solo los atributos más importantes para la simulación (nombre, cargo, salario, hora_entrada).

2. **Método simplificado**: El método `trabajar()` es una representación simplificada del trabajo que realiza.

3. **Representación en cadena**: El método `__str__()` proporciona una representación concisa del estado del trabajador.

4. **La encapsulacion** oculta una accion __quejarseDelJefe que solo se realiza internamente cuando trabaja

```python
class Trabajador:
    def __init__(self, nombre: str, cargo: str, salario: int, nit: int, hora_entrada: int) -> None:
        self.__nombre: int = nombre
        self.__cargo: int = cargo
        self.__salario: int = salario
        self.__nit: int = nit
        self.__hora_entrada= hora_entrada

    def trabajar(self) -> None:
        print(f"empezo a trabajar {self.__cargo}")
        self.__quejarseDelJefe()
        print(f"El trabajador termino de trabajar")

    def cobrar(self) -> None:
        print(f"El trabajador cobro {self.__salario}")

    def __quejarseDelJefe(self):
        print(f"Habla mal del jefe (lo llama sapo-perro)")

    def __str__(self) -> str:
        return f"{self.__nombre}, {self.__cargo}, {self.__salario}, {self.__nit}, {self.__hora_entrada}"
    def cambiarCargo(self, cargoNuevo) -> bool:
        print(f"realiza solicitud para cambiar cargo")
        self.__cargo = cargoNuevo
        return cargoNuevo != ""
```

```python
trabajador = Trabajador(nombre="Juan Villar", cargo="Supervisar", salario=5000, nit=68177374, hora_entrada=6)
print(trabajador)
trabajador.trabajar()
trabajador.cobrar()
print(trabajador.cambiarCargo("Gerente"))
```
