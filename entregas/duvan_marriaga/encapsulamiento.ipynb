### EJEMPLO DE ENCAPSULAMIENTO 

**se expone la funcion publica comer**
**se privatiza el acceso a la validacion de si tiene hambre**

```python
class Persona:
    def __init__(self, nombre: str, tieneHambre: bool) -> None:
        self.__nombre: str = nombre
        self.__tieneHambre: bool = tieneHambre
    def __tieneHambre(self):
        if self.__tieneHambre:
            return True
        else:
            return False
    def comer(self):
        if self.__tieneHambre:
            print(f" {self.__nombre} -> Comer")
        else:
            print(f" {self.__nombre} -> No Comer")
```

```python
per1 = Persona('juana', False)
per2 = Persona('juan', True)

per1.comer()
per2.comer()

```
