Crea tu propio Builder para construir un objeto complejo, como una pizza o un computador.

```python
class Pizza:
    def __init__(self, masa, queso , vegetales):
        self.masa = masa
        self.queso = queso
        self.vegetales = vegetales
    def mostrar(self):
        print(f"Pizza de {self.masa}, {self.queso}, {self.vegetales}")

pizza = Pizza("maiz", "salado", "piña")
pizza.mostrar()
```

```python
class Pizzabuilder:
    def __init__(self):
        self.masa = None
        self.queso = None
        self.vegetales = None

    def set_masa(self, masa):
        self.masa = masa
        return self

    def set_queso(self, queso):
        self.queso = queso
        return self

    def set_vegetales(self, vegetales):
        self.vegetales = vegetales
        return self

    def build(self):
        return Pizza(self.masa, self.queso, self.vegetales)

builder = Pizzabuilder()
pizza = builder.set_masa("maiz").set_queso("salado").set_vegetales("piña").build()
pizza.mostrar()

```

```python
class PizzaMother:
    def __init__(self, builder):
        self.builder = builder

    def contruir_club(self, masa):
        return(self.builder.set_masa("blanco")
               .set_queso(masa)
               .build())

    def construir_nigeriano(self):
        return (self.builder.set_masa("dura")
                .set_queso("dulce")
                .set_vegetales("espinaca")
                .build())

mother = PizzaMother(Pizzabuilder())
pizza = mother.construir_nigeriano()
pizza.mostrar()


```
