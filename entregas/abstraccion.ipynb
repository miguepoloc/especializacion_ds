```python

class Paperplane:
    def __init__(self, speed: int, altitude: int, roll_angle: int) -> None:
        self.speed = speed
        self.altitude = altitude
        self.roll_angle = roll_angle

    def fly(self) -> None:
        print(f"El avion de papel vuela a {self.speed} km/h")

    def __str__(self) -> str:
        return f"Avion de papel: velocidad={self.speed} km/h, altitud={self.altitude} m, angulo de alabeo={self.roll_angle} grados"
```

```python
Paper_plane= Paperplane(speed= 500, altitude= 2000, roll_angle= 20)
print(Paper_plane)
Paper_plane.fly()
```

```python
class Scooter:
    def __init__(self, speed: int, aceleration: int, capacity: int)-> None:
        self.speed = speed
        self.aceleration = aceleration
        self.capacity = capacity
    def acelerate(self) -> None:
        print(f"La scooter acelera a {self.speed} km/h")
    def __str__(self) -> str:
        return f"Scooter: velocidad={self.speed} km/h, aceleracion={self.aceleration} m/s^2, capacidad={self.capacity} kg"
```

```python
scoo_ter= Scooter(speed= 650, aceleration= 40, capacity= 1)
print(scoo_ter)
scoo_ter.acelerate()
```
