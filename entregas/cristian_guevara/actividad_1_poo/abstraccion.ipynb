Abstracción

Vehiculos de mensajería con abstracción desde el punto de vista de la empresa y desde el cliente


```python
class VehiculoMensajeriaEmpresa:
    def __init__(self, tipo, capacidad_carga, costo_operacion):
        self.tipo = tipo
        self.capacidad_carga = capacidad_carga
        self.costo_operacion = costo_operacion

    def informacion_empresa(self):
        return (
            f"Vehículo: {self.tipo}\n"
            f"Capacidad de carga: {self.capacidad_carga} kg\n"
            f"Costo de operación: ${self.costo_operacion}"
        )
```

```python
class VehiculoMensajeriaCliente:
    def __init__(self, costo_envio, tiempo_entrega):
        self.costo_envio = costo_envio
        self.tiempo_entrega = tiempo_entrega

    def informacion_cliente(self):
        return (
            f"Medio de transporte: {self.tipo}\n"
            f"Costo del envío: ${self.costo_envio}\n"
            f"Tiempo estimado: {self.tiempo_entrega} horas"
        )
```
