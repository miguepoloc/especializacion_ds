```python
class VehiculoMensajeria:

    def __init__(self, tipo, capacidad_carga, costo_operacion):
        self.__tipo = tipo
        self.__capacidad_carga = capacidad_carga
        self.__costo_operacion = costo_operacion

    # Getter
    def get_costo_operacion(self):
        return self.__costo_operacion

    # Setter
    def set_costo_operacion(self, nuevo_costo):
        if nuevo_costo > 0:
            self.__costo_operacion = nuevo_costo
        else:
            print("El costo debe ser mayor a 0")

    def informacion_empresa(self):
        return (
            f"Vehículo: {self.__tipo}\n"
            f"Capacidad de carga: {self.__capacidad_carga} kg\n"
            f"Costo de operación: ${self.__costo_operacion}"
        )
```

```python
class Carro(VehiculoMensajeria):

    def calcular_tiempo_entrega(self, distancia):
        # Velocidad promedio de 60 km/h
        return distancia / 60
```

```python
class Avion(VehiculoMensajeria):

    def calcular_tiempo_entrega(self, distancia):
        # Velocidad promedio de 800 km/h
        return distancia / 800
```

```python
class Barco(VehiculoMensajeria):

    def calcular_tiempo_entrega(self, distancia):
        # Velocidad promedio de 40 km/h
        return distancia / 40
```

```python
carro = Carro("Carro", 500, 100000)
avion = Avion("Avión", 5000, 5000000)
barco = Barco("Barco", 20000, 2000000)

print(carro.calcular_tiempo_entrega(600))
print(avion.calcular_tiempo_entrega(600))
print(barco.calcular_tiempo_entrega(600))
```
