**Polimorfismo**

```python
from abc import ABC, abstractmethod


class VehiculoMensajeriaEmpresa(ABC):

    def __init__(self, tipo, capacidad_carga, costo_operacion):
        self.__tipo = tipo
        self.__capacidad_carga = capacidad_carga
        self.__costo_operacion = costo_operacion

    @abstractmethod
    def calcular_costo_envio(self, distancia):
        pass


class Carro(VehiculoMensajeriaEmpresa):

    def calcular_costo_envio(self, distancia):
        return distancia * 500
```

```python
carro = Carro("Carro", 500, 100000)

print(carro.calcular_costo_envio(100))
```
