crear un factory method para una aplicacion de notificaciones (emails,sms,push)


```python
import abc
class Notifier(abc.ABC):
    @abc.abstractmethod
    def send(self, message: str) -> None:
        pass

class SMS(Notifier):
    def send(self, message: str) -> None:
        print(f"Enviando SMS: {message}")

class Email(Notifier):
    def send(self, message: str) -> None:
        print(f"Enviando email: {message}")
```

```python
import abc

class Notificador(abc.ABC):
    @abc.abstractmethod
    def crear_notificador(self) -> Notifier:
        pass

    def enviar_notificacion(self, message: str) -> None:
        notificador = self.crear_notificador()
        notificador.send(message)
        print("Notificación enviada.")

class EmailNotifier(Notificador):
    def crear_notificador(self) -> Notifier:
        print("Enviando email...")
        return Email()

class SMSNotifier(Notificador):
    def crear_notificador(self) -> Notifier:
        print("Enviando SMS...")
        return SMS()


notificacion = EmailNotifier()
notificacion.enviar_notificacion("Sergio hola")
```
