La abstracción permite representar el mismo objeto real según el punto de vista de quien lo utiliza. En este caso, el médico se modela de dos formas:

- **Desde el paciente:** interesan su nombre, especialidad, consultorio y la acción de recetar.
- **Desde la EPS que lo contrata:** interesan su identificación profesional, especialidad, tarifa, disponibilidad y la acción de validar su contratación.

Ambas clases representan al mismo médico, pero cada una expone únicamente la información necesaria para su contexto. Esta es la diferencia entre el objeto real y su abstracción en un sistema.

```plantuml
@startuml
left to right direction

class MedicoParaPaciente {
  - nombre: str
  - especialidad: str
  - consultorio: str
  + recetar(medicamento: str): str
}

class MedicoContratadoPorEPS {
  - identificacion_profesional: str
  - nombre: str
  - especialidad: str
  - tarifa_consulta: float
  - disponible: bool
  + validar_contratacion(): str
}

MedicoParaPaciente .. MedicoContratadoPorEPS : representan al mismo médico
@enduml

````python
from dataclasses import dataclass


@dataclass
class MedicoParaPaciente:
    __nombre: str
    __especialidad: str
    __consultorio: str

    def recetar(self, medicamento: str) -> str:
        print(f"El Dr. {self.__nombre} receta {medicamento}.")


@dataclass
class MedicoContratadoPorEPS:
    __identificacion_profesional: str
    __nombre: str
    __especialidad: str
    __tarifa_consulta: float
    __disponible: bool

    def validar_contratacion(self) -> str:
        estado = "disponible" if self.__disponible else "no disponible"
        print(f"Médico {self.__nombre}: {estado} para contratación.")
````

````python
medico_para_paciente = MedicoParaPaciente(
    "Laura Gómez",
    "Medicina interna",
    "Consultorio 204",
)

medico_para_paciente.recetar("Paracetamol")

medico_para_eps = MedicoContratadoPorEPS(
    "RM-12345",
    "Laura Gómez",
    "Medicina interna",
    85000.0,
    True,
)

medico_para_eps.validar_contratacion()
````

### Comparación de la abstracción

| Perspectiva del paciente | Perspectiva de la EPS |
| --- | --- |
| Nombre del médico | Nombre del médico |
| Especialidad | Especialidad |
| Consultorio | Identificación profesional |
| Recetar un medicamento | Tarifa de consulta |
|  | Disponibilidad |
|  | Validar contratación |

La relación punteada indica que no son dos médicos diferentes: son dos representaciones del mismo profesional. La clase de cada contexto omite los datos y comportamientos que no necesita conocer.
