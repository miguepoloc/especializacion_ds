El polimorfismo permite utilizar objetos de diferentes subclases mediante una misma interfaz. En este dominio, `Medico` y `Paciente` son tipos de `Persona` y ambos implementan el método `presentarse()` de manera diferente.

El código cliente solo conoce que trabaja con personas. No necesita preguntar si el objeto es un médico o un paciente; Python selecciona automáticamente la implementación correcta de `presentarse()`.

````python
from dataclasses import dataclass
from abc import ABC, abstractmethod

@dataclass
class Persona(ABC):
    nombre: str
    apellido: str
    edad: int

    def nombre_completo(self) -> str:
        return f"{self.nombre} {self.apellido}"

    @abstractmethod
    def presentarse(self) -> str:
        pass


class Medico(Persona):
    def __init__(
        self,
        nombre: str,
        apellido: str,
        edad: int,
        especialidad: str,
        consultorio: str,
    ) -> None:
        super().__init__(nombre, apellido, edad)
        self.especialidad = especialidad
        self.consultorio = consultorio

    def presentarse(self) -> str:
        return super().presentarse() + (
            f" Soy médico de {self.especialidad} y atiendo en "
            f"el consultorio {self.consultorio}."
        )


class Paciente(Persona):
    def __init__(
        self,
        nombre: str,
        apellido: str,
        edad: int,
        eps: str,
        estado: str = "activo",
    ) -> None:
        super().__init__(nombre, apellido, edad)
        self.eps = eps
        self.estado = estado

    def presentarse(self) -> str:
        return super().presentarse() + (
            f" Soy paciente de la EPS {self.eps} y mi estado es "
            f"{self.estado}."
        )


medico = Medico(
    "Laura",
    "Gómez",
    42,
    "Medicina interna",
    "204",
)
paciente = Paciente("Carlos", "Pérez", 35, "SaludTotal")

personas: list[Persona] = [medico, paciente]

for persona in personas:
    print(persona.presentarse())
````

## Evidencia del polimorfismo

- `Persona` define la interfaz común `presentarse()`.
- `Medico` y `Paciente` sobrescriben el mismo método con información propia.
- Ambas implementaciones reutilizan el comportamiento común mediante `super().presentarse()`.
- El cliente recorre `personas` y llama a `persona.presentarse()` de forma uniforme.
- No se utilizan condicionales por tipo para seleccionar el comportamiento; Python aplica el despacho dinámico.
