En el dominio del consultorio, `Medico` y `Paciente` son personas. Por eso se define la superclase `Persona`, que concentra los atributos y comportamientos comunes: nombre, apellido y edad.

- Un médico **es una** persona.
- Un paciente **es una** persona.

Las subclases reutilizan el constructor y el método `nombre_completo()` de `Persona` mediante `super()`. Cada una agrega únicamente sus datos y comportamientos particulares.

````python
from dataclasses import dataclass
from datetime import datetime


@dataclass
class Persona:
    nombre: str
    apellido: str
    edad: int

    def nombre_completo(self) -> str:
        return f"{self.nombre} {self.apellido}"

class Medico(Persona):
    def __init__(
        self,
        nombre: str,
        apellido: str,
        edad: int,
        especialidad: str,
        consultorio: str,
    ) -> None:
        super().__init__(nombre, apellido, edad)
        self.especialidad = especialidad
        self.consultorio = consultorio

    def recetar(self, medicamento: str) -> str:
        return f"El Dr. {self.nombre_completo()} receta {medicamento}."


class Paciente(Persona):
    def __init__(
        self,
        nombre: str,
        apellido: str,
        edad: int,
        eps: str,
        estado: str = "activo",
    ) -> None:
        super().__init__(nombre, apellido, edad)
        self.eps = eps
        self.estado = estado

    def solicitar_cita(self, medico: Medico, fecha: datetime) -> str:
        return (
            f"Cita solicitada con el Dr. {medico.nombre_completo()} "
            f"para el {fecha:%d/%m/%Y %H:%M}."
        )

# La cita conserva la relación entre un paciente y un médico.
@dataclass
class Cita:
    paciente: Paciente
    medico: Medico
    fecha: datetime


medico = Medico(
    "Laura",
    "Gómez",
    42,
    "Medicina interna",
    "204",
)
paciente = Paciente("Carlos", "Pérez", 35, "SaludTotal")
cita = Cita(paciente, medico, datetime(2026, 9, 15, 10, 30))

print(medico.presentarse())
print(paciente.presentarse())
print(medico.recetar("Acetaminofén"))
print(paciente.solicitar_cita(medico, cita.fecha))
````

## Evidencia de la herencia

- `Medico` y `Paciente` heredan los atributos `nombre`, `apellido` y `edad` de `Persona`.
- Ambas subclases reutilizan `nombre_completo()` sin duplicar su implementación.
- Sus constructores llaman a `super().__init__()` para inicializar la parte común.
- `presentarse()` se sobrescribe porque cada tipo de persona debe mostrar información diferente, pero conserva la presentación base llamando a `super().presentarse()`.
- `Cita` no hereda de `Persona`, porque una cita no es una persona; únicamente se relaciona con `Medico` y `Paciente`.
