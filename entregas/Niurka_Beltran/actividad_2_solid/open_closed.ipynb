# __OCP__

Para esta segunda actividad se estará ejecutando el principio de abierto/cerrado. El primer ejemplo representa la versión que viola el principio, porque la lista de tipos de panel está "quemada" dentro del método por lo que agregar un tipo de panel nuevo obliga a tocar código existente, en vez de simplemente añadir algo nuevo sin afectar lo anterior. Por lo tanto, si se desea agregar un nuevo tipo tocaria modificar dentro de __def calcular_energia_generada__ para agregar un __elif__

__VERSIÓN QUE VIOLA__

```python
class CalculadorEnergia:
    def __init__(self, tipo_panel, potencia):
        self.tipo_panel = tipo_panel
        self.potencia = potencia

    def calcular_energia_generada(self, horas_sol):
        if self.tipo_panel == "monocristalino":
            eficiencia = 0.20
        elif self.tipo_panel == "policristalino":
            eficiencia = 0.16
        else:
            raise ValueError("Tipo de panel no válido")

        return self.potencia * horas_sol * eficiencia
```

```python

panel_mono = CalculadorEnergia("monocristalino", 300)
panel_poli = CalculadorEnergia("policristalino", 300)

print("Energía panel monocristalino:", panel_mono.calcular_energia_generada(5))
print("Energía panel policristalino:", panel_poli.calcular_energia_generada(5))
```

__VERSIÓN CORREGIDA__

```python
from abc import ABC, abstractmethod


class Panel(ABC):
    def __init__(self, potencia):
        self.potencia = potencia

    @abstractmethod
    def eficiencia(self):
        pass

    def calcular_energia_generada(self, horas_sol):
        return self.potencia * horas_sol * self.eficiencia()


class PanelMonocristalino(Panel):
    def eficiencia(self):
        return 0.20


class PanelPolicristalino(Panel):
    def eficiencia(self):
        return 0.16

class PanelFlexible(Panel):
    def eficiencia(self):
        return 0.12

```

```python
panel_mono = PanelMonocristalino(300)
panel_poli = PanelPolicristalino(300)
panel_flexible = PanelFlexible(300)

print("Energía panel monocristalino:", panel_mono.calcular_energia_generada(5))
print("Energía panel policristalino:", panel_poli.calcular_energia_generada(5))
print("Energía panel flexible:", panel_flexible.calcular_energia_generada(5))
```

Con esta version se comprueba que para agregar un nuevo tipo de panel se añade sin tener que intervenir y romper el sistema que ya venia funcionando bien, se organizó por clase los tipos de paneles que se usan y si se necesita agregar otro se crearia una nueva clase.
