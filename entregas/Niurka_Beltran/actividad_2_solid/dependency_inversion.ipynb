En este ejercicio la clase de __FacturaSolar__ viola este principio, porque declara que metodo de cobro se va usar, por eso esta clase depende de la clase __PagoEfectivo__.

A pesar de que no se declaro que tipo de metodo se iba a usar, este tomo que el metodo era en efectivo.

__VERSIÓN QUE VIOLA__

```python
class PagoEfectivo:
    def procesar(self, monto):
        print(f"Procesando pago en efectivo de ${monto}")

class FacturaSolar:
    def __init__(self, monto_total):
        self.monto_total = monto_total
        self.metodo_pago = PagoEfectivo()

    def pagar(self):
        self.metodo_pago.procesar(self.monto_total)
```

```python

factura = FacturaSolar(monto_total=200000)
factura.pagar()
```

__VERSIÓN CORREGIDA__

En esta versión __FacturaSolar__ ya no viola el principio porque dejó de decidir por si mismo que metodo de pago se usaria. Ahora solo recibe la decision que se declare por fuera, y usando el método que se pida.

```python
from abc import ABC, abstractmethod

class MetodoPago(ABC):
    @abstractmethod
    def procesar(self, monto):
        pass

class PagoEfectivo(MetodoPago):
    def procesar(self, monto):
        print(f"Procesando pago en efectivo de ${monto}")

class PagoTransferencia(MetodoPago):
    def procesar(self, monto):
        print(f"Procesando pago por transferencia de ${monto}")

class FacturaSolar:
    def __init__(self, monto_total, metodo_pago: MetodoPago):
        self.monto_total = monto_total
        self.metodo_pago = metodo_pago

    def pagar(self):
        self.metodo_pago.procesar(self.monto_total)
```

```python
factura_efectivo = FacturaSolar(monto_total=200000, metodo_pago=PagoEfectivo())
factura_efectivo.pagar()

factura_transferencia = FacturaSolar(monto_total=200000, metodo_pago=PagoTransferencia())
factura_transferencia.pagar()
```
