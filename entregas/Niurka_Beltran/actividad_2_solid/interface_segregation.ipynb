# __ISP__

En este caso la clase __DipositivoSolar__ viola el principio porque esta mezclando dos responsabilidades distintas, que son generar y almacenar, obligando a un panel almacenar energia cuando esa no es su funcion, igual con bateria la obliga a generar cuando esta no lo hace.

En el ejemplo ded uso se coloca a prueba la parte del panel donde se le pide almacenar energia, al no ser esa su función lo que hace es pasar.

__VERSIÓN QUE VIOLA__

```python
from abc import ABC, abstractmethod


class DispositivoSolar(ABC):
    @abstractmethod
    def generar_energia(self):
        pass

    @abstractmethod
    def almacenar_energia(self, cantidad_kwh):
        pass


class PanelSolar(DispositivoSolar):
    def __init__(self, potencia):
        self.potencia = potencia

    def generar_energia(self):
        return self.potencia

    def almacenar_energia(self, cantidad_kwh):
        pass


class Bateria(DispositivoSolar):
    def __init__(self, nivel_actual):
        self.nivel_actual = nivel_actual

    def generar_energia(self):
        pass

    def almacenar_energia(self, cantidad_kwh):
        self.nivel_actual += cantidad_kwh
        return self.nivel_actual
```

```python
panel = PanelSolar(300)
panel.almacenar_energia(5)
```

__VERSIÓN CORREGIDA__

Acá se construyen dos interfaces, donde cada una cumpla dicha función y asi poder heradar para panel el generador y para la bateria el almacenamiento sin que se obligue a pasar.

```python
from abc import ABC, abstractmethod


class Generador(ABC):
    @abstractmethod
    def generar_energia(self):
        pass


class Almacenador(ABC):
    @abstractmethod
    def almacenar_energia(self, cantidad_kwh):
        pass


class PanelSolar(Generador):
    def __init__(self, potencia):
        self.potencia = potencia

    def generar_energia(self):
        return self.potencia


class Bateria(Almacenador):
    def __init__(self, nivel_actual):
        self.nivel_actual = nivel_actual

    def almacenar_energia(self, cantidad_kwh):
        self.nivel_actual += cantidad_kwh
        return self.nivel_actual
```

```python
panel = PanelSolar(300)
bateria = Bateria(nivel_actual=5)

print("Energía generada por el panel:", panel.generar_energia())
print("Nivel de la batería tras almacenar:", bateria.almacenar_energia(1))
```
