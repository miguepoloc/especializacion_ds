# __LISKOV__
En este ejemplo tenemos la clase padre __Bateria__, la cual se encarga de cargar, sumando una cantidad al nivel actual, devolviendo el resultado. Después, tenemos una subclase __BateriaDefectuosa(Bateria)__ que a pesar que identifique una bateria no podria reemplazar la clase padre porque en vez de sumar y devolver un resultado, ella detiene el sistema mostrando un error.

__VERSIÓN QUE VIOLA__


```python
class Bateria:
    def __init__(self, capacidad_kwh, nivel_actual):
        self.capacidad_kwh = capacidad_kwh
        self.nivel_actual = nivel_actual

    def cargar(self, cantidad_kwh):
        self.nivel_actual += cantidad_kwh
        return self.nivel_actual


class BateriaDefectuosa(Bateria):
    def cargar(self, cantidad_kwh):
        raise Exception("Esta batería no puede cargarse")
```

```python
def cargar_bateria(bateria):
    nivel = bateria.cargar(1)
    print(f"Nivel actual ({type(bateria).__name__}):", nivel)

bateria_normal = Bateria(capacidad_kwh=10, nivel_actual=5)
cargar_bateria(bateria_normal)

bateria_mala = BateriaDefectuosa(capacidad_kwh=10, nivel_actual=5)
cargar_bateria(bateria_mala)
```

__VERSIÓN CORREGIDA__

Al agregara la subclase __BateriaLitio__ se observa que esta cumple con la funcion de la clase padre, tiene una misma entrada y un mismo tipo de salida por lo que tendrian el mismo efecto, sumando y mostrando la cantidad de bateria actual despues de la operación.

```python
class Bateria:
    def __init__(self, capacidad_kwh, nivel_actual):
        self.capacidad_kwh = capacidad_kwh
        self.nivel_actual = nivel_actual

    def cargar(self, cantidad_kwh):
        self.nivel_actual += cantidad_kwh
        return self.nivel_actual


class BateriaLitio(Bateria):
    def cargar(self, cantidad_kwh):
        self.nivel_actual += cantidad_kwh
        return self.nivel_actual


def cargar_bateria(bateria):
    nivel = bateria.cargar(1)
    print(f"Nivel actual ({type(bateria).__name__}):", nivel)
```

```python
bateria_normal = Bateria(capacidad_kwh=10, nivel_actual=5)
bateria_litio = BateriaLitio(capacidad_kwh=10, nivel_actual=5)

cargar_bateria(bateria_normal)
cargar_bateria(bateria_litio)
```
