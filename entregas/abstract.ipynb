Crea tu propia Abstract Factory para una familia de productos tecnológicos (por ejemplo, Laptop y Smartphone de diferentes marcas).

```python
import abc

class Tecnologiafactory(abc.ABC):
    @abc.abstractmethod
    def crear_tecnologia(self):
        pass

class Laptopfactory(Tecnologiafactory):
    def crear_tecnologia(self):
        return "Laptop creada"

class Celularfactory(Tecnologiafactory):
    def crear_tecnologia(self):
        return "Celular creado"

factory = Laptopfactory()
laptop = factory.crear_tecnologia()
print(laptop)  # Output: Laptop creada

factory = Celularfactory()
celular = factory.crear_tecnologia()
print(celular)  # Output: Celular creado
```
