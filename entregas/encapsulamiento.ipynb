Encapsulamiento

```python
class Cajafuerte:
    def __init__(self, titular: str, saldo_inicial: float, contraseña: str = "1234") -> None:
        self.__titular = titular
        self.__saldo = saldo_inicial
        self.__contraseña = contraseña  # Contraseña personalizada
    def depositar(self, cantidad: float) -> None:
        if cantidad > 0:
            self.__saldo += cantidad
            return True
        return False
    def retirar(self, cantidad:float) -> bool:
        if 0 < cantidad <= self.__saldo:
            self.__saldo -= cantidad
            return True
        return False
    def obtener_saldo(self, contraseña: str) -> float:
        if contraseña == self.__contraseña:
            return self.__saldo
        return 0.0
    def obtener_titular(self) -> str:
        return self.__titular
```

```python
cuenta = Cajafuerte(titular="Sergio", saldo_inicial=10000, contraseña="contrasena123")
print(f"saldo inicial.: {cuenta.obtener_saldo('contrasena123')}")  # Output: 10000

cuenta.depositar(5000)
print(f"saldo despues de depositar 5000.: {cuenta.obtener_saldo('contrasena123')}")  # Output: 15000

cuenta.retirar(3000)
print(f"saldo despues de retirar 3000,: {cuenta.obtener_saldo('contrasena123')}")  # Output: 12000
```
