Crea tu propio Adapter para adaptar una clase de pagos de terceros a una interfaz de pagos esperada por tu sistema.

```python
class TerceroPagos:
    def pagar(self, monto):
        print(f"Pagando {monto} a un tercero.")

    def recibir_pago(self, monto):
        print(f"Recibiendo pago de {monto} de un tercero.")


class PagosAdapter:
    def __init__(self, adaptee):
        self.adaptee = adaptee

    def pagar(self, monto):
        self.adaptee.pagar(monto)

    def imprimirpago(self, monto):
        self.adaptee.recibir_pago(monto)


pagos = TerceroPagos()
adapter = PagosAdapter(pagos)
adapter.pagar(100)

```
