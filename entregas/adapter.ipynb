```python
class ReproductorMP3:
    def reproducir_mp3(self,archivo):
        print(f'Reproduciendo mp3: {archivo}')

class AdaptadorAudio:
    def __init__(self,reproductor):
        self.reproductor=reproductor
    def play(self,archivo):
        self.reproductor.reproducir_mp3(archivo)

def reproducir_audio(reproductor,archivo):
    reproductor.play(archivo)

Reproductor = ReproductorMP3()
adaptado=AdaptadorAudio(Reproductor)
reproducir_audio(adaptado,'cancion.mp3')
```
