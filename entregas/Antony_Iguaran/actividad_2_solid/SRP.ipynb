# __PRINCIPIO DE RESPONSABILIDAD UNICA__

Para esta primera parte, se realizará una clase con más de una responsabilidad y luego, se dividirá cada responsabilidad para que cada uno tenga una responsabilidad única.

Para esto, usaré un sistema de peaje.

## Código sin utilizar este principio.
Trabajaré con una clase que maneje la información del vehiculo, los datos del sistema de cobro, la lógica de validación del cobro el cual imprimirá una notificación.

```python
class Peaje:
    def __init__(self, placa, categoria_veh, id, saldo):
        self.placa = placa
        self.categoria_veh = categoria_veh
        self.id = id
        self.saldo = saldo

    def obtener_tarifa(self):
        tarifas = {"Categoria 1": 10500, "Categoria II":13500, "Categoria III": 18500, "Categoria IV": 23500}
        return tarifas.get(self.categoria_veh, 0)

    def validar_y_cobrar(self):
        tarifa = self.obtener_tarifa()
        if self.saldo >= tarifa:
            self.saldo -= tarifa
            print(f"Cobro exitoso a {self.placa}. Nuevo saldo: {self.saldo}")
            return True, tarifa
        else:
            print(f"Saldo insuficiente para {self.placa}. Tarifa: {tarifa}")
            return False, tarifa

#Ejemplo de uso
vehiculo_exitoso = Peaje("ABC123", "Categoria 1", "123456789", 20000)
vehiculo_sin_saldo = Peaje("XYZ789", "Categoria II", "987654321", 10000)

print("----Procesando información Vehiculo 1 ---")
vehiculo_exitoso.validar_y_cobrar()
print("----Procesando información Vehiculo 2 ---")
vehiculo_sin_saldo.validar_y_cobrar()
```

## Aplicando SRP

Haciendo la correción al código anterior, separamos los datos del cliente (tag/saldo) de la lógica transaccional del peaje.

```python
class CuentaPeaje:
    def __init__(self, iden, saldo_inicial):
        self.iden = iden
        self.saldo = saldo_inicial

    def descontar_saldo(self, monto):
        if self.tiene_saldo_suficiente(monto):
            self.saldo -= monto
            return True
        return False

    def tiene_saldo_suficiente(self, monto):
        return self.saldo >= monto

class EstacionPeaje:
    def __init__(self, nombre_estacion, ubicacion):
        self.nombre_estacion = nombre_estacion
        self.ubicacion = ubicacion

    def procesar_cobro(self, cuenta: CuentaPeaje, tarifa):
        if cuenta.descontar_saldo(tarifa):
            return self.generar_recibo(cuenta.iden, tarifa)
        return "Transacción rechazada"

    def generar_recibo(self, iden, monto):
        return f"[{self.nombre_estacion}] Cobro de {monto} al TAG {iden} exitoso."

#Ejemplo de uso
cuenta = CuentaPeaje(1001, 20000)
estacion = EstacionPeaje("Peaje Troncal", "Km 15")
print(estacion.procesar_cobro(cuenta, 12000))
```
