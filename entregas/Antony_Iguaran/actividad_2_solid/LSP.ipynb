# __Principio de Sustitución de Liskov__

No se debe romper el contrato o comportamiento esperado de su clase padre, en esto se aplicará no a un sistema de peaje, sino a la reparación de un vehiculo

## Código sin utilizar este principio.
Una bicicleta va a heredar de vehiculo, pero al momento de modificarla lanzará un error en un método que el código espera que funcione, porque la bicicleta no es un automotor, por ende, haremos que tenga un fallo en el método del cambio de aceite.

```python
class VehiculoTaller:
    def __init__(self, vin, marca):
        self.vin = vin
        self.marca = marca

    def registrar_ingreso(self):
        return f"Ingreso de {self.marca}"

    def realizar_cambio_aceite(self):
        return "Aceite cambiado correctamente."

class BicicletaTaller(VehiculoTaller):
    def __init__(self, vin, marca, tipo_frenos):
        super().__init__(vin, marca)
        self.tipo_frenos = tipo_frenos

    def registrar_ingreso(self):
        return f"Ingreso de Bicicleta {self.marca}"

    def realizar_cambio_aceite(self):
        raise Exception("¡Las bicicletas no usan aceite de motor!")


# 1. Probando el vehículo base como un automor
auto = VehiculoTaller("VIN123", "Toyota")
print(auto.registrar_ingreso())
print(auto.realizar_cambio_aceite())

print("-" * 30)

# 2. Probando la bicicleta
bici = BicicletaTaller("VIN987", "Trek", "Disco")
print(bici.registrar_ingreso())

    # 3. Demostración de la violación del LSP
print("Intentando cambiar aceite a la bicicleta...")

print(bici.realizar_cambio_aceite())
```

## Utilizando el principio actual.

Tenemos la clase padre, luego de ahí se crea una clase especifica para cuando el vehiculo es motorizado o cuando es una bicicleta.

Si intento ejecutar bici.realizar_cambio_aceite(), Python lanzará un AttributeError, evitando que rompa el principio.

```python
class VehiculoTaller:
    def __init__(self, vin, marca):
        self.vin = vin
        self.marca = marca

    def registrar_ingreso(self):
        return f"Ingreso de {self.marca}"

class VehiculoMotorizado(VehiculoTaller):
    def realizar_cambio_aceite(self):
        return "Aceite cambiado correctamente."

class AutoTaller(VehiculoMotorizado):
    def registrar_ingreso(self):
        return f"Ingreso de Auto {self.marca}"

class BicicletaTaller(VehiculoTaller):
    def __init__(self, vin, marca, tipo_frenos):
        super().__init__(vin, marca)
        self.tipo_frenos = tipo_frenos

    def registrar_ingreso(self):
        return f"Ingreso de Bicicleta {self.marca}"

    def ajustar_cadena(self):
        return "Cadena ajustada."

# Bloque de prueba para ejecutar en la terminal

auto = AutoTaller("VIN123", "Toyota")
bici = BicicletaTaller("VIN456", "Trek", "Disco")

print("--- Probando el Auto ---")
print(auto.registrar_ingreso())
print(auto.realizar_cambio_aceite())

print("\n--- Probando la Bicicleta ---")
print(bici.registrar_ingreso())
print(bici.ajustar_cadena())

```
