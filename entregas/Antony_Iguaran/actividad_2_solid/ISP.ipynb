# __Principio de segregación de interfaces__

Para este principio, no se obliga a los usuarios a depender de interfaces que no utilizan, por esto se delimita el contexto para separar las funciones, y así tener beneficios como el mantenimiento de cada interfaz.

## Código sin aplicar este principio.

## Código aplicando el principio actual

Apartamos las interfaces según el contexto específico que se consumirá. El método diagnosticar_motor() ya no existe en esta clase y nos libramos del código basura y del NotImplementedError.

```python
from abc import ABC, abstractmethod

# 1. Separamos las responsabilidades en interfaces distintas
class IPeaje(ABC):
    @abstractmethod
    def pagar_telepeaje(self, monto):
        pass

class IDiagnostico(ABC):
    @abstractmethod
    def diagnosticar_motor(self):
        pass

# 2. Clase base para mantener los datos del vehículo
class VehiculoBase:
    def __init__(self, placa, modelo):
        self.placa = placa
        self.modelo = modelo

# 3. El cliente ahora hereda la clase base y SOLO la interfaz que le interesa
class ClienteSoloPeaje(VehiculoBase, IPeaje):
    def pagar_telepeaje(self, monto):
        return f"Vehículo placa {self.placa} pagó {monto} en el telepeaje."

class ClienteTaller(VehiculoBase, IDiagnostico):
    def diagnosticar_motor(self):
        return f"Vehículo placa {self.placa} está siendo diagnosticado por nuestro equipo."
# Parte de la ejecución

vehiculo_peaje = ClienteSoloPeaje("SMR-456", 2024)
vehiculo_taller= ClienteTaller("TOYOTA TXL", 2025)

#Ejecutando el código de prueba para el cliente que solo necesita pagar peaje y el cliente que necesita diagnóstico
print("--- Transacción ---")
resultado = vehiculo_peaje.pagar_telepeaje("$12.000")
print(resultado)

#Ejecutando el código de prueba para el cliente que necesita diagnóstico
print("\n--- Diagnóstico ---")
resultado_diagnostico = vehiculo_taller.diagnosticar_motor()
print(resultado_diagnostico)
```
