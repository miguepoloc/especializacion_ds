# __Principio Abierto / Cerrado__

Para esta segunda parte, si deseo agregar una categoria nueva de vehiculo, no debo modificar el código principal de la calculadora de tarifas.

## Código sin utilizar este principio
Si necesito agregar una nueva actualización al diccionario que tengo o agregar un condicional "if" dentro del método existente.

Por ejemplo, deseo agregar una moto o vehículo eléctrico.


```python
class CalculadoraTarifasPeaje:
    def __init__(self, tarifa_base, temporada):
        self.tarifa_base = tarifa_base
        self.temporada = temporada

    def calcular_por_categoria(self, categoria):
        # Rompe OCP: Modificar este método por cada nueva categoría
        if categoria == "I":
            return self.tarifa_base
        elif categoria == "II":
            return self.tarifa_base * 1.5
        elif categoria == "especial":
            return self.tarifa_base * 0.8
        return 0

    def actualizar_temporada(self, nueva_temporada):
        self.temporada = nueva_temporada



mi_peaje = CalculadoraTarifasPeaje(tarifa_base=10000, temporada="Alta")

# 2. Calcular e imprimir la tarifa para la Categoría I
tarifa_cat1 = mi_peaje.calcular_por_categoria("I")
print(f"Tarifa Categoría I: ${tarifa_cat1}")

# 3. Calcular e imprimir la tarifa para la Categoría II
tarifa_cat2 = mi_peaje.calcular_por_categoria("II")
print(f"Tarifa Categoría II: ${tarifa_cat2}")

# 4. Calcular e imprimir la tarifa para la Categoría especial
tarifa_cat_especial = mi_peaje.calcular_por_categoria("especial")
print(f"Tarifa Categoría Especial: ${tarifa_cat_especial}")

# 5. Actualizar la temporada, esto es opcional y es algo agregado para mostrar que se puede cambiar la temporada sin afectar el cálculo de tarifas.
mi_peaje.actualizar_temporada("Baja")
print(f"La temporada actual ahora es: {mi_peaje.temporada}")
```

## Empleando el principio O/C
Utilizamos un patrón visto en la clase pasada que es el polimorfismo. Se puede extender el sistema creando nuevas clases sin tocar el procesador principal.

```python
from abc import ABC, abstractmethod

class CategoriaVehiculo(ABC):
    def __init__(self, codigo, descripcion):
        self.codigo = codigo
        self.descripcion = descripcion

    @abstractmethod
    def calcular_tarifa(self, tarifa_base):
        pass

    def obtener_info(self):
        return f"Categoría: {self.descripcion} ({self.codigo})"

class CategoriaI(CategoriaVehiculo):
    def __init__(self, codigo, descripcion, recargo_ambiental=0):
        super().__init__(codigo, descripcion)
        self.recargo_ambiental = recargo_ambiental

    def calcular_tarifa(self, tarifa_base):
        return tarifa_base + self.recargo_ambiental

class CategoriaII(CategoriaVehiculo):
    def __init__(self, codigo, descripcion, ejes_extra=1):
        super().__init__(codigo, descripcion)
        self.ejes_extra = ejes_extra

    def calcular_tarifa(self, tarifa_base):
        return tarifa_base * (1 + (0.2 * self.ejes_extra))

class CategoriaEspecial(CategoriaVehiculo):
    def __init__(self, codigo, descripcion, descuento=0.2):
        super().__init__(codigo, descripcion)
        self.descuento = descuento

    def calcular_tarifa(self, tarifa_base):
        return tarifa_base * (1 - self.descuento)

# Ejecución
auto = CategoriaI("I", "Automóvil Estándar", recargo_ambiental=500)
camion = CategoriaII("II", "Camión Pequeño", ejes_extra=2)
especial = CategoriaEspecial("Especial", "Moto", descuento=0.2)


print(auto.obtener_info(), "Paga:", auto.calcular_tarifa(10500))
print(camion.obtener_info(), "Paga:", camion.calcular_tarifa(13500))
print(especial.obtener_info(), "Paga:", especial.calcular_tarifa(10500))
```
