# Primer contexto.

Para este ejercicio, escogeré un objeto como a un automóvil y partire de dos contextos diferentes el cual se establecerán en dos clases.

El primer contexto será al automóvil como un sistema de cobro, es decir, un peaje. 
Para esto estableceré los siguientes atributos:
   -Placa
   -Categoria del automóvil
   -ID del tag del telepeaje.
Los anteriores atributos se tendrán en cuenta para realizar si debe pagar y cuanto estaria pagando aproximadamente.

@startuml
class Peaje {
    + Placa: str
    + Categoria del vehiculo: str
    + iden: int
    + saldo: float
    + obtener_tarifa()
    + validar_saldo()
}
@enduml

```python
class Peaje:

    def __init__(self, placa, categoria_vehiculo, iden, saldo):
        self.placa = placa
        self.categoria_vehiculo = categoria_vehiculo
        self.iden = iden
        self.saldo = saldo

    def obtener_tarifa(self):
        tarifas = {
            "I": 12000,       # Automóvil
            "II": 18000,      # Camioneta o vehículo pequeño con remolque
            "III": 25000,     # Bus o camión
            "IV": 35000       # Vehículo de mayor tamaño
        }

        return tarifas.get(self.categoria_vehiculo.upper(), 0)

    def validar_saldo(self):
        tarifa = self.obtener_tarifa()

        if tarifa == 0:
            print("Categoría de vehículo no válida.")
            return False

        if self.saldo >= tarifa:
            self.saldo -= tarifa

            print("Pago realizado correctamente.")
            print(f"Placa: {self.placa}")
            print(f"ID del tag: {self.iden}")
            print(f"Valor pagado: ${tarifa}")
            print(f"Saldo restante: ${self.saldo}")

            return True

        else:
            print("Saldo insuficiente para realizar el pago.")
            print(f"Saldo disponible: ${self.saldo}")
            print(f"Valor del peaje: ${tarifa}")

            return False


vehiculo1 = Peaje(
    placa="ABC123",
    categoria_vehiculo="I",
    iden=1001,
    saldo=20000
)

# Validar si puede pagar el peaje
vehiculo1.validar_saldo()
```

# Segundo contexto.

Ahora, se utilizará el mismo objeto pero en otro contexto, que será la clase de reparación.

Para esto, se proponen los siguientes atributos y metodos.

@startuml
class MaquinaAReparar {
    + vin: str
    + marca: str
    + kilometraje: int
    + diagnostico: float
    + registrar_diagnostico()
    + verificar_diagnostico()
}
@enduml

```python
class MaquinaAReparar:

    def __init__(self, vin, marca, kilometraje, diagnostico):
        self.vin = vin
        self.marca = marca
        self.kilometraje = kilometraje
        self.diagnostico = diagnostico

    def registrar_diagnostico(self):
        print(f"Diagnóstico del vehículo: {self.diagnostico}")

    def verificar_mantenimiento(self):
        if self.kilometraje >= 10000:
            print("El vehículo requiere revisión de mantenimiento.")
        else:
            print("El vehículo aún no requiere mantenimiento.")

automovil = MaquinaAReparar(
    vin="9BWZZZ377VT004251",
    marca="Toyota",
    kilometraje=25000,
    diagnostico="Desgaste en las pastillas de freno"
)

automovil.registrar_diagnostico()
automovil.verificar_mantenimiento()
```
