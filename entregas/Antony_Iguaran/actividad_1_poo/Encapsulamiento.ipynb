# Una incubadora inteligente.

El software de una incubadora requiere un control preciso de su temperatura. Por seguridad, no se debe dar acceso directo a este parámetro, evitando así que otras partes del programa asignen valores peligrosamente altos o bajos.

A continuación, describiré los atributos y los metodos:

 @startuml
        class IncubadoraInteligente {
            + especie: str
            + __temperatura: str
            + temperatura: float 
            + mostrar_estado():
        }
        @enduml

```python
class IncubadoraInteligente:

    def __init__(self, especie, temperatura):
        self.especie = especie
        self._temperatura = 0
        self.temperatura = temperatura


    @property
    def temperatura(self):
        return self._temperatura

    @temperatura.setter
    def temperatura(self, nueva_temperatura):
        if 35 <= nueva_temperatura <= 40:
            self._temperatura = nueva_temperatura
        else:
            print("Error: la temperatura debe estar entre 35 °C y 40 °C.")

    def mostrar_estado(self):
        print(f"Especie: {self.especie}")
        print(f"Temperatura actual: {self._temperatura} °C")


incubadora1 = IncubadoraInteligente("Gallina", 37.5)

print(f"Temperatura: {incubadora1.temperatura} °C")

incubadora1.temperatura = 38

incubadora1.temperatura = 50 #Intentando establecer una temperatura fuera del rango permitido

incubadora1.mostrar_estado()
```
