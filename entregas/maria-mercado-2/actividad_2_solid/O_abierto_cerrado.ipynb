# O - Open/Closed (Abierto/Cerrado)
Ejemplo: cálculo de daño con múltiples tipos de arma. Violación: if/elif en clase central; Corrección: usar estrategias extensibles.


```python
# --- Viola O/C: DamageCalculator con if/elif ---
class DamageCalculatorViolation:
    def __init__(self, player_level: int):
        self.player_level = player_level
        self.base = 5
    def compute(self, weapon_type: str, power: int) -> int:
        # cada vez que agregas un arma hay que modificar este método
        if weapon_type == 'sword':
            return self.base + power + self.player_level
        elif weapon_type == 'bow':
            return int((self.base + power) * 0.8) + self.player_level
        elif weapon_type == 'magic':
            return self.base + power * 2 + self.player_level
        else:
            raise ValueError('Unknown weapon')

dcv = DamageCalculatorViolation(3)
print(dcv.compute('sword', 10))
# Imagínate ahora añadir 'laser' -> hay que editar compute -> viola O/C
```

```python
# --- Versión corregida: usar clases de estrategia y registro dinámico ---
from typing import Protocol, Dict

class DamageStrategy(Protocol):
    def compute(self, power: int, level: int) -> int:
        ...

class SwordStrategy:
    def __init__(self, base=5):
        self.base = base
    def compute(self, power: int, level: int) -> int:
        return self.base + power + level

class BowStrategy:
    def __init__(self, base=5):
        self.base = base
    def compute(self, power: int, level: int) -> int:
        return int((self.base + power) * 0.8) + level

class MagicStrategy:
    def __init__(self, base=5):
        self.base = base
    def compute(self, power: int, level: int) -> int:
        return self.base + power * 2 + level

class DamageCalculatorFixed:
    def __init__(self, level: int):
        self.level = level
        self.strategies: Dict[str, DamageStrategy] = {}
    def register(self, name: str, strat: DamageStrategy):
        self.strategies[name] = strat
    def compute(self, weapon_type: str, power: int) -> int:
        strat = self.strategies.get(weapon_type)
        if strat is None:
            raise ValueError('Unknown weapon')
        return strat.compute(power, self.level)

# uso y extensión sin modificar DamageCalculatorFixed
dcf = DamageCalculatorFixed(4)
dcf.register('sword', SwordStrategy())
dcf.register('bow', BowStrategy())
dcf.register('magic', MagicStrategy())
print(dcf.compute('magic', 7))
# ahora si agrego 'laser' solo creo LaserStrategy y lo registro externamente
```
