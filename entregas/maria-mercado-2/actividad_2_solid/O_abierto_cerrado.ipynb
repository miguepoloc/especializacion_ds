# O - Open/Closed (Abierto/Cerrado)


```python
from abc import ABC, abstractmethod

class Jugador(ABC):
    def __init__(self, name: str, level: int = 1):
        self.name = name
        self.level = level
        self.status = 'active'

    @abstractmethod
    def calcular_recompensa(self):
        pass

    def descripcion(self):
        return f"{self.name} (lvl {self.level})"

class JugadorNormal(Jugador):
    def __init__(self, name='Normal', level=1):
        super().__init__(name, level)
        self.rank = 'bronze'
        self.bonus = 0

    def calcular_recompensa(self):
        return 100

    def aplicar_bonus(self, amount):
        self.bonus += amount
        return self.bonus

class JugadorVIP(Jugador):
    def __init__(self, name='VIP', level=2):
        super().__init__(name, level)
        self.vip_tier = 1
        self.extra = 50

    def calcular_recompensa(self):
        return 300

    def subir_tier(self):
        self.vip_tier += 1
        self.extra += 20
        return self.vip_tier

class JugadorStreamer(Jugador):
    def __init__(self, name='Streamer', level=1, followers=1000):
        super().__init__(name, level)
        self.followers = followers
        self.stream_hours = 0

    def calcular_recompensa(self):
        return 500

    def agregar_stream(self, hours):
        self.stream_hours += hours
        return self.stream_hours

class JugadorEsports(Jugador):
    def __init__(self, name='Esports', level=5, team='TeamA'):
        super().__init__(name, level)
        self.team = team
        self.ranking = 100

    def calcular_recompensa(self):
        return 1000

    def mejorar_ranking(self, delta):
        self.ranking = max(1, self.ranking - delta)
        return self.ranking

# Uso
jugadores = [JugadorNormal(), JugadorVIP(), JugadorStreamer(), JugadorEsports()]
for j in jugadores:
    print(f"Recompensa: {j.calcular_recompensa()} - {j.descripcion()}")
```
