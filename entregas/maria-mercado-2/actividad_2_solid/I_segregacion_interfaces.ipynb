# I - Interface Segregation (Segregación de Interfaces)
Ejemplo: un interfaz grande que obliga a implementaciones con métodos innecesarios. Se muestra la violación y la corrección separando interfaces.


```
# --- Viola IS: interfaz 'Device' muy gorda y clases que no necesitan todo ---
class DeviceViolation:
    def __init__(self, name: str, version: str):
        self.name = name
        self.version = version
    def render(self, data):
        print('rendering', data)
    def save(self, obj):
        print('saving', obj)
    def send_network(self, packet):
        print('sending', packet)

class LocalRenderer(DeviceViolation):
    def __init__(self, name, version):
        super().__init__(name, version)
    def render(self, data):
        print('[LocalRenderer] show', data)
    # No necesita send_network ni save pero los hereda/implementa vacíos o con stubs -> mala práctica

lr = LocalRenderer('Local', '1.0')
lr.render({'hp': 10})
```

```
# --- Corrección: segregar interfaces (protocolos) y sólo implementar lo necesario ---
from typing import Protocol

class Renderer(Protocol):
    def render(self, data): ...
    def set_style(self, style: str): ...

class Saver(Protocol):
    def save(self, obj): ...
    def load(self, key): ...

class NetworkSender(Protocol):
    def send(self, packet): ...
    def connect(self, addr): ...

class LocalRendererFixed:
    def __init__(self, name: str):
        self.name = name
        self.style = 'plain'
    def render(self, data):
        print(f'[LocalRendererFixed] {self.name} ->', data)
    def set_style(self, style: str):
        self.style = style

class CloudSaver:
    def __init__(self, endpoint: str):
        self.endpoint = endpoint
        self.last = None
    def save(self, obj):
        self.last = obj
        return True
    def load(self, key):
        return {'dummy': True}

lr2 = LocalRendererFixed('Viewport')
lr2.render({'score': 100})
s = CloudSaver('https://api')
s.save({'player':'Hero'})
```
