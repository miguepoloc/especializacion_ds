# I - Interface Segregation Principle (ISP) (Segregación de Interfaces)
Contexto

Dentro de la plataforma existen usuarios normales, streamers y organizadores de torneos. No todos necesitan las mismas funcionalidades.


Antes de aplicar ISP

```python
class UsuarioGaming:

    def jugar(self):
        pass

    def transmitir_stream(self):
        pass

    def crear_torneo(self):
        pass
```

¿Por qué viola ISP?

```python
# APLICANDO ISP
from abc import ABC, abstractmethod

class Jugable(ABC):
    @abstractmethod
    def jugar(self):
        pass

class Streamer(ABC):
    @abstractmethod
    def transmitir_stream(self):
        pass

class OrganizadorTorneos(ABC):
    @abstractmethod
    def crear_torneo(self):
        pass

class JugadorBasico(Jugable):
    def __init__(self, username: str = 'basico', nivel: int = 1):
        self.username = username
        self.nivel = nivel
        self.progreso = {}

    def jugar(self):
        print('Jugador básico jugando')

    def guardar_progreso(self):
        self.progreso['last_play'] = 'now'
        print('Progreso guardado')

class UsuarioStreamer(Jugable, Streamer):
    def __init__(self, username: str = 'streamer', followers: int = 1000):
        self.username = username
        self.followers = followers
        self.stream_settings = {'quality': '1080p'}

    def jugar(self):
        print('Streamer jugando')

    def transmitir_stream(self):
        print('Transmitiendo partida en Twitch')

    def configurar_stream(self, quality: str):
        self.stream_settings['quality'] = quality
        print(f'Stream configurado a {quality}')

class ModeradorTorneos(Jugable, OrganizadorTorneos):
    def __init__(self, username: str = 'moderador', organizados: int = 0):
        self.username = username
        self.organizados = organizados
        self.permissions = ['manage']

    def jugar(self):
        print('Moderador jugando')

    def crear_torneo(self):
        self.organizados += 1
        print('Creando torneo internacional')

# Uso de ejemplo
basico = JugadorBasico()
streamer = UsuarioStreamer()
moderador = ModeradorTorneos()
basico.jugar()
basico.guardar_progreso()
streamer.jugar()
streamer.transmitir_stream()
streamer.configurar_stream('720p')
moderador.jugar()
moderador.crear_torneo()
```
