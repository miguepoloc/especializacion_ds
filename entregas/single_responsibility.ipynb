```python
class pasajero:
    def __init__(self, nombre_pasajero: str, edad_pasajero: int) -> None:
        self.nombre_pasajero: str = nombre_pasajero
        self.edad_pasajero: int = edad_pasajero
    def mostrar_info(self) -> None:
        print(f"Nombre del pasajero: {self.nombre_pasajero}, Edad: {self.edad_pasajero}")

class Avion:
    def __init__(self, nombre_avion: str, max_pasajeros_avion: int, min_longitud_pista: float) -> None:
        self.nombre_avion: str = nombre_avion
        self.max_pasajeros_avion: int = max_pasajeros_avion
        self.min_longitud_pista: float = min_longitud_pista
        self.pasajeros: list = []
    def agregar_pasajero(self, pasajero: str) -> None:
        if len(self.pasajeros) < self.max_pasajeros_avion:
            self.pasajeros.append(pasajero)
            print(f"Pasajero {pasajero} agregado al avión {self.nombre_avion}.")
        else:
            print(f"No se puede agregar al pasajero {pasajero}. El avión {self.nombre_avion} está lleno.")
class Pista:
    def __init__(self, numero_pista: int, longitud_pista: float) -> None:
        self.numero_pista: int = numero_pista
        self.longitud_pista: float = longitud_pista
    def mostrar_info(self) -> None:
        print(f"Pista número: {self.numero_pista}, Longitud: {self.longitud_pista} metros")

class Aeropuerto:
    def __init__(self, nombre_aeropuerto: str, pistas: list) -> None:
        self.nombre_aeropuerto: str = nombre_aeropuerto
        self.pistas: list = pistas
    def agregar_pista(self, pista: Pista) -> None:
        self.pistas.append(pista)
        print(f"Pista {pista.numero_pista} agregada al aeropuerto {self.nombre_aeropuerto}.")
    def mostrar_info(self) -> None:
        print(f"Aeropuerto: {self.nombre_aeropuerto}, Número de pistas: {len(self.pistas)}")
        for pista in self.pistas:
            pista.mostrar_info()
    def volar(self, avion: Avion, pista: Pista, pasajeros: list) -> None:
        if len(pasajeros) > avion.max_pasajeros_avion:
            print(f"No se puede volar con el avión {avion.nombre_avion} ya que excede la capacidad de {avion.max_pasajeros_avion} pasajeros.")
            return
        if pista.longitud_pista < avion.min_longitud_pista:
            print(f"No se puede volar con el avión {avion.nombre_avion} en la pista {pista.numero_pista} ya que requiere una longitud de pista de {avion.min_longitud_pista} metros.")
            return
        print(f"El avión {avion.nombre_avion} ha despegado desde la pista {pista.numero_pista} con los pasajeros: {', '.join(pasajeros)}.")

pasajeros_iniciales = [pasajero("Juan", 25), pasajero("María", 30), pasajero("Pedro", 35)]
pistas = [Pista(1, 3000), Pista(2, 4000), Pista(3, 5000)]
aviones = [Avion("Boeing 747", 3, 3000), Avion("Airbus A380", 2, 4000), Avion("Embraer E195", 1, 4000)]
aeropuerto = Aeropuerto("Aeropuerto Internacional", pistas)
aeropuerto.agregar_pista(Pista(4, 6000))

aeropuerto.volar(aviones[0], pistas[0], ["Juan", "María", "Pedro"])
aeropuerto.volar(aviones[1], pistas[1], ["Luis", "María"])
aeropuerto.volar(aviones[2], pistas[2], ["Pedro", "Luis"])



```

### Ejemplo Incorrecto donde se viola el SRP

En este caso, la clase Book tiene múltiples responsabilidades:

```python
class Book:
    def __init__(self, title: str, author: str) -> None:
        self.title: str = title
        self.author: str = author

    def print_book(self) -> None:
        print(f"Title: {self.title}, Author: {self.author}")

    def save_book(self) -> None:
        # Código para guardar el libro en una base de datos
        pass
```

```python
# Ejemplo de uso
book = Book(title="1984", author="George Orwell")
book.print_book()
book.save_book()
```

#### Análisis del Ejemplo Incorrecto

- **Book**: Mezcla tres responsabilidades: representar los datos del libro, imprimir su información y guardar el libro en una base de datos.

- Cualquier cambio en el formato de impresión o en el motor de persistencia obliga a modificar la clase `Book`, aumentando el acoplamiento y el riesgo de introducir errores.

Este diseño viola el SRP porque la clase tiene más de una razón para cambiar.

## Conclusión

1. **Reducción de Complejidad**: Aplicar el SRP reduce la complejidad del código al dividir responsabilidades en clases separadas.

2. **Mejora de la Cohesión**: Las clases con una única responsabilidad tienen una cohesión más alta, lo que facilita su comprensión y mantenimiento.

3. **Facilidad de Prueba**: Las clases con una única responsabilidad son más fáciles de probar de manera aislada.

4. **Reusabilidad**: Las clases con responsabilidades claras y únicas son más fáciles de reutilizar en diferentes contextos.

Aplicar el SRP puede requerir más clases y una planificación cuidadosa, pero los beneficios a largo plazo en términos de calidad del código y facilidad de mantenimiento son significativos.
