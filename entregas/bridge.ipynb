```python
class CuentasBancariasEmail:
    def autentificar(self, mensaje):
        print(f"Email: {mensaje}")

class CuentasBancariasSMS:
    def autentificar(self, mensaje):
        print(f"SMS: {mensaje}")


class Autentificacion:
    def autentificar(self, mensaje):
        pass

class Email(Autentificacion):
    def autentificar(self, mensaje):
        print(f"Email: {mensaje}")

class SMS(Autentificacion):
    def autentificar(self, mensaje):
        print(f"SMS: {mensaje}")

class CuentaBancaria:
    def __init__(self, autentificacion):
        self.autentificacion = autentificacion

    def enviar_mensaje(self, mensaje):
        self.autentificacion.autentificar(mensaje)


cuenta = CuentaBancaria(Email())
cuenta.enviar_mensaje("Bienvenido a tu cuenta bancaria por Email")


```
