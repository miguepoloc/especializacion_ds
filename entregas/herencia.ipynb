Herencia

```python
class Animal:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def hace_sonido(self):
        return "Este animal hace un sonido"

    def info(self):
        return f"{self.nombre} tiene {self.edad} años."


class Perro(Animal):
    def __init__(self, nombre, edad, raza):
        super().__init__(nombre, edad)
        self.raza = raza
    def hacer_sonido(self):
        return "Guau"
    def traer_pelota(self):
        return f"{self.nombre} esta trayendo la pelota"



class Gato(Animal):
    def __init__(self, nombre, edad, es_curioso):
        super().__init__(nombre, edad)
        self.es_curioso = es_curioso
    def hacer_sonido(self):
        return "miau"
    def ronronear(self):
        return f"{self.nombre} esta ronroneando."

mi_perro = Perro("Firulais", 3 , "Labrador")
mi_gato = Gato("Michi", 2 , True)

print(mi_perro.info())  # Firulais tiene 3 años.
print(mi_perro.hacer_sonido())  # Guau
print(mi_perro.traer_pelota())  # Firulais está trayendo la pelota.

print(mi_gato.info())  # Michi tiene 2 años.
print(mi_gato.hacer_sonido())  # Miau
print(mi_gato.ronronear())  # Michi está ronroneando.
```
