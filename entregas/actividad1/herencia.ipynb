```python
#se realizo una clase padre la cual se llama abanico que tiene los siguientes atributos:
#velocidad y el encendido, ademas de eso tiene 3 funciones el cual son encender, cambiar de velocidad
#y apagarse de esta clase heredan ds clases que son la clase abanicotecho y abanicodepata, usamos la herencia
#y usamos el metodo super para llamar el metodo de la clase padre a las clases hijas
```

```python
class Abanico:
    def __init__(self, velocidad: int=1, encendido:bool= False):
        self.velocidad = velocidad
        self.encendido = encendido

    def encender(self)->None:
        self.encendido=True

    def cambiar_de_velocidad(self,nueva_velocidad:int)->None:
        self.velocidad=nueva_velocidad

    def apagar(self)->None:
        self.encendido=False


class AbanicoTecho(Abanico):
    def __init__(self, velocidad: int =1, tiene_luz:bool=True):
        super().__init__(velocidad)
        self.tiene_luz=tiene_luz

    def encender(self)->None:
        print("Abanico encendido")

class AbanicoDePata(Abanico):
    def __init__(self, velocidad: int =1, altura: float=1.2):
        super().__init__(velocidad)
        self.altura = altura

    def oscilar(self)->None:
        print("el abanico esta girando")

```
