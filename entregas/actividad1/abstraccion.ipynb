```python
class Abanico:
    def __init__(self, velocidad: int, angulo_oscilacion: int, encendido: bool) -> None:
        self.velocidad: int = velocidad
        self.angulo_oscilacion: int = angulo_oscilacion
        self.encendido: bool = encendido

    def soplar(self) -> None:
        if self.encendido:
            print(f"El abanico está soplando a nivel {self.velocidad}")
        else:
            print("El abanico está apagado")

    def __str__(self) -> str:
        estado = "Encendido" if self.encendido else "Apagado"
        return f"Estado: {estado}, Velocidad: Nivel {self.velocidad}, Ángulo: {self.angulo_oscilacion}°"



mi_abanico = Abanico(velocidad=2, angulo_oscilacion=45, encendido=True)

# dEjecuta el método principalll (Abstracción)
mi_abanico.soplar()

# Imprime la representación en textoooo
print(mi_abanico)
```
