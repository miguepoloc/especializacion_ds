    + describir(): str

    + encender(): str

    + encender(): str

Abanico <|-- AbanicoDePata

Abanico <|-- AbanicoDeMano

@enduml

```python
from abc import ABC, abstractmethod


class Abanico(ABC):

    def __init__(self, marca: str, modelo: str) -> None:
        self.marca: str = marca
        self.modelo: str = modelo

    @abstractmethod
    def encender(self) -> str:
        """Método abstracto que cada tipo de abanico debe implementar."""
        pass

    def describir(self) -> str:
        """Devuelve '{marca} {modelo}'."""
        return f"{self.marca} {self.modelo}"


```

```python
class AbanicoDePata(Abanico):

    def encender(self) -> str:
        return "El abanico de pata empieza a girar y a oscilar"
```

```python
class AbanicoDeTecho(Abanico):

    def encender(self) -> str:
        return "El abanico de techo gira suavemente e ilumina la habitación"
```

```python
class AbanicoDeMano(Abanico):

    def encender(self) -> str:
        return "El abanico de mano se despliega manualmente"
```

```python
def probar_abanico(abanico: Abanico) -> None:
    """Función polimórfica: procesa cualquier subclase de Abanico de la misma forma."""
    print(f"Probando {abanico.describir()}:")
    print(abanico.encender())
    print()  # Línea en blanco
```

```python
#
abanico1 = AbanicoDePata("Oster", "Pedestal 16")
abanico2 = AbanicoDeTecho("Whirlpool", "Breeze 2000")
abanico3 = AbanicoDeMano("Artesanal", "Madera")

# Polimorfismo
probar_abanico(abanico1)
probar_abanico(abanico2)
probar_abanico(abanico3)
```
