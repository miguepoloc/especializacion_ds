```python
class Abanico:
    def __init__(self, marca: str, velocidad: int = 0):
        self.marca = marca
        self.__velocidad = velocidad      # Atributo privado
        self.__funciona = False           # Atributo privado

    def get_velocidad(self) -> int:
        return self.__velocidad

    def set_velocidad(self, nueva_velocidad: int) -> None:
        if 0 <= nueva_velocidad <= 3:
            self.__velocidad = nueva_velocidad
        else:
            print("Error: Velocidad no permitida (0 a 3).")

    def is_funciona(self) -> bool:
        return self.__funciona

    def set_funciona(self, estado: bool) -> None:
        self.__funciona = estado


class Cajera:
    def __init__(self, nombre: str):
        self.nombre = nombre

    def registrar_ingreso(self, abanico: Abanico, cliente: str) -> str:
        turno = "TURNO-1"
        print(f"[{self.nombre}] Registró ingreso del abanico '{abanico.marca}' de {cliente}. Asignado: {turno}")
        return turno

    def cobrar_reparacion(self, turno: str, costo: float) -> None:
        print(f"[{self.nombre}] Cobro procesado para el {turno}. Total: ${costo:.2f}")


class Tecnico:
    def __init__(self, nombre: str):
        self.nombre = nombre

    def diagnosticar(self, abanico: Abanico) -> None:
        print(f"[{self.nombre}] Revisando el abanico {abanico.marca}...")
        if not abanico.is_funciona():
            print(f"[{self.nombre}] Diagnóstico: El abanico está averiado y no enciende.")
        else:
            print(f"[{self.nombre}] Diagnóstico: El abanico funciona correctamente.")

    def reparar(self, abanico: Abanico) -> None:
        print(f"[{self.nombre}] Reparando el motor y ajustando velocidades...")
        # métodos encapsulados de Abanico
        abanico.set_funciona(True)
        abanico.set_velocidad(1)
        print(f"[{self.nombre}] ¡Reparación terminada! Estado funcional: {abanico.is_funciona()}.")
```

```python

abanico_cliente = Abanico(marca="Oster", velocidad=0)

cajera = Cajera("María")
tecnico = Tecnico("Carlos")

print("--- 1. INGRESO AL LOCAL ---")
turno = cajera.registrar_ingreso(abanico_cliente, cliente="Josué")

print("\n--- 2. TALLER DE REPARACIÓN ---")
tecnico.diagnosticar(abanico_cliente)
tecnico.reparar(abanico_cliente)

print("\n--- 3. ENTREGA Y COBRO ---")
print(f"Velocidad actual en prueba final: {abanico_cliente.get_velocidad()}")
cajera.cobrar_reparacion(turno, costo=45000.0)
```
