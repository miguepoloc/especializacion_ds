#EJEMPLO CORRECTO

```python
from abc import ABC, abstractmethod

class CanalNotificacion(ABC):
    @abstractmethod
    def enviar(self, mensaje:str):
        pass

class NotificacionEmail(CanalNotificacion):
    def enviar(self, mensaje:str):
        print(f"Enviando notificación por email: {mensaje}")

class NotificacionSMS(CanalNotificacion):
    def enviar(self, mensaje:str):
        print(f"Enviando notificación por SMS: {mensaje}")

class NotificacionWsp(CanalNotificacion):
    def enviar(self, mensaje:str):
        print(f"Enviando notificación por WhatsApp: {mensaje}")
```

```python
mensaje = "¡Hola! Tienes una nueva notificación de WhatsApp"
notificacion_wsp = NotificacionWsp()
notificacion_wsp.enviar(mensaje)

mensaje2 = "¡Hola! Tienes una nueva notificación de SMS"
notificacion_sms = NotificacionSMS()
notificacion_sms.enviar(mensaje2)

```

#INCORRECTO

```python
class Notificador:
    def enviar_notificacion(self, tipo_canal: str, mensaje: str):
        canal = None
        if tipo_canal == "email":
            canal = NotificacionEmail()
        elif tipo_canal == "sms":
            canal = NotificacionSMS()
        elif tipo_canal == "wsp":
            canal = NotificacionWsp()
        else:
            raise ValueError("Tipo de canal no válido")

        canal.enviar(mensaje)
```

```python
noti = Notificador()
noti.enviar_notificacion("email", "Hola, Tienes una nueva notificación")
noti.enviar_notificacion("sms", "Hola, Tienes una nueva notificación de SMS")
```
