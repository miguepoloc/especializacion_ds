```python
from abc import ABC, abstractmethod

class Volar(ABC):
    @abstractmethod
    def volar(self) -> None:
        pass

class Comer(ABC):
    @abstractmethod
    def comer(self) -> None:
        pass

class Caminar(ABC):
    @abstractmethod
    def caminar(self) -> None:
        pass


class Golondrina(Volar, Comer, Caminar):
    def volar(self) -> None:
        print("La golondrina está volando")

    def comer(self) -> None:
        print("La golondrina está comiendo")

    def caminar(self) -> None:
        print("La golondrina está caminando")

class PinguinoMocho(Comer):
    def comer(self) -> None:
        print("El pinguino mocho está comiendo")

class Humano(Comer, Caminar):
    def comer(Self) -> None:
        print("El humano está comiendo")

    def caminar(self) -> None:
        print("El humano está caminando")


```

```python
from abc import ABC, abstractmethod

class Servivo:
    def comer(self) -> None:
        pass
    def volar(self) -> None:
        pass
    def caminar(self) -> None:
        pass

class Golondrina(Servivo):
    def volar(self) -> None:
        print("La golondrina está volando")

    def comer(self) -> None:
        print("La golondrina está comiendo")

    def caminar(self) -> None:
        print("La golondrina está caminando")

class PinguinoMocho(Servivo):
    def comer(self) -> None:
        print("El pinguino mocho está comiendo")

    def caminando(self) -> None:
        print("El pinguino mocho está caminando")
    def volar(self) -> None:
        print("El pinguino esta volando ")

```

```python
g = Golondrina()
p = PinguinoMocho()

print("Golondrina:")
g.volar()
g.comer()
g.caminar()


print("Pinguino Mocho:")
p.comer()
p.volar()
p.caminando()
```
