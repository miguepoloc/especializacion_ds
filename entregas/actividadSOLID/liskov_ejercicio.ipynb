#CORRECTO

```python
from abc import ABC, abstractmethod

class ave(ABC):
    @abstractmethod
    def desplazarse(self) -> str:
        pass

class aveVoladora(ave):
    def desplazarse(self) -> str:
        return "Estoy volando"

class golondrina(aveVoladora):
    pass

class pinguino(ave):
    def desplazarse(self) -> str:
        return "Estoy nadando"


def moverse(a: ave) -> None:
    print(a.desplazarse())

```

```python
g = golondrina()
p = pinguino()

moverse(g)
moverse(p)
```

#INCORRECTO

```python
class ave:
    def volar(self) -> str:
        return "Estoy volando"
class golondrina(ave):
        pass
class pinguino(ave):
        def volar(self) -> None:
            print("Los pingüinos no pueden volar")

def hacer_volar_ave(ave: ave) -> str:
        print(ave.volar().upper())
```

```python

p=pinguino()

hacer_volar_ave(p)
```
