#COREECTO

```python
from abc import ABC, abstractmethod

class MetodoPago(ABC):
    @abstractmethod
    def pagar(self, monto: float) -> None:
        pass
# estas son las de bajo nivel
class PagoPse(MetodoPago):
    def pagar(self, monto: float) -> None:
        print(f"Procesando pago de {monto} a través de PSE.")

class PagoEfectivo(MetodoPago):
    def pagar(self, monto: float) -> None:
        print(f"Procesando pago de {monto} en efectivo.")

class PagoTarjeta(MetodoPago):
    def pagar(self, monto:float) -> None:
        print(f"Procesando pago de {monto} con tarjeta.")
#ALTO nivel
class ProcesoPAgo:
    def __init__(self, metodo: MetodoPago) -> None:
        self.metodo = metodo

    def finalizar_pago(self, monto: float) -> None:
        print("Iniciando proceso de pago...")
        self.metodo.pagar(monto)
        print("Pago finalizado exitosamente.")
```

```python
pago_pse = PagoPse()
comprador1 = ProcesoPAgo(pago_pse)
comprador1.finalizar_pago(100)

pago_efectivo = PagoEfectivo()
comprador2 = ProcesoPAgo(pago_efectivo)
comprador2.finalizar_pago(50)
```

#INCORRECTO

```python
class MetodoPag:
    def pagar(self, monto: float) -> None:
        print(f"Procesando pago de {monto} a través de PSE.")

class ProcesoPag:
    def __init__(self, metodo: MetodoPag) -> None:
        self.metodo = metodo

    def finalizar_pago(self, monto: float) -> None:
        print("Iniciando proceso de pago...")
        self.metodo.pagar(monto)
        print("Pago finalizado exitosamente.")
```

```python
metodo = MetodoPag()

compra = ProcesoPag(metodo)
compra.finalizar_pago(200)
```
