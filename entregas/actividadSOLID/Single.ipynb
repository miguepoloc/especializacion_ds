```python
class Factura:
    def __init__(self, cliente: str, cantidad:float):
        self.cliente = cliente
        self.cantidad = cantidad

    def calcular_toal(self) -> float:
        return self.cantidad * 1.21

class ImpresoraFactura:

    def imprimir(self,factura: Factura):
        print(f"Factura para: {factura.cliente}")

class GuardarFactura:

    def guardar(self, factura: Factura):
        print(f"Guardando factura para: {factura.cliente} con cantidad: {factura.cantidad}")

```

```python
f = Factura("josue",100)

impresora = ImpresoraFactura()
impresora.imprimir(f)

gua=GuardarFactura()
gua.guardar(f)
```
