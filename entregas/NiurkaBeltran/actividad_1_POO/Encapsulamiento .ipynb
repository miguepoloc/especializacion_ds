```python

class Cuenta:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.__saldo = saldo

    def consultar_saldo(self):
        return self.__saldo

    def depositar(self, cantidad):
        if cantidad > 0:
            self.__saldo += cantidad
```

```python

cuenta1 = Cuenta("Laura", 100000)

print("Titular:", cuenta1.titular)
print("Saldo:", cuenta1.consultar_saldo())
```

```python
cuenta1.depositar(50000)

print("Nuevo saldo:", cuenta1.consultar_saldo())
```
