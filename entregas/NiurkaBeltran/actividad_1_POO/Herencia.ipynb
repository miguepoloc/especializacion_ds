```python

class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def comer(self):
        print(self.nombre, "está comiendo")
```

```python

class Perro(Animal):
    def ladrar(self):
        print(self.nombre, "está ladrando")
```

```python
perro1 = Perro("Max")

perro1.comer()
perro1.ladrar()
```

```python
class Gato(Animal):
    def maullar(self):
        print(self.nombre, "está maullando")
```

```python
gato1 = Gato("Michi")

gato1.comer()
gato1.maullar()
```
