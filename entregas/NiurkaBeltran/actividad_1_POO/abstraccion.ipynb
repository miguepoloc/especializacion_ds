ABSTRACCIÓN
Para este tema tome la idea de un libro, porque estos se encuentras en dos sistemas, uno de esos es la bliblioteca que a esta le interesa la parte del título, autor y disponibilidad. Y la otra seria una libreria, que tambien le interesa titulo, autor y precio.

con esto podemos demostrar que un mismo libro puede ser representado de diferentes maneras dependiedno del problema.


```python
#en este caso tendriamos una clase que representa a un libro en una biblioteca#

class LibroBiblioteca:
    def __init__(self, titulo, autor, disponible):
        self.titulo = titulo
        self.autor = autor
        self.disponible = disponible

    def mostrar(self):
        print("Título:", self.titulo)
        print("Autor:", self.autor)
        print("Disponible:", self.disponible)
```

```python

libro1 = LibroBiblioteca("Cien años de soledad", "Gabriel García Márquez", True)

libro1.mostrar()
```

```python
#en este caso tendriamos una clase que representa a un libro en una libreria#
class LibroLibreria:
    def __init__(self, titulo, autor, precio):
        self.titulo = titulo
        self.autor = autor
        self.precio = precio

    def mostrar(self):
        print("Título:", self.titulo)
        print("Autor:", self.autor)
        print("Precio:", self.precio)
```

```python
libro2 = LibroLibreria("Cien años de soledad", "Gabriel García Márquez", 45000)

libro2.mostrar()
```
