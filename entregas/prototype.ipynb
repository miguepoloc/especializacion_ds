Crea tu propio Prototype para clonar objetos de una clase personalizada (por ejemplo, usuario, producto, etc.).


```python
import copy
class Usuario:
    def __init__(self, texto):
        self.texto= texto
    def clonar (self):
        return copy.deepcopy(self)

usu1 = Usuario ("original")

usu2 = usu1.clonar()
print(usu2.texto)
usu2.texto = "copia modificada"
print(usu1.texto)
print(usu2.texto)
```
