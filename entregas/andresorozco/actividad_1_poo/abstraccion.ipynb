### Ejemplo 1: Musico

En este ejemplo, la clase `Musico` es una abstracción de un musico. Observemos cómo se aplica la abstracción:

1. **Atributos relevantes**: Se han seleccionado solo los atributos más importantes para la simulación (nombre, edad, experencia,genero).

2. **Método simplificado**: El método `tocar()` es una representación simplificada de hacer musica.

3. **Representación en cadena**: El método `__str__()` proporciona una representación concisa del estado del avión.

@startuml
' Ejemplo 1: AirPlane (simulación de vuelo)
class Musico {
    + nombre: str
    + edad: int
    + experiencia: int
    + genero: str
    + tocar()
}
@enduml


```python
class Musico:
    def __init__(self, nombre: int, edad: int, experiencia : int, genero : str ) -> None:
        self.nombre: int = nombre
        self.edad: int = edad
        self.experiencia: int = experiencia
        self.genero : str = genero


    def tocar(self) -> None:
        print(f"Musico esta tocando {self.genero}.")

    def __str__(self) -> str:
        return f"Hola, soy {self.nombre} , {self.edad} años, {self.experiencia} años, Ritmo {self.genero}"
```

```python
musico = Musico("Juan",25,3,"cumbia")
print(musico)
musico.tocar()
```


### Ejemplo 2: Reserva de musico

En este segundo ejemplo, la clase `MusicoAgenda` es una abstracción centrada en la reserva de un show para un musico. Veamos cómo se aplica la abstracción:

1. **Atributos simplificados**: Solo se consideran el cuantos que dias tiene disponibles y el mes

2. **Métodos específicos**: Se incluyen métodos relevantes para la reserva (`agendar_show`, `pagar_show`).

3. **Encapsulación**: La validación de disponibilidad



@startuml
class MusicoAgenda {
    + cupos: int
    + mes: str
    + agendar()
    + valida_disponibilidad()
}
@enduml

```python
class MusicoAgenda:
    def __init__(self, cupos: int, mes: str) -> None:
        self.cupos: int = cupos
        self.mes: str = mes

    def agendar(self) -> None:
        if self.valida_disponibilidad():
            print(f"Agendado")
        else:
            print(f"No hay cupos en {self.mes}")

    def valida_disponibilidad(self) -> bool:
        return self.cupos > 0

```

```python
musico_agenda = MusicoAgenda(30, "junio")
musico_agenda.agendar()
```

```python
musico_agenda = MusicoAgenda(0, "junio")
musico_agenda.agendar()
```
