### Ejemplo : Herencia básica y uso de `super()`

Veamos un ejemplo de herencia utilizando el concepto de vehículos:

## Explicación del código

1. **Clase base `Muisico`**: Define atributos y métodos comunes a todos los vehículos.

2. **Clase derivada `Guitarrista`**: Hereda de `Musico` y añade un atributo específico (`tipo`).

3. **Clase derivada `Corista`**: Hereda de `Musico`, añade un atributo (`Voz`) y un método específico (`armonizar`).

4. **Uso de `super()`**: En las clases derivadas, se usa para llamar al constructor de la clase base.

5. **Sobrescritura de métodos**: `Guitarrista` sobrescribe `tocar()` para incluir información adicional.

6. **Polimorfismo**: El bucle for trata tanto a `Guitarrista` como a `Corista` como `musico`.

@startuml
class Musico {
    + nombre: str
    + edad: int
    + experiencia: int
    + genero: str
    + tocar()
}

class Guitarrista {
    + tipo: str
}
class Corista {
    + voz: str
    + armonizar()
}

Musico <|-- Guitarrista
Musico <|-- Corista
@enduml

```python
class Musico:
    def __init__(self, nombre: int, edad: int, experiencia : int, genero : str ) -> None:
        self.nombre: int = nombre
        self.edad: int = edad
        self.experiencia: int = experiencia
        self.genero : str = genero


    def tocar(self) -> None:
        print(f"Musico esta tocando {self.genero}.")

    def __str__(self) -> str:
        return f"Hola, soy {self.nombre} , {self.edad} años, {self.experiencia} años, Ritmo {self.genero}"


class Guitarrista(Musico):
    def __init__(self, nombre, edad, experiencia, genero, tipo: str):
        self.tipo :str = tipo
        super().__init__(nombre, edad, experiencia, genero)

    def tocar(self) -> None:
            print(f"Musico {type(self).__name__} esta tocando {self.genero}")

class Corista(Musico):
    def __init__(self, nombre, edad, experiencia, genero, voz: str):
        self.voz : str = voz
        super().__init__(nombre, edad, experiencia, genero)

    def armonizar(self):
        print("Estoy armonizando")
```

```python
musico = Guitarrista("Juan",25,3,"cumbia","acompañante")
print(musico)
musico.tocar()

musico = Corista("Juan",25,3,"Salsa","Primera")
print(musico)
musico.tocar()
```
