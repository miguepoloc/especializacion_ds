### correcto:

```python
class Animal:
    def __init__(self,nombre:str):
        self.nombre = nombre
    def comer(self):
        return f"{self.nombre} come"


class Ave(Animal):
    def __init__(self, nombre: str, como_vuela: str):
        super().__init__(nombre)
        self.como_vuela = como_vuela
    def volar(self):
        return f"{self.nombre} {self.como_vuela}"
class Mamifero(Animal):
    def __init__(self, nombre: str, como_corre: str):
        super().__init__(nombre)
        self.como_corre = como_corre
    def correr(self):
        return f"{self.nombre} {self.como_corre}"
class Pez(Animal):
    def __init__(self, nombre: str, como_nada: str):
        super().__init__(nombre)
        self.como_nada = como_nada
    def nadar(self):
        return f"{self.nombre} {self.como_nada}"

bocachico = Pez("bocachico", "nada muy bien")
print(bocachico.nombre)
print(bocachico.nadar())
print(bocachico.como_nada)
print(bocachico.comer())


```
