# Encapsulamiento - Videojuegos

En el contexto de un videojuego, un personaje no debe permitir que su vida o mana se cambien libremente desde fuera.
Por eso el estado interno se protege y solo se expone a trav?s de una interfaz p?blica con getters y setters.




```python
@startuml
class Personaje {
  - _vida: int
  - _mana: int
  + get_nombre(): str
  + get_vida(): int
  + set_vida(valor: int): void
  + get_mana(): int
  + set_mana(valor: int): void
  + recibir_dano(dano: int): void
  + lanzar_hechizo(costo: int): str
}
@enduml
```

```python
class Personaje:
    def __init__(self, nombre: str, vida: int, mana: int) -> None:
        self.__nombre = nombre
        self._vida = vida
        self._mana = mana

    def get_nombre(self) -> str:
        return self.__nombre

    def get_vida(self) -> int:
        return self._vida

    def set_vida(self, valor: int) -> None:
        if valor < 0:
            raise ValueError("La vida no puede ser negativa")
        self._vida = valor

    def get_mana(self) -> int:
        return self._mana

    def set_mana(self, valor: int) -> None:
        if valor < 0:
            raise ValueError("El mana no puede ser negativo")
        self._mana = valor

    def recibir_dano(self, dano: int) -> None:
        self._vida = max(0, self._vida - dano)

    def lanzar_hechizo(self, costo: int) -> str:
        if self._mana < costo:
            raise ValueError("No tienes suficiente mana")
        self._mana -= costo
        return "Hechizo lanzado con exito"


personaje = Personaje("Nyra", 100, 40)
print(personaje.get_nombre())
personaje.recibir_dano(25)
print(personaje.get_vida())
print(personaje.lanzar_hechizo(10))
print(personaje.get_mana())
```
