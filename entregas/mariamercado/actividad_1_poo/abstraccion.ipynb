# Abstraccion - Videojuegos

En este notebook se modela la misma entidad del mundo real (un jugador) en dos contextos distintos.
La idea es mostrar que la abstracci?n depende del problema que se quiere resolver.




```python
@startuml
class JugadorPartida {
  + nombre: str
  + nivel: int
  + vida: int
  + mostrar_estado(): str
}

class JugadorInventario {
  + nombre: str
  + oro: int
  + skin: str
  + arma: str
  + mostrar_inventario(): str
}
@enduml
```

```python
class JugadorPartida:
    def __init__(self, nombre: str, nivel: int, vida: int) -> None:
        self.nombre = nombre
        self.nivel = nivel
        self.vida = vida

    def mostrar_estado(self) -> str:
        return f"{self.nombre} esta en nivel {self.nivel} y tiene {self.vida} de vida."


class JugadorInventario:
    def __init__(self, nombre: str, oro: int, skin: str, arma: str) -> None:
        self.nombre = nombre
        self.oro = oro
        self.skin = skin
        self.arma = arma

    def mostrar_inventario(self) -> str:
        return f"{self.nombre} tiene {self.oro} oro, skin {self.skin} y arma {self.arma}."


jugador_partida = JugadorPartida("Astra", 14, 120)
jugador_inventario = JugadorInventario("Astra", 5800, "Neon", "Katana")

print(jugador_partida.mostrar_estado())
print(jugador_inventario.mostrar_inventario())
```
