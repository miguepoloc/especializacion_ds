# Herencia - Videojuegos

Se modela una jerarquia real: En el contexto de un video juego todo personaje es un personaje,
pero el guerrero y el mago tienen comportamientos distintos.




```python
@startuml
class Personaje {
  + nombre: str
  + vida: int
  + atacar(): str
}

class Guerrero {
  + arma: str
  + atacar(): str
}

class Mago {
  + mana: int
  + atacar(): str
}

Personaje <|-- Guerrero
Personaje <|-- Mago
@enduml
```

```python
class Personaje:
    def __init__(self, nombre: str, vida: int) -> None:
        self.nombre = nombre
        self.vida = vida

    def atacar(self) -> str:
        return f"{self.nombre} realiza un ataque b?sico."


class Guerrero(Personaje):
    def __init__(self, nombre: str, vida: int, arma: str) -> None:
        super().__init__(nombre, vida)
        self.arma = arma

    def atacar(self) -> str:
        return f"{self.nombre} golpea con {self.arma} y hace mucho daño."


class Mago(Personaje):
    def __init__(self, nombre: str, vida: int, mana: int) -> None:
        super().__init__(nombre, vida)
        self.mana = mana

    def atacar(self) -> str:
        return f"{self.nombre} lanza una bola de fuego con {self.mana} de mana."


equipo = [Guerrero("Ragnar", 150, "hacha"), Mago("Elira", 90, 40)]

for personaje in equipo:
    print(personaje.atacar())
```
