# Polimorfismo - Videojuegos

En el contexto de los videojuegos los personajes tienen una habilidad especial con el mismo nombre,
pero cada subclase implementa ese comportamiento de manera distinta.




```python
@startuml
class Personaje {
  + nombre: str
  + habilidad_especial(): str
}

class Tanque {
  + habilidad_especial(): str
}

class Curandero {
  + habilidad_especial(): str
}

class Asesino {
  + habilidad_especial(): str
}

Personaje <|-- Tanque
Personaje <|-- Curandero
Personaje <|-- Asesino
@enduml
```

```python
class Personaje:
    def __init__(self, nombre: str) -> None:
        self.nombre = nombre

    def habilidad_especial(self) -> str:
        return f"{self.nombre} usa una habilidad general."


class Tanque(Personaje):
    def habilidad_especial(self) -> str:
        return f"{self.nombre} activa su escudo y bloquea el daño."


class Curandero(Personaje):
    def habilidad_especial(self) -> str:
        return f"{self.nombre} cura a todo el equipo."


class Asesino(Personaje):
    def habilidad_especial(self) -> str:
        return f"{self.nombre} ataca por sorpresa y gana ventaja."


equipo = [Tanque("Bram"), Curandero("Lia"), Asesino("Vex")]

for personaje in equipo:
    print(personaje.habilidad_especial())
```
