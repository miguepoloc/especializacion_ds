### Explicación
El encapsulamiento sirve para proteger las variables internas de un objeto y evitar que se modifiquen de forma indebida desde afuera.

En esta clase `TermostatoIndustrial`:
- Los atributos como `__temperatura_setpoint` y `__esta_bloqueado` son privados (tienen doble guion bajo `__`).
- Para leer o cambiar valores se usan métodos como `obtener_setpoint()` y `configurar_setpoint()`.
- `configurar_setpoint()` valida con `if/else` que el valor esté en un rango seguro (-35°C a 15°C) y que el panel no esté bloqueado.

### Diagrama UML
```plantuml

    - __temperatura_setpoint: float

    + desbloquear_panel(clave: str): bool

```

````python
class TermostatoIndustrial:
    def __init__(self, id_dispositivo: str, setpoint_inicial: float, limite_min: float = -35.0, limite_max: float = 15.0) -> None:
        self.__id_dispositivo: str = id_dispositivo
        self.__temperatura_limite_min: float = limite_min
        self.__temperatura_limite_max: float = limite_max
        self.__esta_bloqueado: bool = False
        self.__clave: str = "1234"
        self.__historial: list[str] = []

        # revisar que el valor inicial tenga sentido
        if self.__temperatura_limite_min <= setpoint_inicial <= self.__temperatura_limite_max:
            self.__temperatura_setpoint: float = setpoint_inicial
        else:
            print(f"[{self.__id_dispositivo}] Valor inicial fuera de rango, se deja en {self.__temperatura_limite_min}°C")
            self.__temperatura_setpoint = self.__temperatura_limite_min
        
        self.__historial.append(f"Iniciado en {self.__temperatura_setpoint}°C")

    def obtener_id(self) -> str:
        return self.__id_dispositivo

    def obtener_setpoint(self) -> float:
        return self.__temperatura_setpoint

    def esta_bloqueado(self) -> bool:
        return self.__esta_bloqueado

    def configurar_setpoint(self, nuevo_setpoint: float) -> bool:
        # si esta bloqueado no deja cambiar
        if self.__esta_bloqueado:
            print(f"[{self.__id_dispositivo}] Panel bloqueado. No se puede cambiar temperatura.")
            return False

        # validar rango permitido
        if self.__temperatura_limite_min <= nuevo_setpoint <= self.__temperatura_limite_max:
            msg: str = f"Setpoint cambiado de {self.__temperatura_setpoint}°C a {nuevo_setpoint}°C"
            self.__temperatura_setpoint = nuevo_setpoint
            self.__historial.append(msg)
            print(f"[{self.__id_dispositivo}] {msg}")
            return True
        else:
            print(f"[{self.__id_dispositivo}] Error: {nuevo_setpoint}°C fuera de rango [{self.__temperatura_limite_min}°C a {self.__temperatura_limite_max}°C]")
            return False

    def bloquear_panel(self, clave: str) -> bool:
        if clave == self.__clave:
            self.__esta_bloqueado = True
            self.__historial.append("Panel bloqueado")
            print(f"[{self.__id_dispositivo}] Panel bloqueado.")
            return True
        print(f"[{self.__id_dispositivo}] Clave incorrecta.")
        return False

    def desbloquear_panel(self, clave: str) -> bool:
        if clave == self.__clave:
            self.__esta_bloqueado = False
            self.__historial.append("Panel desbloqueado")
            print(f"[{self.__id_dispositivo}] Panel desbloqueado.")
            return True
        print(f"[{self.__id_dispositivo}] Clave incorrecta.")
        return False

    def obtener_historial(self) -> list[str]:
        return list(self.__historial)

    def __str__(self) -> str:
        estado = "Bloqueado" if self.__esta_bloqueado else "Desbloqueado"
        return f"Termostato {self.__id_dispositivo} | Setpoint: {self.__temperatura_setpoint}°C | {estado}"

print("Clase de encapsulamiento definida correctamente.")
````

````python
# Pruebas de encapsulamiento
term = TermostatoIndustrial("TERM-01", setpoint_inicial=-18.0)
print(term)

# Cambio normal
term.configurar_setpoint(-20.0)
print(f"Setpoint leido con get: {term.obtener_setpoint()}°C")

# Intento con temperatura invalida
term.configurar_setpoint(50.0)

# Bloqueo del panel
term.bloquear_panel("1234")
term.configurar_setpoint(-22.0)  # no debe dejar

# Desbloqueo y cambio
term.desbloquear_panel("1234")
term.configurar_setpoint(-22.0)

# Ver historial registrado
print("\nHistorial de eventos:")
for h in term.obtener_historial():
    print(" -", h)

print("\nEstado final:", term)
````
