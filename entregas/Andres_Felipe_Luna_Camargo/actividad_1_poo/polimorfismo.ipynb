### Explicación
El polimorfismo permite que distintas clases tengan métodos con el mismo nombre pero con comportamiento propio de cada una. La función cliente puede recorrer una lista de objetos y llamar esos métodos sin preguntar de qué tipo es cada objeto.

- Clase abstracta base: `SensorAlarma(ABC)` con los métodos abstractos `evaluar_condicion()` y `activar_protocolo_respuesta()`.
- Subclases: `SensorTemperaturaCritica`, `SensorPresionRefrigerante` y `SensorFugaGasAmoniaco`.
- Función cliente: `auditar_seguridad_planta(sensores)` recibe la lista y ejecuta los métodos en un ciclo `for` sin condicionales de tipo.

````python
from abc import ABC, abstractmethod

# Clase base abstracta
class SensorAlarma(ABC):
    def __init__(self, id_sensor: str, ubicacion: str, valor_inicial: float) -> None:
        self.id_sensor: str = id_sensor
        self.ubicacion: str = ubicacion
        self.valor_actual: float = valor_inicial

    def actualizar_lectura(self, nuevo_valor: float) -> None:
        self.valor_actual = float(nuevo_valor)

    @abstractmethod
    def evaluar_condicion(self) -> str:
        pass

    @abstractmethod
    def activar_protocolo_respuesta(self) -> str:
        pass

    def __str__(self) -> str:
        return f"Sensor {self.id_sensor} ({self.ubicacion}) | Lectura: {self.valor_actual}"


# Subclase 1: Sensor de temperatura
class SensorTemperaturaCritica(SensorAlarma):
    def __init__(self, id_sensor: str, ubicacion: str, temp_inicial: float, temp_umbral_max: float = 4.0) -> None:
        super().__init__(id_sensor, ubicacion, temp_inicial)
        self._temp_umbral_max: float = temp_umbral_max

    def evaluar_condicion(self) -> str:
        if self.valor_actual > self._temp_umbral_max:
            return f"Alerta caliente: {self.valor_actual:.1f}°C (umbral {self._temp_umbral_max:.1f}°C)"
        return f"Temperatura ok: {self.valor_actual:.1f}°C"

    def activar_protocolo_respuesta(self) -> str:
        if self.valor_actual > self._temp_umbral_max:
            return "Prender compresor auxiliar al 100% y avisar a bodega"
        return "Ninguna accion"


# Subclase 2: Sensor de presion
class SensorPresionRefrigerante(SensorAlarma):
    def __init__(self, id_sensor: str, ubicacion: str, presion_inicial: float, presion_min_psi: float = 20.0) -> None:
        super().__init__(id_sensor, ubicacion, presion_inicial)
        self._presion_minima_segura_psi: float = presion_min_psi

    def evaluar_condicion(self) -> str:
        if self.valor_actual < self._presion_minima_segura_psi:
            return f"Alerta baja presion: {self.valor_actual:.1f} PSI (minimo {self._presion_minima_segura_psi:.1f} PSI)"
        return f"Presion ok: {self.valor_actual:.1f} PSI"

    def activar_protocolo_respuesta(self) -> str:
        if self.valor_actual < self._presion_minima_segura_psi:
            return "Apagar compresor y cerrar valvulas de paso"
        return "Ninguna accion"


# Subclase 3: Sensor de fuga de amoniaco
class SensorFugaGasAmoniaco(SensorAlarma):
    def __init__(self, id_sensor: str, ubicacion: str, ppm_inicial: float, ppm_evacuacion: float = 25.0) -> None:
        super().__init__(id_sensor, ubicacion, ppm_inicial)
        self._ppm_evacuacion: float = ppm_evacuacion

    def evaluar_condicion(self) -> str:
        if self.valor_actual >= self._ppm_evacuacion:
            return f"PELIGRO FUGA: {self.valor_actual:.1f} PPM de amoniaco"
        return f"Gas normal: {self.valor_actual:.1f} PPM"

    def activar_protocolo_respuesta(self) -> str:
        if self.valor_actual >= self._ppm_evacuacion:
            return "Activar extractores y sonar alarma de evacuacion"
        return "Ninguna accion"


# Funcion cliente (polimorfismo puro: no revisa tipos con if)
def auditar_seguridad_planta(sensores: list[SensorAlarma]) -> None:
    print("--- REVISION DE SENSORES EN PLANTA ---")
    for s in sensores:
        print(f"[{s.id_sensor}] Ubicacion: {s.ubicacion}")
        print("  Estado:", s.evaluar_condicion())
        print("  Accion:", s.activar_protocolo_respuesta())
        print("-" * 45)

print("Clases y función de polimorfismo definidas correctamente.")
````

````python
# Creamos varios sensores de diferente tipo
s1 = SensorTemperaturaCritica("S-TEMP-01", "Cuarto Carnes", temp_inicial=6.5, temp_umbral_max=4.0)
s2 = SensorPresionRefrigerante("S-PRES-01", "Sala Maquinas", presion_inicial=15.0, presion_min_psi=20.0)
s3 = SensorFugaGasAmoniaco("S-GAS-01", "Pasillo Central", ppm_inicial=30.0, ppm_evacuacion=25.0)
s4 = SensorTemperaturaCritica("S-TEMP-02", "Cuarto Lacteos", temp_inicial=2.0, temp_umbral_max=4.0)

# Lista con diferentes tipos de sensores
lista_sensores: list[SensorAlarma] = [s1, s2, s3, s4]

# Pasamos la lista a la funcion cliente
auditar_seguridad_planta(lista_sensores)
````
