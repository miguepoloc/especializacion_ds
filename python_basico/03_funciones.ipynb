# Python de 0 a experto — 3. Funciones

## Introducción
Las funciones en Python comparten la idea general con cualquier lenguaje
(nombre, parámetros, retorno), pero tienen características propias que se
usan todo el tiempo en código real: **valores por defecto**, **`*args`**,
**`**kwargs`** y funciones como *first-class citizens*.

## Objetivos
- Definir funciones con parámetros posicionales, por nombre y con valores
  por defecto.
- Usar `*args` y `**kwargs` para aceptar un número variable de argumentos.
- Explicar por qué las funciones son objetos de primera clase en Python.

## 1. Definición básica y valores por defecto

`def nombre(parametros) -> tipo_retorno:` — el retorno es opcional; si no
hay `return`, la función devuelve `None` implícitamente.

```python
def saludar(nombre: str, saludo: str = "Hola") -> str:
    return f"{saludo}, {nombre}!"

print(saludar("Ana"))                 # usa el valor por defecto
print(saludar("Luis", "Buenas"))      # lo sobreescribe
print(saludar(nombre="Marta", saludo="Hey"))  # por nombre (keyword)
```

**Cuidado**: un valor por defecto mutable (`lista=[]`) se crea **una
sola vez**, no en cada llamada — es uno de los errores más comunes al
migrar desde otros lenguajes.

```python
def agregar_mal(item, lista=[]):     # trampa: la lista se comparte entre llamadas
    lista.append(item)
    return lista

print(agregar_mal("a"))
print(agregar_mal("b"))   # uno esperaría ['b'], pero acumula ['a', 'b']

def agregar_bien(item, lista=None):
    if lista is None:
        lista = []
    lista.append(item)
    return lista

print(agregar_bien("a"))
print(agregar_bien("b"))  # cada llamada parte de una lista nueva
```

## 2. `*args` — argumentos posicionales variables

Empaqueta cualquier cantidad de argumentos posicionales extra en una
tupla dentro de la función.

```python
def sumar_todos(*numeros) -> float:
    print("recibidos como tupla:", numeros)
    return sum(numeros)

print(sumar_todos(1, 2, 3))
print(sumar_todos(10, 20, 30, 40))
print(sumar_todos())  # tupla vacía, suma 0
```

## 3. `**kwargs` — argumentos por nombre variables

Empaqueta cualquier cantidad de argumentos `clave=valor` extra en un
diccionario dentro de la función.

```python
def crear_perfil(nombre: str, **datos) -> dict:
    perfil = {"nombre": nombre}
    perfil.update(datos)
    return perfil

p = crear_perfil("Ana", edad=30, ciudad="Bogotá", activo=True)
print(p)

# se pueden combinar posicionales + *args + **kwargs, en ese orden
def resumen(titulo, *args, **kwargs):
    print("Titulo:", titulo)
    print("Extras posicionales:", args)
    print("Extras nombrados:", kwargs)

resumen("Reporte", 1, 2, autor="Ana", version=2)
```

## 4. Desempaquetado al llamar (`*` y `**` en la llamada)

El mismo `*`/`**` sirve para "desempaquetar" una lista o diccionario ya
existente al llamar una función.

```python
def punto3d(x, y, z):
    return f"({x}, {y}, {z})"

coords = [1, 2, 3]
print(punto3d(*coords))          # equivale a punto3d(1, 2, 3)

datos = {"x": 5, "y": 6, "z": 7}
print(punto3d(**datos))          # equivale a punto3d(x=5, y=6, z=7)
```

## 5. Funciones como objetos de primera clase

En Python una función es un valor más: se puede guardar en una variable,
pasar como argumento a otra función o retornar desde otra función.

```python
def cuadrado(n):
    return n ** 2

def cubo(n):
    return n ** 3

def aplicar(func, valor):
    return func(valor)

print(aplicar(cuadrado, 5))
print(aplicar(cubo, 3))
print(aplicar(lambda n: n + 1, 10))   # función anónima (lambda)

operaciones = {"cuadrado": cuadrado, "cubo": cubo}
print(operaciones["cuadrado"](4))
```
