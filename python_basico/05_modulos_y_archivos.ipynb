## 1. Formas de `import`

- `import modulo` — usas `modulo.funcion()`.
- `import modulo as alias` — alias corto (`import numpy as np`).
- `from modulo import funcion` — traes el símbolo directo al espacio de
  nombres actual.
- `from modulo import funcion as alias` — combinación de las dos anteriores.

```python
import math
import math as m
from math import sqrt, pi
from math import factorial as fact

print(math.sqrt(16))     # módulo completo
print(m.pi)               # módulo con alias
print(sqrt(25), pi)       # símbolo importado directo
print(fact(5))             # símbolo importado con alias
```

## 4. Manejo de archivos con `with open(...)`

`with` es un *context manager*: garantiza que el archivo se cierre
automáticamente al salir del bloque, **incluso si ocurre una excepción**
dentro. Es la forma idiomática — evita el patrón propenso a errores de
`open()` ... `close()` manual (donde una excepción antes del `close()`
deja el archivo abierto).

```python
import tempfile
import os

ruta = os.path.join(tempfile.gettempdir(), "notas_curso.txt")

# escribir
with open(ruta, "w", encoding="utf-8") as f:
    f.write("Python\n")
    f.write("POO\n")
    f.write("SOLID\n")
    f.write("Patrones\n")

print("archivo escrito en:", ruta)
print("¿sigue abierto el archivo tras el 'with'? ->", f.closed)  # True: se cerró solo

# leer todo el contenido
with open(ruta, "r", encoding="utf-8") as f:
    contenido = f.read()
print("--- contenido completo ---")
print(contenido)
```

```python
# leer línea por línea (más eficiente en memoria para archivos grandes)
with open(ruta, "r", encoding="utf-8") as f:
    for numero, linea in enumerate(f, start=1):
        print(f"{numero}: {linea.strip()}")

# anexar (append) sin borrar lo anterior
with open(ruta, "a", encoding="utf-8") as f:
    f.write("Refactoring\n")

with open(ruta, "r", encoding="utf-8") as f:
    lineas = f.readlines()
print("total de líneas tras el append:", len(lineas))

os.remove(ruta)  # limpieza: era un archivo temporal de ejemplo
```
