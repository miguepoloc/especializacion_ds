## Introducción
Python trae 4 estructuras de datos "de fábrica" con sintaxis literal
propia: **listas**, **tuplas**, **diccionarios** y **sets**. Elegir la
correcta no es un detalle menor: afecta legibilidad, rendimiento y hasta
corrección (mutabilidad).

## 1. Listas — `list`

Colección **ordenada**, **mutable**, permite duplicados. Es el "array
dinámico" de Python: se usa cuando el orden importa y vas a modificar el
contenido (agregar, quitar, reordenar).

```python
frutas = ["manzana", "pera", "uva"]
frutas.append("mango")           # agrega al final
frutas.insert(1, "kiwi")         # inserta en una posición
frutas.remove("pera")            # quita por valor
ultimo = frutas.pop()            # quita y devuelve el último

print(frutas)
print("se quitó:", ultimo)
print("longitud:", len(frutas))
print("ordenada:", sorted(frutas))
print("slice [0:2]:", frutas[0:2])
print("invertida:", frutas[::-1])
```

## 2. Tuplas — `tuple`

Colección **ordenada**, **inmutable**, permite duplicados. Se usa para
datos que no deben cambiar (coordenadas, registros fijos) y es más liviana
que una lista. También habilita el *unpacking*.

```python
punto = (3, 4)
x, y = punto              # unpacking
print(f"x={x}, y={y}")

# una tupla de 1 elemento necesita coma
un_elemento = (5,)
print(type(un_elemento), un_elemento)

try:
    punto[0] = 99          # TypeError: las tuplas son inmutables
except TypeError as e:
    print("Error esperado:", e)

# unpacking extendido
primero, *resto = (1, 2, 3, 4)
print(primero, resto)
```

## 3. Diccionarios — `dict`

Colección de pares **clave-valor**, ordenada por inserción desde Python
3.7+, mutable, claves únicas. Se usa cuando accedes por nombre/identificador
en vez de por posición (registros, configuración, conteos).

```python
persona = {"nombre": "Ana", "edad": 30, "ciudad": "Bogotá"}
persona["profesion"] = "ingeniera"   # agregar/actualizar
edad = persona.get("edad")            # acceso seguro (no lanza KeyError)
pais = persona.get("pais", "Colombia")  # con valor por defecto

print(persona)
print("edad:", edad, "| pais (default):", pais)

for clave, valor in persona.items():
    print(f"  {clave} -> {valor}")

print("claves:", list(persona.keys()))
print("¿tiene 'edad'?:", "edad" in persona)
```

## 4. Sets — `set`

Colección **no ordenada**, **mutable**, **sin duplicados**. Se usa para
verificar pertenencia rápidamente o para operaciones de teoría de
conjuntos (unión, intersección, diferencia).

```python
vocales_en_texto = set("otorrinolaringologo")
print("letras únicas:", vocales_en_texto)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("unión:", a | b)
print("intersección:", a & b)
print("diferencia a-b:", a - b)
print("diferencia simétrica:", a ^ b)

a.add(10)
a.discard(1)   # no lanza error si no existe (a diferencia de .remove())
print("a modificado:", a)
```

## 5. ¿Cuál uso y cuándo?

| Necesito...                              | Estructura |
|-------------------------------------------|------------|
| Orden + voy a modificar + duplicados OK    | `list`     |
| Orden + NO debe cambiar                    | `tuple`    |
| Acceso por nombre/clave                    | `dict`     |
| Solo valores únicos / pertenencia rápida   | `set`      |

```python
# ejemplo: eliminar duplicados de una lista preservando la idea de "únicos"
compras = ["pan", "leche", "pan", "huevos", "leche", "pan"]
compras_unicas = list(set(compras))
print("original:", compras)
print("únicas (sin orden garantizado):", sorted(compras_unicas))

# ejemplo: contar ocurrencias con un dict
conteo = {}
for item in compras:
    conteo[item] = conteo.get(item, 0) + 1
print("conteo:", conteo)
```
