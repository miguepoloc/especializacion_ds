# Python de 0 a experto — 4. Comprehensions y excepciones

## Introducción
Dos herramientas que se usan constantemente en código Python idiomático:
las **comprehensions** (crear listas/dicts/sets en una línea, en vez de un
`for` con `.append()`) y el **manejo de excepciones** (`try/except/finally`
y excepciones propias).

## Objetivos
- Reescribir bucles `for` + `.append()` como list/dict/set comprehensions.
- Aplicar filtros (`if`) dentro de una comprehension.
- Manejar errores con `try/except/finally` capturando excepciones
  específicas.
- Crear y lanzar excepciones propias heredando de `Exception`.

## 1. List comprehension

Forma compacta de construir una lista a partir de otro iterable, con
filtro opcional. Sintaxis: `[expresion for item in iterable if condicion]`.

```python
numeros = range(1, 11)

# forma "tradicional"
cuadrados_v1 = []
for n in numeros:
    cuadrados_v1.append(n ** 2)

# la misma idea como list comprehension
cuadrados_v2 = [n ** 2 for n in numeros]

print(cuadrados_v1 == cuadrados_v2)
print(cuadrados_v2)

# con filtro: solo los cuadrados de los números pares
pares_al_cuadrado = [n ** 2 for n in numeros if n % 2 == 0]
print(pares_al_cuadrado)
```

## 2. Dict comprehension

Misma idea, pero construyendo un diccionario: `{clave: valor for ... }`.

```python
palabras = ["python", "es", "un", "lenguaje", "genial"]

longitudes = {palabra: len(palabra) for palabra in palabras}
print(longitudes)

# invertir un diccionario (valor -> clave)
invertido = {valor: clave for clave, valor in longitudes.items()}
print(invertido)
```

## 3. Set comprehension

Igual, pero para construir un `set` (elimina duplicados automáticamente):
`{expresion for item in iterable}`.

```python
frase = "el veloz murcielago hindu comia feliz cardillo y kiwi"
letras_unicas = {c for c in frase if c != " "}
print(sorted(letras_unicas))
print("cantidad de letras distintas:", len(letras_unicas))
```

## 4. `try` / `except` / `finally`

- `try`: el código que puede fallar.
- `except TipoDeError:` captura un tipo específico (evita `except:` a
  secas, que atrapa TODO, incluso `KeyboardInterrupt`).
- Se pueden encadenar varios `except`.
- `finally`: se ejecuta siempre, haya habido error o no (cerrar recursos).

```python
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero")
        return None
    except TypeError:
        print("Error: los operandos deben ser numéricos")
        return None
    else:
        print("División exitosa")
        return resultado
    finally:
        print("-- fin del intento de división --")

print(dividir(10, 2))
print(dividir(10, 0))
print(dividir(10, "dos"))
```

## 5. Excepciones propias

Se heredan de `Exception` (o de una subclase más específica). Permiten
representar errores de tu propio dominio, distinguibles de los errores
genéricos del lenguaje.

```python
class SaldoInsuficienteError(Exception):
    """Se lanza cuando una cuenta no tiene fondos suficientes."""
    def __init__(self, saldo: float, monto: float):
        self.saldo = saldo
        self.monto = monto
        super().__init__(
            f"Saldo insuficiente: tienes {saldo}, intentaste retirar {monto}"
        )


def retirar(saldo: float, monto: float) -> float:
    if monto > saldo:
        raise SaldoInsuficienteError(saldo, monto)
    return saldo - monto


try:
    nuevo_saldo = retirar(saldo=100, monto=150)
except SaldoInsuficienteError as e:
    print("Operación rechazada:", e)
    print("Saldo actual:", e.saldo, "| Monto solicitado:", e.monto)
else:
    print("Nuevo saldo:", nuevo_saldo)
```
