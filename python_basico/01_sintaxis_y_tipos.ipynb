## 1. Indentación en vez de llaves

En C/Java/JavaScript los bloques se delimitan con `{ }`. En Python **la
indentación ES la sintaxis**: no es una convención de estilo, es lo que el
intérprete usa para saber dónde empieza y termina un bloque.

Reglas prácticas:
- Se usan 4 espacios por nivel (no tabs, no mezclar).
- Todo lo que abre un bloque (`if`, `for`, `while`, `def`, `class`, `try`...)
  termina en `:` y el bloque va indentado en la línea siguiente.
- Un error de indentación es un `IndentationError` en tiempo de "compilación"
  (en realidad al parsear), no un bug silencioso.

```python
def clasificar(nota: float) -> str:
    if nota >= 4.5:
        return "excelente"
    elif nota >= 3.0:
        return "aprobado"
    else:
        return "reprobado"

for nota in [4.8, 3.2, 2.0]:
    print(nota, "->", clasificar(nota))
```

## 2. Tipado dinámico (que no es lo mismo que débil)

Python es **dinámicamente tipado**: una variable no tiene un tipo fijo
declarado, el tipo lo tiene el *valor* que apunta en cada momento, y puede
cambiar. Pero es **fuertemente tipado**: no convierte tipos incompatibles a
la fuerza (a diferencia de JavaScript, `"3" + 2` no funciona en Python).

```python
x = 10          # x apunta a un int
print(type(x))
x = "diez"      # ahora x apunta a un str; válido, no hay error
print(type(x))

try:
    resultado = "3" + 2   # TypeError: no hay conversión implícita
except TypeError as e:
    print("Error esperado:", e)
```

## 3. Tipos primitivos

| Tipo    | Ejemplo         | Notas |
|---------|-----------------|-------|
| `int`   | `42`, `-7`      | precisión arbitraria, no hay overflow |
| `float` | `3.14`, `1e-3`  | doble precisión (64 bits) |
| `str`   | `'hola'`, `"hola"` | inmutable, comillas simples o dobles son equivalentes |
| `bool`  | `True`, `False` | subtipo de `int` (`True == 1`) |
| `NoneType` | `None`       | equivalente a `null`/`nil` |
| `complex` | `2 + 3j`      | poco común fuera de matemáticas/ingeniería |

```python
edad: int = 30
altura: float = 1.72
nombre: str = "Ana"
es_activo: bool = True
sin_valor = None

print(edad, altura, nombre, es_activo, sin_valor)
print([type(v).__name__ for v in (edad, altura, nombre, es_activo, sin_valor)])

# int de Python no tiene límite de tamaño (a diferencia de un int32/int64)
print(2 ** 100)
```

## 4. Operadores

- **Aritméticos**: `+ - * / // % **` — ojo con `/` (siempre devuelve
  `float`) vs `//` (división entera, "floor division").
- **Comparación**: `== != < > <= >=`.
- **Lógicos**: `and`, `or`, `not` (palabras, no `&& || !`).
- **Pertenencia**: `in`, `not in` — funcionan sobre listas, tuplas, sets,
  strings y diccionarios (sobre las claves).
- **Identidad**: `is`, `is not` — comparan identidad de objeto, no valor
  (se usan sobre todo para comparar con `None`).

```python
a, b = 7, 2
print("7 / 2 =", a / b)      # 3.5 (float)
print("7 // 2 =", a // b)    # 3 (entero)
print("7 % 2 =", a % b)      # 1 (resto)
print("2 ** 10 =", 2 ** 10)  # potencia

edades = [18, 22, 30]
print(25 in edades, "25 in edades")
print(22 in edades, "22 in edades")

valor = None
print(valor is None, "-> se compara con 'is', no con '=='")
```

## 5. Control de flujo: `if` / `for` / `while`

Diferencias clave frente a otros lenguajes:
- No hay `switch`/`case` clásico (desde 3.10 existe `match`, fuera del
  alcance básico de hoy).
- `for` itera sobre **iterables** (listas, rangos, strings...), no es un
  bucle de contador con `for(i=0; i<n; i++)`.
- `range(inicio, fin, paso)` genera la secuencia de índices cuando sí
  necesitas contar.
- `while` existe igual que en otros lenguajes, con `break`/`continue`.

```python
# for sobre un iterable directamente
frutas = ["manzana", "pera", "uva"]
for fruta in frutas:
    print("Fruta:", fruta)

# for con índice explícito via range()
for i in range(0, 10, 2):
    print("par:", i)

# while con break/continue
n = 0
while True:
    n += 1
    if n % 2 == 0:
        continue          # salta pares
    if n > 7:
        break              # corta el ciclo
    print("impar:", n)
```
