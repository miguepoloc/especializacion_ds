# 3. Liskov

Aplicamos el principio de Liskov

1.1 Ejemplo malo


```python
class Telefono:
    def __init__(self, marca: str):
        self.__marca = marca

    @property
    def get_marca(self) -> str:
        return self.__marca

    def encender(self) -> None:
        print(f'Encendiendo el móvil de marca: {self.__marca}')


class Iphone(Telefono):

    def encender(self) -> None:
        print('Encendiendo iPhone...')


class Samsung(Telefono):

    def encender(self) -> None:
        raise Exception('Los teléfonos Samsung no pueden encenderse')


class Usuario:
    def __init__(self, telefono: Telefono):
        self.__telefono = telefono

    def usar_telefono(self) -> None:
        self.__telefono.encender()
```

```python
iphone = Iphone('Apple')
samsung = Samsung('Samsung')

usuario1 = Usuario(iphone)
usuario2 = Usuario(samsung)

usuario1.usar_telefono()
usuario2.usar_telefono()
```

1.2 Ejemplo Bueno

```python
class Iphone(Telefono):
    def encender(self) -> None:
        print(f'Encendiendo el movil de marca: {self.get_marca}')
```

```python
class Usuario:
    def __init__(self, telefono:Telefono):
        self.__telefono = telefono

    def usar_telefono(self) -> None:
        self.__telefono.encender()
```

```python
telefono_normal:Telefono = Telefono('Normal')
telefono_iphone:Iphone = Iphone('Iphone')

usuario_normal:Usuario = Usuario(telefono=telefono_normal)
usuario_iphone:Usuario = Usuario(telefono=telefono_iphone)

usuario_normal.usar_telefono()
usuario_iphone.usar_telefono()
```
