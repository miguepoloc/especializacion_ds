# 5. Dependency Inversion

Aplicamos el principio de Inversion de Dependencia al sieguiente ejemplo que se basa en el pago de un carrito de compra por diferentes medios de pago

1.1 Ejemplo malo

```python
class PagoConTarjeta:
    def procesar_pago(self, monto: float) -> None:
        print(f'Procesando pago por tarjeta con el monto: {monto}')

class PagoPayPal:
    def procesar_pago(self, monto: float) -> None:
        print(f'Procesando pago por PayPal con el monto: {monto}')

class CarritoCompras:
    def __init__(self, tipo_pago: str):
        if tipo_pago == "tarjeta":
            self.metodo_pago = PagoConTarjeta()
        elif tipo_pago == "paypal":
            self.metodo_pago = PagoPayPal()
        else:
            raise ValueError("Método de pago no soportado")

    def pagar(self, monto: float) -> None:
        self.metodo_pago.procesar_pago(monto)


pago_carrito_tarjeta = CarritoCompras("tarjeta")
pago_carrito_tarjeta.pagar(15000)

pago_carrito_paypal = CarritoCompras("paypal")
pago_carrito_paypal.pagar(19600)
```

1.2 Ejemplo bueno

```python
from abc import ABC, abstractmethod

class MetodoPago(ABC):
    @abstractmethod
    def procesar_pago(self, monto:float) -> None:
        pass

class PagoConTarjeta(MetodoPago):
    def procesar_pago(self, monto) -> None:
        print(f'Procesando pago por tarjeta con el monto: {monto}')

class PagoPayPal(MetodoPago):
    def procesar_pago(self, monto) -> None:
        print(f'Procensando pago por PayPal con el monto: {monto}')

class CarritoCompras():
    def __init__(self, metodo_pago:MetodoPago):
        self.metodo_pago = metodo_pago

    def pagar(self, monto:float) -> None:
        self.metodo_pago.procesar_pago(monto)
```

```python
pago_paypal:PagoPayPal = PagoPayPal()
pago_tarjeta:PagoConTarjeta = PagoConTarjeta()
```

```python
pago_carrito_compras = CarritoCompras(pago_tarjeta)
pago_carrito_compras.pagar(15000)

pago_carrito_compras = CarritoCompras(pago_paypal)
pago_carrito_compras.pagar(19600)
```
