# 1. Single Responsibility Actividad

Aplicacamos el principio de responsabilidad unica para las siguientes clase:

Tomaremos como clase de ejemplo una tienda

1.1 Ejemplo malo

```python
class Tienda:
    def __init__(self, nombre:str, admin:str):
        self.nombre = nombre
        self.admin = admin
        self.productos = []

    def agregar_producto(self, producto):
        self.productos.append(producto)

    def mostrar_productos(self):
        for producto in self.productos:
            print(f"Producto: {producto.nombre}, Precio: {producto.precio}")

    def agregar_admin(self, nuevo_admin:str):
        self.admin = nuevo_admin

```

1.2 Ejemplo bueno

```python
class Tienda:
    def __init__(self, nombre):
        self.__nombre = nombre
        self.productos = []


class Producto:
    def __init__(self,tienda:Tienda):
        self.__tienda = tienda

    def get_productos(self) -> list:
        return self.__tienda.productos

    def agregar_producto(self, producto:str) -> None:
        self.__tienda.productos.append(producto)
        return f"Producto agregado: {producto}"

    def eliminar_producto(self, producto:str) -> None:
        if producto in self.__tienda.productos:
            self.__tienda.productos.remove(producto)

class Cliente:
    def __init__(self, nombre):
        self.__nombre = nombre

    def comprar_producto(self, producto:str) -> None:
        print(f"{self.__nombre} compró {producto}.")
```

```python
tienda:Tienda = Tienda("Mi Tienda")
agregar:Producto = Producto(tienda)
cliente:Cliente = Cliente("Juan")

agregar.agregar_producto("Arroz")
agregar.agregar_producto("Frijoles")
agregar.agregar_producto("Aceite")

agregar.get_productos()

cliente.comprar_producto("Arroz")
cliente.comprar_producto("Frijoles")
```
