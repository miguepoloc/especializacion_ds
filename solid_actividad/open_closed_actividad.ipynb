# 2. Open/Closed

Aplicamos el principio de Abierto / Cerrado

1.1 Ejemplo malo

```python
class Banco:
    def __init__(self, tipo_metodo: str):
        self.saldo = 0
        self.tipo_metodo = tipo_metodo

    def depositar(self, valor: float):
        if self.tipo_metodo == "PSE":
            self.saldo += valor
            print(f"Depósito de {valor} realizado con sucesso via PSE.")
        elif self.tipo_metodo == "PayPal":
            self.saldo += valor
            print(f"Depósito de {valor} realizado con sucesso via PayPal.")
        else:
            print("Método de pagamento no soportado.")

    def sacar(self, valor: float):
        if self.tipo_metodo == "PSE":
            if self.saldo >= valor:
                self.saldo -= valor
                print(f"Saque de {valor} realizado con sucesso via PSE.")
            else:
                print("Saldo insuficiente para saque via PSE.")

        elif self.tipo_metodo == "PayPal":
            if self.saldo >= valor:
                self.saldo -= valor
                print(f"Saque de {valor} realizado con sucesso via PayPal.")
            else:
                print("Saldo insuficiente para saque via PayPal.")
        else:
            print("Método de pagamento no soportado.")
```

```python
cuenta_pse = Banco("PSE")
cuenta_pse.depositar(500)
cuenta_pse.sacar(200)
cuenta_pse.sacar(400)

cuenta_paypal = Banco("PayPal")
cuenta_paypal.depositar(1000)
cuenta_paypal.sacar(300)
```

1.2 Ejemplo bueno

```python
from abc import ABC, abstractmethod

class Banco(ABC):
    def __init__(self):
            self.saldo = 0

    @abstractmethod
    def depositar(self, valor):
        pass

    @abstractmethod
    def sacar(self, valor):
        pass

class Pse(Banco):
    def depositar(self, valor):
        self.saldo += valor
        print(f"Depósito de {valor} realizado con sucesso.")

    def sacar(self, valor):
        if self.saldo >= valor:
            self.saldo -= valor
            print(f"Saque de {valor} realizado con sucesso.")
        else:
            print("Saldo insuficiente para saque.")

class PayPal(Banco):
    def depositar(self, valor):
        self.saldo += valor
        print(f"Depósito de {valor} realizado con sucesso via PayPal.")

    def sacar(self, valor):
        if self.saldo >= valor:
            self.saldo -= valor
            print(f"Saque de {valor} realizado con sucesso via PayPal.")
        else:
            print("Saldo insuficiente para saque via PayPal.")
```

```python
cuenta_pse = Pse()
cuenta_pse.depositar(500)
cuenta_pse.sacar(200)
cuenta_pse.sacar(400)

cuenta_paypal = PayPal()
cuenta_paypal.depositar(1000)
cuenta_paypal.sacar(300)
```
