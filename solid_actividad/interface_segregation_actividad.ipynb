# 4. Interface Segregation

Aplicamos el principio de Interface Segregation a la idea de un dispositivo que cumple con ciertas tareas o funciones, como lo es un telefono y una impresora

1.1 Ejemplo malo

```python
from abc import ABC, abstractmethod


class Dispositivo(ABC):

    @abstractmethod
    def encender_dispositivo(self) -> None:
        pass

    @abstractmethod
    def imprimir_desde_dispositivo(self) -> None:
        pass

    @abstractmethod
    def escanear_desde_dispositivo(self) -> None:
        pass

    @abstractmethod
    def llamar_desde_dispositivo(self) -> None:
        pass


class Telefono(Dispositivo):

    def encender_dispositivo(self) -> None:
        print('Encendiendo el telefono...')

    def llamar_desde_dispositivo(self) -> None:
        print('Llamando desde el telefono...')

    def imprimir_desde_dispositivo(self) -> None:
        pass

    def escanear_desde_dispositivo(self) -> None:
        pass


class Impresora(Dispositivo):

    def encender_dispositivo(self) -> None:
        print('Encendiendo la impresora...')

    def imprimir_desde_dispositivo(self) -> None:
        print('Imprimiendo desde la impresora...')

    def escanear_desde_dispositivo(self) -> None:
        print('Escaneando desde la impresora...')

    def llamar_desde_dispositivo(self) -> None:
        pass
```

```python
telefono = Telefono()
impresora = Impresora()


telefono.encender_dispositivo()
telefono.llamar_desde_dispositivo()


impresora.encender_dispositivo()
impresora.imprimir_desde_dispositivo()
impresora.escanear_desde_dispositivo()
```

1.2 Ejemplo Bueno

```python
from abc import ABC, abstractmethod

class Encender(ABC):
    @abstractmethod
    def encender_dispositivo(self):
        pass

class Imprimir(ABC):
    @abstractmethod
    def imprimir_desde_dispositivo(self):
        pass

class Escanear(ABC):
    @abstractmethod
    def escanear_desde_dispositivo(self):
        pass

class Llamar(ABC):
    @abstractmethod
    def llamar_desde_dispositivo(self):
        pass

class Telefono(Encender, Llamar):
    def encender_dispositivo(self) -> None:
        print('Encendiendo el telefono...')

    def llamar_desde_dispositivo(self) -> None:
        print('Llamando desde el telefono...')

class Impresora(Encender, Imprimir, Escanear):
    def encender_dispositivo(self) -> None:
        print('Encendiendo impresora...')

    def imprimir_desde_dispositivo(self) -> None:
        print('Imprimiendo desde la impresora...')

    def escanear_desde_dispositivo(self) -> None:
        print('Escaneando desde la impresora...')
```

```python
telefono:Telefono = Telefono()
impresora:Impresora = Impresora()

telefono.encender_dispositivo()
telefono.llamar_desde_dispositivo()

impresora.encender_dispositivo()
impresora.imprimir_desde_dispositivo()
impresora.escanear_desde_dispositivo()
```
